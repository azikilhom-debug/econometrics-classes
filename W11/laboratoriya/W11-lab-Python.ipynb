```python
# --- Kerakli kutubxonalar ---------------------------------------------
import math
import numpy as np                  # sonli massivlar
import pandas as pd                 # jadval koʻrinishidagi maʼlumot
import matplotlib
import matplotlib.pyplot as plt     # grafiklar
from scipy import stats             # qt / pt / pchisq ekvivalentlari

# Okabe–Ito palitrasi: rang koʻrmaslikka moslashgan.
oiBlue = "#0072B2"; oiOrange = "#E69F00"; oiGreen = "#009E73"
oiPink = "#CC79A7"; oiVerm   = "#D55E00"; oiSky   = "#56B4E9"
oiYell = "#F0E442"; oiGrey   = "#5A6470"

plt.rcParams.update({
    "figure.dpi": 110, "figure.figsize": (7, 4.2), "font.size": 10,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.alpha": 0.25, "axes.titleweight": "bold"})

# QOIDA (2-haftadan): grafik yorliqlarida faqat ASCII harflar.
print("MUHIT: matplotlib", matplotlib.__version__)
```

```python
# --- IKKI DAFTAR BIR XIL MATN CHIQARISHI UCHUN YORDAMCHILAR ------------
# (1) ha(): R «TRUE», Python «True» yozadi. Mantiqiy qiymatni
#     OʻZIMIZ formatlaymiz.
def ha(b):
    return "HA" if bool(b) else "YOʻQ"

# (2) pad()/padl(): R ning sprintf("%-16s") formati satrni BAYT boʻyicha
#     toʻldiradi, Python esa BELGI boʻyicha. «YOʻQ» da 4 ta belgi, ammo
#     5 ta bayt bor.
def pad(s, n):
    return s + " " * max(0, n - len(s))

def padl(s, n):
    return " " * max(0, n - len(s)) + s

# (3) P_(): R dagi sprintf + cat ekvivalenti.
def P_(fmt, *a):
    print(fmt % a)

# (4) Dispersiya: MAXRAJNI HAR DOIM OSHKORA YOZAMIZ.
def varS(x):                        # maxraj n-1
    x = np.asarray(x, dtype=float)
    return float(((x - x.mean()) ** 2).sum() / (len(x) - 1))

def varP(x):                        # maxraj n
    x = np.asarray(x, dtype=float)
    return float(((x - x.mean()) ** 2).sum() / len(x))

def sdS(x):
    return math.sqrt(varS(x))

def sdP(x):
    return math.sqrt(varP(x))

# (5) p-qiymat funksiyalari.
def p_t2(t, df):                    # ikki tomonlama t
    return float(2 * stats.t.sf(abs(t), df))

def p_z2(z):                        # ikki tomonlama z
    return float(2 * stats.norm.cdf(-abs(z)))

def p_chi(x, df):                   # HAR DOIM oʻng dum
    return float(stats.chi2.sf(x, df))

# (6) Kritik qiymatlar.
Z975 = 1.959964                                    # z_{0,975} — LITERAL

def tkr(daraja, df):
    return float(stats.t.ppf(daraja, df))

def chikr(daraja, df):
    return float(stats.chi2.ppf(daraja, df))

# (7) Welch: bitta funksiya, uchta qiymat (SE, t, df).
def welch(x, y):
    x = np.asarray(x, dtype=float); y = np.asarray(y, dtype=float)
    n1 = len(x); n2 = len(y)
    v1 = varS(x); v2 = varS(y)
    se = math.sqrt(v1 / n1 + v2 / n2)
    df = (v1 / n1 + v2 / n2) ** 2 / \
         ((v1 / n1) ** 2 / (n1 - 1) + (v2 / n2) ** 2 / (n2 - 1))
    return {"se": se, "t": (x.mean() - y.mean()) / se, "df": df}

# (8) Korrelyatsiya — formulani oshkora yozamiz (ikki tilda bir xil bit).
def korr(x, y):
    x = np.asarray(x, dtype=float); y = np.asarray(y, dtype=float)
    dx = x - x.mean(); dy = y - y.mean()
    return float((dx * dy).sum() / math.sqrt((dx ** 2).sum() * (dy ** 2).sum()))

# (9) Birinchi raqam: "%.0f" orqali — int() katta sonlarda ham ishlaydi,
#     lekin ikki tilda AYNAN bir xil boʻlishi uchun formatdan foydalanamiz.
def birinchiRaqam(x):
    return np.array([int(("%.0f" % v)[0]) for v in np.asarray(x, dtype=float)])

# (10) Kontingensiya jadvali uchun kutilgan chastotalar (F148).
def kutilgan(M):
    M = np.asarray(M, dtype=float)
    return np.outer(M.sum(axis=1), M.sum(axis=0)) / M.sum()
```

```python
# --- MINSTD generatori (2-haftadan beri oʻzgarmagan) -------------------
#     x <- (48271 * x) mod 2147483647,    u <- x / 2147483647
A_ = 48271; M_ = 2147483647

def minstd(urug):
    holat = {"x": urug % M_}
    def gen(n):
        n = int(n); o = np.empty(n); x = holat["x"]
        for i in range(n):
            x = (A_ * x) % M_
            o[i] = x
        holat["x"] = x
        return o / M_
    return gen

# Indeks tanlash: 0 dan hi-1 gacha butun son (R bilan bir xil).
def indeks(gen, n, hi):
    return (gen(n) * hi).astype(int)

# Fisher–Yates aralashtirish. DIQQAT — tartib QATʼIY QAYD ETILGAN:
# u oqimidan n-1 ta son olinadi, teskari yoʻnalishda ishlatiladi.
def aralash(gen, n):
    n = int(n)
    a = np.arange(n)
    u = gen(n - 1)
    for i in range(n - 1, 0, -1):
        j = int(u[n - 1 - i] * (i + 1))
        a[i], a[j] = a[j], a[i]
    return a                          # 0 dan boshlangan indekslar

g = minstd(2026); u3 = g(3)
P_("=== 0-BOSQICH: MINSTD NAZORATI ===")
P_("urugʻ 2026 -> u1 = %.10f  u2 = %.10f  u3 = %.10f", u3[0], u3[1], u3[2])
g = minstd(11)
P_("aralashtirish nazorati (urugʻ 11, n = 6): %s",
   " ".join(str(v) for v in aralash(g, 6)))
P_("bu ikki qator ikkinchi tildagi daftarda ham AYNAN shunday chiqishi shart")
rng = np.random.default_rng(2026)
print("MUHIT: Python ning default_rng(2026).random(1) = %.10f  "
      "(R da boshqa son chiqadi)" % rng.random())
```

# 1-topshiriq. A/B test: dizayn savoli va randomizatsiya {#t1}

> **Blum:** Qoʻllash · **Qiyinlik:** 3 (Amaliy) · **Ball:** 16 ·
> **Vaqt:** 16 daqiqa

**Dizayn savoli.** Horror janridagi film byudjeti Comedy janridagidan
farq qiladimi? Har bir Horror filmi *aniq bitta* Comedy filmi bilan
bogʻlanganmi? **Yoʻq.** Demak bu — **bogʻliqsiz** ikki tanlanma;
klassik A/B test tuzilishi.

**Statistik tushuncha.** $H_0$ aytadiki, guruh yorligʻi hech narsani
anglatmaydi. Agar bu rost boʻlsa, $30$ ta byudjetni bitta qutiga
solib, yorliqlarni **qayta tashlash** mumkin. Har tashlashda
oʻrtachalar farqini yozamiz — bu **nol taqsimoti**. $p$-qiymat
(F153) shundan **sanaladi**:

$$p_{\text{rand}}=\frac{\#\{|d^{*}|\ge|d_{\text{kuz}}|\}}{M}.$$

```python
# --- 1A: tanlanma (urugʻ 20261101, har janrdan 15 ta film) -------------
hor = kino.loc[kino["janr"] == "Horror", "byudjet"].values / 1e6   # mln $
com = kino.loc[kino["janr"] == "Comedy", "byudjet"].values / 1e6
g = minstd(20261101)
nA = 15
xa = hor[indeks(g, nA, len(hor))]
xb = com[indeks(g, nA, len(com))]
dobs = xa.mean() - xb.mean()

P_("=== 1-TOPSHIRIQ: A/B TEST --- DIZAYN SAVOLI VA RANDOMIZATSIYA ===")
P_("--- 1A: tanlanma (urugʻ 20261101, har janrdan 15 ta film, mln $) ---")
P_("Horror : %s", ", ".join("%g" % v for v in xa))
P_("Comedy : %s", ", ".join("%g" % v for v in xb))
P_("xbar_Horror = %.6f   s = %.6f   n = %d", xa.mean(), sdS(xa), nA)
P_("xbar_Comedy = %.6f   s = %.6f   n = %d", xb.mean(), sdS(xb), nA)
P_("kuzatilgan farq d_kuz = %.6f mln $", dobs)

# --- 1B: yorliqni aralashtirish ---------------------------------------
qutiq = np.concatenate([xa, xb])
M = 20000
gp = minstd(20261102)
farqlar = np.empty(M)
for i in range(M):
    sh = qutiq[aralash(gp, 2 * nA)]
    farqlar[i] = sh[:nA].mean() - sh[nA:].mean()
keskin = int((np.abs(farqlar) >= abs(dobs) - 1e-12).sum())
p_rand = keskin / M
P_("--- 1B: yorliqni aralashtirish (%d marta, urugʻ 20261102) ---", M)
P_("nol taqsimoti: markaz = %.6f (nazariy 0)   SD = %.6f",
   farqlar.mean(), sdS(farqlar))
P_("|farq| >= %.6f boʻlgan holatlar: %d / %d", abs(dobs), keskin, M)
P_("p_randomizatsiya = %.6f   ->  %s", p_rand,
   "H0 RAD ETILADI" if p_rand < 0.05 else "H0 RAD ETILMAYDI")

# --- 1C: Welch t-testi bilan solishtirish ------------------------------
w1 = welch(xa, xb)
pw1 = p_t2(w1["t"], w1["df"])
tc1 = tkr(0.975, w1["df"])
lo1 = dobs - tc1 * w1["se"]; hi1 = dobs + tc1 * w1["se"]
P_("--- 1C: Welch t-testi (F129-F131) ---")
P_("SE = %.6f   t = %.6f   df = %.6f   p = %.6f   ->  %s",
   w1["se"], w1["t"], w1["df"], pw1,
   "H0 RAD ETILADI" if pw1 < 0.05 else "H0 RAD ETILMAYDI")
P_("ikki usul farqi = %.6f (%.2f foiz nisbiy)",
   abs(p_rand - pw1), abs(p_rand - pw1) / pw1 * 100)
P_("95 foiz IO (F132): [%.6f; %.6f]  nol %s", lo1, hi1,
   "ICHIDA" if (lo1 <= 0 <= hi1) else "TASHQARIDA")

# --- 1D: xuddi shu savol, BARCHA maʼlumot bilan ------------------------
w2 = welch(hor, com)
d2 = hor.mean() - com.mean()
pw2 = p_t2(w2["t"], w2["df"])
tc2 = tkr(0.975, w2["df"])
sp2 = math.sqrt(((len(hor) - 1) * varS(hor) + (len(com) - 1) * varS(com)) /
                (len(hor) + len(com) - 2))
P_("--- 1D: xuddi shu savol, BARCHA maʼlumot bilan (tanlanma emas) ---")
P_("Horror: n = %d  xbar = %.6f  s = %.6f", len(hor), hor.mean(), sdS(hor))
P_("Comedy: n = %d  xbar = %.6f  s = %.6f", len(com), com.mean(), sdS(com))
P_("farq = %.6f mln $   SE = %.6f   df = %.6f", d2, w2["se"], w2["df"])
P_("t = %.6f   p = %.6e   ->  %s", w2["t"], pw2,
   "H0 RAD ETILADI" if pw2 < 0.05 else "H0 RAD ETILMAYDI")
P_("95 foiz IO: [%.6f; %.6f]  nol %s",
   d2 - tc2 * w2["se"], d2 + tc2 * w2["se"],
   "ICHIDA" if ((d2 - tc2 * w2["se"]) <= 0 <= (d2 + tc2 * w2["se"]))
   else "TASHQARIDA")
P_("Koen d (F135) = %.6f", d2 / sp2)
P_("n = 15 da p = %.4f, toʻliq maʼlumotda p = %.3e --- EFFEKT OʻZGARMADI,",
   p_rand, pw2)
P_("oʻzgargani faqat uni koʻrish qobiliyati (10-haftaning quvvat darsi).")
```

```python
# --- Nol taqsimotining rasmi ------------------------------------------
mask = np.abs(farqlar) >= abs(dobs) - 1e-12
fig, ax = plt.subplots()
qir = np.linspace(farqlar.min(), farqlar.max(), 46)
ax.hist(farqlar[~mask], bins=qir, color=oiBlue, edgecolor="white",
        linewidth=0.15, label="markaz")
ax.hist(farqlar[mask], bins=qir, color=oiVerm, edgecolor="white",
        linewidth=0.15, label="as-or-more extreme")
for v in (dobs, -dobs):
    ax.axvline(v, color=oiVerm, linestyle="--", linewidth=0.9)
ax.set_title("Randomizatsiya nol taqsimoti (20 000 aralashtirish)")
ax.set_xlabel("aralashtirilgan guruhlar ortachalari farqi (mln $)")
ax.set_ylabel("chastota")
ax.legend(frameon=False)
fig.text(0.01, -0.02, "Vertikal chiziqlar: kuzatilgan farq -13.633 va uning aksi",
         fontsize=8, color="grey")
plt.show()
```

# 2-topshiriq. Ikki bogʻliqsiz oʻrtacha: Welch va pooled {#t2}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 14 ·
> **Vaqt:** 14 daqiqa

**Dizayn savoli.** Oʻzbekiston va Qozogʻistonning bugʻdoy
hosildorligi. $1994$-yildagi oʻzbek qiymati $1994$-yildagi qozoq
qiymati bilan **juftmi**? Ular bir xil yilga tegishli — lekin ular
*bir xil kuzatuv birligining* ikki oʻlchovi emas: bu ikki alohida
mamlakat. Biz ularni **bogʻliqsiz** deb qaraymiz — va 2C bosqichida
bu qarorning narxini koʻramiz.

**Ikki formula, bitta savol.**

$$\SE_{\text{Welch}}=\sqrt{\frac{s_1^2}{n_1}+\frac{s_2^2}{n_2}}\ \ (\text{F129}),
\qquad
s_p^2=\frac{(n_1-1)s_1^2+(n_2-1)s_2^2}{n_1+n_2-2}\ \ (\text{F133}).$$

```python
# --- Maʼlumot: bugʻdoy hosildorligi 1992-2018 (t/ga) -------------------
u = uzq["bugdoy"].values
k = kzq["bugdoy"].values
n1 = len(u); n2 = len(k)

P_("=== 2-TOPSHIRIQ: IKKI BOGʻLIQSIZ OʻRTACHA --- WELCH va POOLED ===")
P_("%s: n = %d  xbar = %.6f  s = %.6f  s^2 = %.6f",
   pad(UZ, 11), n1, u.mean(), sdS(u), varS(u))
P_("%s: n = %d  xbar = %.6f  s = %.6f  s^2 = %.6f",
   pad(KZ, 11), n2, k.mean(), sdS(k), varS(k))
P_("farq = %.6f t/ga", u.mean() - k.mean())
P_("dispersiyalar nisbati s1^2/s2^2 = %.6f (SD lar nisbati %.6f)",
   varS(u) / varS(k), sdS(u) / sdS(k))

# --- 2A: ikki hisoblash ------------------------------------------------
wA = welch(u, k)
pA = p_t2(wA["t"], wA["df"])
sp2v = ((n1 - 1) * varS(u) + (n2 - 1) * varS(k)) / (n1 + n2 - 2)
seP = math.sqrt(sp2v * (1 / n1 + 1 / n2))
tP = (u.mean() - k.mean()) / seP
dfP = n1 + n2 - 2
pP = p_t2(tP, dfP)
P_("--- 2A: ikki hisoblash ---")
P_("Welch  : SE = %.6f  t = %.6f  df = %.6f  p = %.6e",
   wA["se"], wA["t"], wA["df"], pA)
P_("Pooled : s_p^2 = %.6f  SE = %.6f  t = %.6f  df = %d  p = %.6e",
   sp2v, seP, tP, dfP, pP)
P_("t statistikalari AYNAN tengmi (|farq| < 1e-9)? %s   <- chunki n1 = n2",
   ha(abs(wA["t"] - tP) < 1e-9))
P_("df esa juda farq qiladi: %.4f va %d", wA["df"], dfP)
P_("pooled p ni %.1f marta kichraytiradi --- tekin emas, foydasi yoʻq",
   pA / pP)

# --- 2B: effekt va oraliq ----------------------------------------------
tcA = tkr(0.975, wA["df"])
loA = (u.mean() - k.mean()) - tcA * wA["se"]
hiA = (u.mean() - k.mean()) + tcA * wA["se"]
P_("--- 2B: effekt hajmi va ishonch oraligʻi ---")
P_("95 foiz IO (Welch): [%.6f; %.6f]  nol %s", loA, hiA,
   "ICHIDA" if (loA <= 0 <= hiA) else "TASHQARIDA")
P_("Koen d = %.6f  (juda katta)", (u.mean() - k.mean()) / math.sqrt(sp2v))
P_("Qaror (alpha = 0,05): %s",
   "H0 RAD ETILADI" if pA < 0.05 else "H0 RAD ETILMAYDI")

# --- 2C: SHARTLARNI TEKSHIRISH -----------------------------------------
# Bu ikki qator — VAQT QATORI. Ketma-ket yillar bogʻliqsizmi?
# Lag-1 avtokorrelyatsiya chegarasi: z_{0,975}/sqrt(n) = 1,959964/sqrt(n).
r1 = korr(u[:-1], u[1:])
r2 = korr(k[:-1], k[1:])
lim = Z975 / math.sqrt(n1)
P_("--- 2C: SHARTLARNI TEKSHIRISH (bu ikki qator VAQT QATORI) ---")
P_("lag-1 avtokorrelyatsiya: %s %.6f | %s %.6f",
   pad(UZ, 11), r1, pad(KZ, 11), r2)
P_("chegara z_{0,975}/sqrt(n) = %.6f/sqrt(%d) = %.6f", Z975, n1, lim)
P_("  %s: |r| = %.6f -> %s", pad(UZ, 11), abs(r1),
   "BOGʻLIQSIZLIK SHARTI BUZILGAN" if abs(r1) > lim
   else "sezilarli bogʻliqlik yoʻq")
P_("  %s: |r| = %.6f -> %s", pad(KZ, 11), abs(r2),
   "BOGʻLIQSIZLIK SHARTI BUZILGAN" if abs(r2) > lim
   else "sezilarli bogʻliqlik yoʻq")
P_("Buzilish faqat BITTA qatorda. Bitta qatorning buzilishi ham yetarli:")
P_("p-qiymat rasmiy maʼnoda haqiqiy emas. Hisob toʻgʻri, shart buzilgan.")
```

```python
# --- Ikki qatorning rasmi: nega biri avtokorrelyatsiyalangan? ----------
fig, ax = plt.subplots()
ax.plot(uzq["yil"].values, u, color=oiBlue, marker="o", markersize=4,
        linewidth=1.0, label="1-pozitsiya")
ax.plot(kzq["yil"].values, k, color=oiOrange, marker="^", markersize=4,
        linewidth=1.0, label="3-pozitsiya")
ax.set_title("Bugdoy hosildorligi 1992-2018 (t/ha)")
ax.set_xlabel("yil"); ax.set_ylabel("t/ha")
ax.legend(frameon=False)
fig.text(0.01, -0.02,
         "1-pozitsiya: kuchli trend (lag-1 r = 0.966). "
         "3-pozitsiya: trendsiz (r = -0.000)", fontsize=8, color="grey")
plt.show()
```

# 3-topshiriq. XATO OVI: juft maʼlumot bogʻliqsiz deb tahlil qilingan {#t3}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 18 ·
> **Vaqt:** 18 daqiqa

## Sizga berilgan hisobot

> **«Qozogʻistonda bugʻdoy va arpa hosildorligini solishtirdik
> ($1992$–$2018$). Ikki tanlanmali $t$-test oʻtkazdik:
> $t=-2{,}00$, df $=50{,}61$, $p=0{,}0509$. $95\,\%$ ishonch oraligʻi
> $[-0{,}3104;\ 0{,}0006]$ nolni oʻz ichiga oladi. **Xulosa: bugʻdoy
> va arpa hosildorligi orasida statistik jihatdan ahamiyatli farq
> topilmadi.**»**

**Sizning vazifangiz:** bu hisobotdagi xatoni toping, uni
**oʻlchang**, va toʻgʻri tahlilni oʻtkazing.

> **Yordam berilmaydigan maslahat.** Xato formulada emas — hisobotda
> keltirilgan barcha sonlar *oʻsha* formuladan **toʻgʻri** chiqqan.
> Xato bir bosqich yuqorida: **dizayn savoliga** notoʻgʻri javob
> berilgan.

```python
# --- 3A: hamkasbning hisobotini QAYTA HOSIL QILAMIZ (xato yoʻl) --------
bug = kzq["bugdoy"].values
arp = kzq["arpa"].values
nC = len(bug)

P_("=== 3-TOPSHIRIQ: XATO OVI --- JUFT MAʼLUMOT BOGʻLIQSIZ DEB TAHLIL QILINGAN ===")
P_("--- 3A: hamkasbning hisoboti (XATO YOʻL: bogʻliqsiz deb hisoblash) ---")
P_("bugʻdoy: n = %d  xbar = %.6f  s = %.6f", nC, bug.mean(), sdS(bug))
P_("arpa   : n = %d  xbar = %.6f  s = %.6f", nC, arp.mean(), sdS(arp))
wX = welch(bug, arp)
pX = p_t2(wX["t"], wX["df"])
tcX = tkr(0.975, wX["df"])
dX = bug.mean() - arp.mean()
loX = dX - tcX * wX["se"]; hiX = dX + tcX * wX["se"]
P_("SE = %.6f   t = %.6f   df = %.6f   p = %.6f   ->  %s",
   wX["se"], wX["t"], wX["df"], pX,
   "H0 RAD ETILADI" if pX < 0.05 else "H0 RAD ETILMAYDI")
P_("95 foiz IO: [%.6f; %.6f]  nol %s", loX, hiX,
   "ICHIDA" if (loX <= 0 <= hiX) else "TASHQARIDA")
P_("-> hisobotdagi sonlar QAYTA HOSIL BOʻLDI. Demak arifmetika toʻgʻri.")

# --- 3B: xatoni topamiz --- juftlarni tekshiramiz ----------------------
r_juft = korr(bug, arp)
d = bug - arp
P_("--- 3B: XATO QAYERDA? kuzatuv birligini tekshiramiz ---")
P_("Har bir YIL uchun ikkala son ham bor: bir xil tuproq, ob-havo, siyosat.")
P_("Demak kuzatuv birligi --- YIL, son emas. n = %d juft, n = %d emas.",
   nC, 2 * nC)
P_("juftlar korrelyatsiyasi r = %.6f", r_juft)
P_("%d ta farqdan %d tasi manfiy --- tasodifiy boʻlishi mumkinmi?",
   nC, int((d < 0).sum()))

# --- 3C: TOʻGʻRI tahlil --- juftlashgan t-test (F136-F137) -------------
seD = sdS(d) / math.sqrt(nC)
tD = d.mean() / seD
pD = p_t2(tD, nC - 1)
tcD = tkr(0.975, nC - 1)
P_("--- 3C: TOʻGʻRI tahlil --- juftlashgan t-test ---")
P_("D_i = X_i - Y_i ustida BIR TANLANMALI test:")
P_("dbar = %.6f   s_d = %.6f   SE = %.6f", d.mean(), sdS(d), seD)
P_("t = %.6f   df = %d   p = %.6e   ->  %s", tD, nC - 1, pD,
   "H0 RAD ETILADI" if pD < 0.05 else "H0 RAD ETILMAYDI")
P_("95 foiz IO: [%.6f; %.6f]  nol %s",
   d.mean() - tcD * seD, d.mean() + tcD * seD,
   "ICHIDA" if ((d.mean() - tcD * seD) <= 0 <= (d.mean() + tcD * seD))
   else "TASHQARIDA")
P_("effekt hajmi d_z = dbar/s_d = %.6f (F139)", d.mean() / sdS(d))

# --- 3D: NEGA SE shunchalik farq qildi? (F138 ayniyati) ----------------
vD = varS(bug) + varS(arp) - 2 * r_juft * sdS(bug) * sdS(arp)
P_("--- 3D: NEGA? --- Var(D) = s1^2 + s2^2 - 2 r s1 s2  (F138) ---")
P_("%.6f + %.6f - 2*%.6f*%.6f*%.6f = %.6f",
   varS(bug), varS(arp), r_juft, sdS(bug), sdS(arp), vD)
P_("kuzatilgan s_d^2 = %.6f", varS(d))
P_("bu AYNIYATmi (|ayirma| < 1e-12)? %s", ha(abs(vD - varS(d)) < 1e-12))
P_("r > 0 boʻlgani uchun dispersiya %.4f marta kichrayadi,",
   (varS(bug) + varS(arp)) / varS(d))
P_("standart xato esa sqrt(%.4f) = %.4f marta.",
   (varS(bug) + varS(arp)) / varS(d),
   math.sqrt((varS(bug) + varS(arp)) / varS(d)))

# --- 3E: ikki xulosa yonma-yon -----------------------------------------
P_("--- 3E: BIR XIL MAʼLUMOT, IKKI XIL XULOSA ---")
P_("%s%s%s", pad("", 18), padl("bogʻliqsiz (XATO)", 20),
   padl("juftlashgan (TOʻGʻRI)", 24))
P_("%s%s%s", pad("farq bahosi", 18), padl("%.6f" % dX, 20),
   padl("%.6f" % d.mean(), 24))
P_("%s%s%s", pad("standart xato", 18), padl("%.6f" % wX["se"], 20),
   padl("%.6f" % seD, 24))
P_("%s%s%s", pad("erkinlik darajasi", 18), padl("%.4f" % wX["df"], 20),
   padl("%d" % (nC - 1), 24))
P_("%s%s%s", pad("p-qiymat", 18), padl("%.6f" % pX, 20),
   padl("%.6e" % pD, 24))
P_("%s%s%s", pad("qaror", 18),
   padl("RAD ETILADI" if pX < 0.05 else "RAD ETILMAYDI", 20),
   padl("RAD ETILADI" if pD < 0.05 else "RAD ETILMAYDI", 24))
P_("SE nisbati: %.4f marta. Farqning bahosi esa AYNAN bir xil.",
   wX["se"] / seD)
```

```python
# --- Juftlik koʻrinadigan rasm ----------------------------------------
fig, ax = plt.subplots()
lo = min(bug.min(), arp.min()); hi = max(bug.max(), arp.max())
ax.plot([lo, hi], [lo, hi], color=oiGrey, linestyle="--", linewidth=0.9)
ax.scatter(bug, arp, color=oiBlue, s=26)
ax.set_title("Har bir nuqta -- bitta YIL (juft)")
ax.set_xlabel("bugdoy (t/ha)"); ax.set_ylabel("arpa (t/ha)")
fig.text(0.01, -0.02,
         "Juftlar korrelyatsiyasi r = 0.923. "
         "Deyarli hamma nuqta chiziq ostida.", fontsize=8, color="grey")
plt.show()
```

# 4-topshiriq. Ikki ulush A/B testi va $\chi^2=z^2$ {#t4}

> **Blum:** Qoʻllash · **Qiyinlik:** 3 (Amaliy) · **Ball:** 14 ·
> **Vaqt:** 14 daqiqa

**Dizayn savoli.** Natija **son** emas, **toifa**: film «foydali»
boʻldimi yoki yoʻq? (Sohada qabul qilingan qoida: jahon daromadi
byudjetning $2{,}5$ barobaridan koʻp.) Ikki guruh — PG-13 va R —
bogʻliqsiz. Demak: **ikki ulush $z$-testi**.

**Ikki xil standart xato — haftaning eng koʻp uchraydigan texnik
xatosi.**

$$\SE_0=\sqrt{\phat(1-\phat)\Bigl(\tfrac1{n_1}+\tfrac1{n_2}\Bigr)}\ \ (\text{F142})
\quad\text{TEST uchun};\qquad
\SE=\sqrt{\tfrac{\phat_1(1-\phat_1)}{n_1}+\tfrac{\phat_2(1-\phat_2)}{n_2}}\ \ (\text{F144})
\quad\text{ORALIQ uchun}.$$

```python
# --- 4A: maʼlumot ------------------------------------------------------
foyda = (kinoR["jahon_daromadi"] > 2.5 * kinoR["byudjet"]).values
rey = kinoR["reyting"].values
P_("=== 4-TOPSHIRIQ: IKKI ULUSH A/B TESTI ===")
P_("--- 4A: 'foydali' film = jahon daromadi > 2,5 x byudjet ---")
for rr in ["G", "PG", "PG-13", "R"]:
    m_ = (rey == rr)
    P_("  %s: %d / %d = %.6f", pad(rr, 6), int(foyda[m_].sum()),
       int(m_.sum()), float(foyda[m_].mean()))
x1 = int(foyda[rey == "PG-13"].sum()); nn1 = int((rey == "PG-13").sum())
x2 = int(foyda[rey == "R"].sum());     nn2 = int((rey == "R").sum())
p1 = x1 / nn1; p2 = x2 / nn2
P_("A varianti (PG-13): %d/%d = %.6f", x1, nn1, p1)
P_("B varianti (R)    : %d/%d = %.6f", x2, nn2, p2)
P_("farq = %.6f (F140)", p1 - p2)

# --- 4B: shartlar ------------------------------------------------------
pbir = (x1 + x2) / (nn1 + nn2)                       # F141
P_("--- 4B: muvaffaqiyat-muvaffaqiyatsizlik sharti ---")
for nm, nv in [("PG-13", nn1), ("R", nn2)]:
    P_("  %s: n*phat = %.4f, n*(1-phat) = %.4f -> %s", pad(nm, 6),
       nv * pbir, nv * (1 - pbir),
       "BAJARILDI" if min(nv * pbir, nv * (1 - pbir)) >= 10 else "BAJARILMADI")

# --- 4C: TEST (SE_0 birlashtirilgan ulushdan) --------------------------
se0 = math.sqrt(pbir * (1 - pbir) * (1 / nn1 + 1 / nn2))   # F142
zv = (p1 - p2) / se0                                       # F143
pz = p_z2(zv)
P_("--- 4C: TEST --- SE_0 BIRLASHTIRILGAN ulushdan ---")
P_("phat_bir = %.6f   SE_0 = %.6f", pbir, se0)
P_("z = %.6f   p = %.6f   ->  %s", zv, pz,
   "H0 RAD ETILADI" if pz < 0.05 else "H0 RAD ETILMAYDI")

# --- 4D: ISHONCH ORALIGʻI (SE birlashtirilmagan) -----------------------
seio = math.sqrt(p1 * (1 - p1) / nn1 + p2 * (1 - p2) / nn2)   # F144
loP = (p1 - p2) - Z975 * seio; hiP = (p1 - p2) + Z975 * seio
P_("--- 4D: ISHONCH ORALIGʻI --- SE BIRLASHTIRILMAGAN ---")
P_("SE = %.6f   (SE_0 dan %.6f farq)", seio, abs(seio - se0))
P_("95 foiz IO: [%.6f; %.6f]  nol %s", loP, hiP,
   "ICHIDA" if (loP <= 0 <= hiP) else "TASHQARIDA")
P_("Test va oraliq bir xil xulosaga keldimi? %s",
   ha((pz < 0.05) == (not (loP <= 0 <= hiP))))

# --- 4E: effekt hajmi (F145) -------------------------------------------
RR = p1 / p2
OR = (p1 / (1 - p1)) / (p2 / (1 - p2))
P_("--- 4E: EFFEKT HAJMI --- p ni hech qachon yolgʻiz qoldirmang ---")
P_("mutlaq farq = %.6f   nisbiy = %+.2f foiz", p1 - p2, (p1 - p2) / p2 * 100)
P_("nisbiy xavf RR = %.6f   shanslar nisbati OR = %.6f", RR, OR)
P_("Ulushlar uchun MUTLAQ farq eng tushunarli oʻlchov; OR eng chalgʻituvchisi.")

# --- 4F: 2x2 jadval uchun chi^2 = z^2 (F152) ---------------------------
M22 = np.array([[x1, nn1 - x1], [x2, nn2 - x2]], dtype=float)
E22 = kutilgan(M22)
chi22 = float(((M22 - E22) ** 2 / E22).sum())
P_("--- 4F: 2x2 jadval uchun chi^2 = z^2 (F152) ---")
P_("jadval:  foydali  foydasiz | jami")
P_("  PG-13  %7d  %8d | %d", M22[0, 0], M22[0, 1], M22[0].sum())
P_("  R      %7d  %8d | %d", M22[1, 0], M22[1, 1], M22[1].sum())
P_("chi^2 = %.9f   df = 1   p = %.9f", chi22, p_chi(chi22, 1))
P_("z^2   = %.9f            p_z = %.9f", zv ** 2, pz)
P_("|chi^2 - z^2| < 1e-9 ? %s  -> bu AYNIYAT, tasodif emas",
   ha(abs(chi22 - zv ** 2) < 1e-9))
P_("n = %d boʻlsa ham natija ahamiyatsiz --- 10-haftaning teskari tomoni:",
   nn1 + nn2)
P_("katta n har qanday farqni 'ahamiyatli' qilmaydi, faqat kichigini koʻradi.")
```

**Talqin.** Yozma javob bering:

1. Nima uchun **test** birlashtirilgan ulushdan, **oraliq** esa
   birlashtirilmagandan foydalanadi? Javobda «$H_0$ ni rost deb
   faraz qilamiz» iborasi boʻlsin.
2. Bu yerda ikki $\SE$ atigi $0{,}000026$ farq qildi. Qachon ular
   sezilarli farq qiladi?
3. $n=2\,009$ — katta tanlanma, lekin natija ahamiyatsiz.
   10-haftada teskarisini koʻrgan edingiz. Ikkalasini bitta jumlada
   birlashtiring.
4. $OR=1{,}1454$. Talaba buni «$14{,}5\,\%$ koʻp» deb oʻqidi.
   Xatosini tuzating.

**Qaror.** Marketing boʻlimi «R reytingli filmlar kamroq foyda
keltiradi» degan xulosani eʼlon qilmoqchi. Ularga
**effekt + oraliq + $p$** tartibida bitta jumla yozing.

# 5-topshiriq. $\chi^2$ moslik testi: Benford qonuni {#t5}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 16 ·
> **Vaqt:** 16 daqiqa

**Dizayn savoli.** Bitta kategorik oʻzgaruvchi (birinchi raqam,
$9$ toifa) va **nazariyadan** kelgan kutilgan ulushlar. Demak:
**moslik testi** (F147), df $=k-1=8$.

$$\Prob(D=d)=\log_{10}\Bigl(1+\frac1d\Bigr),\qquad
E_d=n\,\Prob(D=d),\qquad
\chi^2=\sum_{d=1}^{9}\frac{(O_d-E_d)^2}{E_d}\ \ (\text{F146}).$$

```python
# --- 5A: bitta test, uchta ustun ---------------------------------------
ben = np.array([math.log10(1 + 1 / d_) for d_ in range(1, 10)])
P_("=== 5-TOPSHIRIQ: XI-KVADRAT MOSLIK TESTI --- BENFORD QONUNI ===")
P_("kutilgan ulushlar: %s",
   " ".join("%d:%.6f" % (d_, b_) for d_, b_ in zip(range(1, 10), ben)))
P_("kritik qiymat chi^2_{0,95}(8) = %.6f", chikr(0.95, 8))
P_("--- 5A: BIR XIL fayl, BIR XIL test, UCHTA ustun ---")
P_("%s%s%s%s%s%s", pad("ustun", 16), padl("n", 7), padl("chi^2", 12),
   padl("p", 11), padl("eng kichik E", 14), padl("qaror", 18))
ustunlar = ["byudjet", "jahon_daromadi", "ichki_daromad"]
natija5 = {}
for nm in ustunlar:
    v = kino[nm].values; v = v[v > 0]
    fd = birinchiRaqam(v)
    O = np.array([int((fd == d_).sum()) for d_ in range(1, 10)], dtype=float)
    nn = int(O.sum()); E = ben * nn
    ch = float(((O - E) ** 2 / E).sum()); pv = p_chi(ch, 8)
    natija5[nm] = {"O": O, "E": E, "ch": ch, "pv": pv, "n": nn}
    P_("%s%s%s%s%s%s", pad(nm, 16), padl("%d" % nn, 7),
       padl("%.6f" % ch, 12), padl("%.6f" % pv, 11),
       padl("%.4f" % E.min(), 14),
       padl("H0 RAD ETILADI" if pv < 0.05 else "H0 RAD ETILMAYDI", 18))
P_("Eng kichik kutilgan chastota hamma joyda >= 5 -> shart BAJARILDI.")

# --- 5B: byudjet ustunining toʻliq jadvali -----------------------------
b5 = natija5["byudjet"]
P_("--- 5B: byudjet ustunining toʻliq hisobi (n = %d) ---", b5["n"])
P_("raqam :  %s", "".join("%7d" % d_ for d_ in range(1, 10)))
P_("kuzat.:  %s", "".join("%7d" % int(o) for o in b5["O"]))
P_("kutil.:  %s", "".join("%7.2f" % e for e in b5["E"]))
hissa = (b5["O"] - b5["E"]) ** 2 / b5["E"]
P_("hissa :  %s", "".join("%7.3f" % h for h in hissa))
P_("chi^2 = %.6f   df = 8   p = %.6f", b5["ch"], b5["pv"])
P_("eng katta hissa: %d-raqam (%.3f)", int(np.argmax(hissa)) + 1, hissa.max())

# --- 5C: NEGA byudjet moslashmaydi? Sabab MAʼLUMOTDA koʻrinadi ---------
bv = kino["byudjet"].values
jv = kino["jahon_daromadi"].values
P_("--- 5C: NEGA? --- sabab statistikada emas, maʼlumotning kelib chiqishida ---")
P_("byudjet 1 mln ning butun karrasi : %.2f foiz",
   float((bv % 1e6 == 0).mean()) * 100)
P_("byudjet 5 mln ning butun karrasi : %.2f foiz",
   float((bv % 5e6 == 0).mean()) * 100)
P_("turli byudjet qiymatlari soni    : %d (%d film uchun)",
   len(set(bv)), len(bv))
P_("jahon daromadi 1 mln ning karrasi: %.2f foiz",
   float((jv % 1e6 == 0).mean()) * 100)
P_("turli jahon daromadi qiymatlari  : %d", len(set(jv)))
P_("Byudjet --- eʼlon qilingan, muzokara qilingan va YAXLITLANGAN son.")
P_("Jahon daromadi --- OʻLCHANGAN son. Bir xil fayl, bir xil test,")
P_("turli natija --- va sabab maʼlumot QANDAY paydo boʻlganida.")
```

# 6-topshiriq. $\chi^2$ bogʻliqsizlik testi: toʻliq oqim {#t6}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 22 ·
> **Vaqt:** 22 daqiqa

**Dizayn savoli.** **Ikki** kategorik oʻzgaruvchi (reyting va janr),
bitta tanlanma. Demak: **bogʻliqsizlik testi** (F148),
$E_{ij}=(\text{satr jami})(\text{ustun jami})/n$,
df $=(r-1)(c-1)$ (F149).

**Toʻliq oqim — oltita bosqich, hech biri tashlab ketilmaydi:**
jadval → kutilganlar → **shart** → $\chi^2$ va $p$ → **effekt hajmi
(Kramer $V$)** → **qayerda? (standartlashgan qoldiqlar)**.

```python
# --- 6A: kuzatilgan jadval ---------------------------------------------
reylev = ["G", "PG", "PG-13", "R"]
janlev = sorted(set(kinoR["janr"]))
rv = kinoR["reyting"].values; jv2 = kinoR["janr"].values
M = np.zeros((4, 5), dtype=int)
for i, rr in enumerate(reylev):
    for j, jj in enumerate(janlev):
        M[i, j] = int(((rv == rr) & (jv2 == jj)).sum())
P_("=== 6-TOPSHIRIQ: XI-KVADRAT BOGʻLIQSIZLIK TESTI ===")
P_("--- 6A: kuzatilgan jadval (reyting x janr, %d film) ---", int(M.sum()))
P_("%s%s%s", pad("", 8), "".join("%11s" % s for s in janlev),
   "%8s" % "jami")
for i, rr in enumerate(reylev):
    P_("%s%s%s", pad(rr, 8), "".join("%11d" % v for v in M[i]),
       "%8d" % M[i].sum())
P_("%s%s%s", pad("jami", 8), "".join("%11d" % v for v in M.sum(axis=0)),
   "%8d" % M.sum())

# --- 6B: kutilgan chastotalar (F148) -----------------------------------
E = kutilgan(M)
P_("--- 6B: kutilgan chastotalar E_ij = (satr jami)(ustun jami)/n ---")
P_("%s%s", pad("", 8), "".join("%11s" % s for s in janlev))
for i, rr in enumerate(reylev):
    P_("%s%s", pad(rr, 8), "".join("%11.4f" % v for v in E[i]))

# --- 6C: SHART --- ATAYLAB BUZILGAN ------------------------------------
chiM = float(((M - E) ** 2 / E).sum())
dfM = (M.shape[0] - 1) * (M.shape[1] - 1)
P_("--- 6C: SHART TEKSHIRUVI (bu bosqich hech qachon tashlanmaydi) ---")
P_("chi^2 = %.6f   df = %d   p = %.6e", chiM, dfM, p_chi(chiM, dfM))
P_("eng kichik kutilgan chastota = %.6f", E.min())
P_("5 dan kichik kataklar: %d / %d (%.2f foiz)",
   int((E < 5).sum()), E.size, int((E < 5).sum()) / E.size * 100)
P_("SHART %s --- p-qiymatga toʻliq ishonib boʻlmaydi.",
   "BUZILGAN" if E.min() < 5 else "BAJARILGAN")

# --- 6D: TUZATISH --- G va PG ni birlashtiramiz ------------------------
# Sabab MAZMUNIY: G va PG ikkalasi ham 'oilaviy' reyting, va G da
# atigi 49 film bor. Sabab p-qiymat EMAS.
M2 = np.vstack([M[0] + M[1], M[2], M[3]])
nomlar2 = ["G/PG", "PG-13", "R"]
E2 = kutilgan(M2)
chi2 = float(((M2 - E2) ** 2 / E2).sum())
df2 = (M2.shape[0] - 1) * (M2.shape[1] - 1)
P_("--- 6D: TUZATISH --- G va PG birlashtiriladi (G da atigi %d film) ---",
   int(M[0].sum()))
P_("%s%s%s", pad("", 8), "".join("%11s" % s for s in janlev),
   "%8s" % "jami")
for i in range(3):
    P_("%s%s%s", pad(nomlar2[i], 8), "".join("%11d" % v for v in M2[i]),
       "%8d" % M2[i].sum())
P_("chi^2 = %.6f   df = %d   p = %.6e", chi2, df2, p_chi(chi2, df2))
P_("eng kichik kutilgan chastota = %.6f -> shart %s", E2.min(),
   "BAJARILDI" if E2.min() >= 5 else "BUZILGAN")
P_("kritik qiymat chi^2_{0,95}(%d) = %.6f", df2, chikr(0.95, df2))
P_("Qaror: %s",
   "H0 RAD ETILADI" if p_chi(chi2, df2) < 0.05 else "H0 RAD ETILMAYDI")
P_("Xulosa OʻZGARMADI --- va bu halol natija, uyushtirilgan emas.")

# --- 6E: EFFEKT HAJMI --- Kramer V (F151) ------------------------------
V = math.sqrt(chi2 / (M2.sum() * (min(M2.shape) - 1)))
P_("--- 6E: EFFEKT HAJMI --- Kramer V ---")
P_("V = sqrt(%.4f / (%d * %d)) = %.6f", chi2, int(M2.sum()),
   min(M2.shape) - 1, V)
P_("Talqin: V ~ 0,1 zaif · 0,3 oʻrtacha · 0,5 kuchli -> bu yerda KUCHLI")
P_("Diqqat: chi^2 n bilan oʻsadi, V oʻsmaydi. 'Ahamiyatli' != 'kuchli'.")

# --- 6F: QAYERDA? --- standartlashgan qoldiqlar (F150) -----------------
Rr = (M2 - E2) / np.sqrt(E2)
P_("--- 6F: QAYERDA? --- standartlashgan qoldiqlar r_ij = (O-E)/sqrt(E) ---")
P_("%s%s", pad("", 8), "".join("%11s" % s for s in janlev))
for i in range(3):
    P_("%s%s", pad(nomlar2[i], 8), "".join("%11.4f" % v for v in Rr[i]))
fl = Rr.flatten()
ordr = np.argsort(-np.abs(fl), kind="stable")[:5]
P_("eng katta 5 ta qoldiq:")
for ix in ordr:
    i, j = int(ix) // Rr.shape[1], int(ix) % Rr.shape[1]
    P_("  %s x %s: kuzatilgan %4d, kutilgan %8.2f, qoldiq %+9.4f",
       pad(nomlar2[i], 6), pad(janlev[j], 10), int(M2[i, j]), E2[i, j],
       Rr[i, j])
P_("chi^2 faqat 'bogʻliqlik bor' deydi; QAYERDA ekanini qoldiqlar aytadi.")

# --- 6G: chi^2 taqsimoti qayerdan keladi? --- simulyatsiya -------------
gs = minstd(20261400)
Ms = 20000
nsim = 300
rowp = M2.sum(axis=1) / M2.sum()
colp = M2.sum(axis=0) / M2.sum()
cum = np.cumsum(np.outer(rowp, colp).flatten())
chis = np.empty(Ms)
for i in range(Ms):
    uu = gs(nsim)
    cells = np.searchsorted(cum, uu, side="right")
    tab = np.bincount(cells, minlength=len(cum)).reshape(3, 5).astype(float)
    ex = kutilgan(tab)
    with np.errstate(divide="ignore", invalid="ignore"):
        cc = np.where(ex > 0, (tab - ex) ** 2 / ex, 0.0)
    chis[i] = cc.sum()
P_("--- 6G: chi^2 taqsimoti qayerdan keladi? (H0 ROST, simulyatsiya) ---")
P_("n = %d kuzatuv, %d takror, urugʻ 20261400", nsim, Ms)
P_("%s%s%s", pad("", 22), padl("simulyatsiya", 16), padl("nazariya", 16))
P_("%s%s%s", pad("oʻrtacha", 22), padl("%.6f" % chis.mean(), 16),
   padl("df = %d" % df2, 16))
P_("%s%s%s", pad("dispersiya", 22), padl("%.6f" % varS(chis), 16),
   padl("2*df = %d" % (2 * df2), 16))
P_("%s%s%s", pad("95 foiz kvantil", 22),
   padl("%.6f" % float(np.quantile(chis, 0.95)), 16),
   padl("%.6f" % chikr(0.95, df2), 16))
P_("%s%s%s", pad("P(chi^2 > kritik)", 22),
   padl("%.6f" % float((chis > chikr(0.95, df2)).mean()), 16),
   padl("0.050000", 16))
```

# 7-topshiriq. Koʻp taqqoslash: Bonferroni, Holm, BH {#t7}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 20 ·
> **Vaqt:** 20 daqiqa

**Dizayn savoli.** $5$ janr $\Rightarrow$ barcha juftlar
$=\binom52=10$ ta test. $H_0$ hamma joyda rost boʻlsa ham
$\Prob(\text{kamida bitta ahamiyatli})=1-0{,}95^{10}$ (F154).

$$\text{Bonferroni (F155): }p^{*}=\min(mp,1);\quad
\text{Holm (F156): }(m-i+1)p_{(i)};\quad
\text{BH (F157): }\tfrac{m}{i}p_{(i)}.$$

```python
# --- 7A: 5 janr, foydalilik ulushi (BARCHA filmlar) --------------------
foydaAll = (kino["jahon_daromadi"] > 2.5 * kino["byudjet"]).values
jz = kino["janr"].values
jl = sorted(set(jz))
P_("=== 7-TOPSHIRIQ: KOʻP TAQQOSLASH ===")
P_("--- 7A: 5 janr, foydalilik ulushi (barcha %d film) ---", len(kino))
for gn in jl:
    v = foydaAll[jz == gn]
    P_("%s n = %4d  foydali = %4d  ulush = %.6f", pad(gn, 11), len(v),
       int(v.sum()), float(v.mean()))

# --- 7B: 10 ta juftlik testi -------------------------------------------
nomlar = []; zlar = []; plar = []
for i in range(4):
    for j in range(i + 1, 5):
        a = foydaAll[jz == jl[i]]; b = foydaAll[jz == jl[j]]
        xA = int(a.sum()); nAv = len(a); xB = int(b.sum()); nBv = len(b)
        pp = (xA + xB) / (nAv + nBv)
        se = math.sqrt(pp * (1 - pp) * (1 / nAv + 1 / nBv))
        zz = (xA / nAv - xB / nBv) / se
        nomlar.append("%s vs %s" % (jl[i], jl[j]))
        zlar.append(zz); plar.append(p_z2(zz))
plar = np.array(plar); zlar = np.array(zlar)
m = len(plar)
ordp = np.argsort(plar, kind="stable")            # barqaror tartiblash

bonf = np.minimum(plar * m, 1.0)                  # F155
holm = np.empty(m); acc = 0.0                     # F156
for rank in range(1, m + 1):
    ix = ordp[rank - 1]
    acc = max(acc, min((m - rank + 1) * plar[ix], 1.0))
    holm[ix] = acc
bh = np.empty(m); prev = 1.0                      # F157
for rank in range(m, 0, -1):
    ix = ordp[rank - 1]
    prev = min(prev, min(plar[ix] * m / rank, 1.0))
    bh[ix] = prev

P_("--- 7B: m = %d ta test, uchta tuzatma ---", m)
P_("%s%s%s%s%s%s", pad("juft", 24), padl("z", 10), padl("p", 12),
   padl("Bonferroni", 13), padl("Holm", 11), padl("BH (FDR)", 11))
for ix in ordp:
    P_("%s%s%s%s%s%s", pad(nomlar[ix], 24), padl("%.4f" % zlar[ix], 10),
       padl("%.6f" % plar[ix], 12), padl("%.6f" % bonf[ix], 13),
       padl("%.6f" % holm[ix], 11), padl("%.6f" % bh[ix], 11))

# --- 7C: nechtasi ahamiyatli -------------------------------------------
P_("--- 7C: alpha = 0,05 da ahamiyatli natijalar ---")
P_("tuzatishsiz %d/%d · Bonferroni %d/%d · Holm %d/%d · BH %d/%d",
   int((plar < 0.05).sum()), m, int((bonf < 0.05).sum()), m,
   int((holm < 0.05).sum()), m, int((bh < 0.05).sum()), m)
P_("Holm >= Bonferroni topilmalari? %s (bu HAR DOIM shunday)",
   ha(int((holm < 0.05).sum()) >= int((bonf < 0.05).sum())))
P_("BH >= Holm topilmalari? %s",
   ha(int((bh < 0.05).sum()) >= int((holm < 0.05).sum())))

# --- 7D: xuddi shu 10 juft, BYUDJET oʻrtachalari boʻyicha --------------
pb = []
for i in range(4):
    for j in range(i + 1, 5):
        a = kino.loc[kino["janr"] == jl[i], "byudjet"].values / 1e6
        b = kino.loc[kino["janr"] == jl[j], "byudjet"].values / 1e6
        ww = welch(a, b)
        pb.append(p_t2(ww["t"], ww["df"]))
pb = np.array(pb)
P_("--- 7D: XUDDI SHU 10 juft, lekin byudjet oʻrtachalari boʻyicha ---")
P_("tuzatishsiz %d/%d · Bonferroni %d/%d",
   int((pb < 0.05).sum()), m, int((np.minimum(pb * m, 1.0) < 0.05).sum()), m)
P_("Effektlar shu qadar kattaki, tuzatish hech narsani oʻzgartirmaydi.")
P_("Tuzatish --- 'natijani yoʻq qiladigan jazo' emas; u faqat")
P_("CHEGARADAGI natijalarga taʼsir qiladi. Kuchli effekt omon qoladi.")

# --- 7E: FWER (F154) ---------------------------------------------------
P_("--- 7E: oilaviy xato ehtimoli FWER (F154) ---")
P_("%s%s%s", pad("m", 6), padl("1 - 0,95^m", 14),
   padl("kutilgan yolgʻon topilma", 26))
for mm in [1, 2, 5, 10, 20, 50]:
    P_("%s%s%s", pad("%d" % mm, 6), padl("%.6f" % (1 - 0.95 ** mm), 14),
       padl("%.4f" % (0.05 * mm), 26))
P_("Bizning holatda m = %d -> FWER = %.6f", m, 1 - 0.95 ** m)
```

```python
# --- BH chizigʻi: p_(i) va i*alpha/m -----------------------------------
ii = np.arange(1, m + 1)
ps = np.sort(plar)
fig, ax = plt.subplots()
ax.plot(ii, ii * 0.05 / m, color=oiGreen, linewidth=1.2, label="BH: i*alpha/m")
ax.axhline(0.05 / m, color=oiVerm, linestyle="--", linewidth=1.0,
           label="Bonferroni: alpha/m")
ax.axhline(0.05, color=oiGrey, linestyle=":", linewidth=1.0, label="alpha")
ax.scatter(ii, ps, color=oiBlue, s=30, zorder=3, label="p_(i)")
ax.set_ylim(0, 0.12)
ax.set_xticks(ii)
ax.set_title("BH: p_(i) <= i*alpha/m bolgan eng katta i")
ax.set_xlabel("tartiblangan p ning orni i"); ax.set_ylabel("p")
ax.legend(frameon=False, fontsize=8)
fig.text(0.01, -0.02,
         "Yashil: BH chizigi (qiya). Qizil: Bonferroni (gorizontal). "
         "Kulrang: alpha.", fontsize=8, color="grey")
plt.show()
```

# Yakuniy tekshiruv roʻyxati

Topshirishdan oldin har bir topshiriq uchun tasdiqlang:

- [ ] **Dizayn savoliga** javob berilgan: juftmi / bogʻliqsizmi /
      kategorikmi?
- [ ] **Shart tekshiruvi** yozma ravishda bor (va buzilgan boʻlsa,
      shunday deb aytilgan).
- [ ] Har bir natijada **effekt hajmi $+$ ishonch oraligʻi $+$ $p$**
      — shu tartibda.
- [ ] $p$ hech qayerda yolgʻiz turmagan.
- [ ] Simulyatsiya natijasi yonida **generator va urugʻ**
      koʻrsatilgan.
- [ ] Talqin va qaror bosqichlari **oʻz soʻzlaringiz** bilan
      yozilgan.
