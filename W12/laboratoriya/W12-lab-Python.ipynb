```python
# --- Kerakli kutubxonalar ---------------------------------------------
import math
import numpy as np                  # sonli massivlar
import pandas as pd                 # jadval koʻrinishidagi maʼlumot
import matplotlib
import matplotlib.pyplot as plt     # grafiklar
from scipy import stats             # shapiro va boshqalar

# Okabe–Ito palitrasi: rang koʻrmaslikka moslashgan (2-haftadan beri).
oiBlue = "#0072B2"; oiOrange = "#E69F00"; oiGreen = "#009E73"
oiPink = "#CC79A7"; oiVerm   = "#D55E00"; oiSky   = "#56B4E9"
oiYell = "#F0E442"; oiGrey   = "#5A6470"

plt.rcParams.update({
    "figure.dpi": 110, "figure.figsize": (7, 4.2), "font.size": 10,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.alpha": 0.25, "axes.titleweight": "bold"})

# QOIDA (2-haftadan): grafik yorliqlarida faqat ASCII harflar.
print("MUHIT: matplotlib", matplotlib.__version__)
```

```python
# --- IKKI DAFTAR BIR XIL MATN CHIQARISHI UCHUN YORDAMCHILAR -----------
# (1) ha(): R «TRUE», Python «True» yozadi — mantiqiy qiymatni
#     OʻZIMIZ formatlaymiz.
def ha(b):
    return "HA" if bool(b) else "YOʻQ"

# (2) pad()/padl(): R ning sprintf("%-16s") formati satrni BAYT
#     boʻyicha, Python niki BELGI boʻyicha toʻldiradi. «YOʻQ» da
#     4 ta belgi, ammo 5 ta bayt bor.
def pad(s, n):
    return s + " " * max(0, n - len(s))

def padl(s, n):
    return " " * max(0, n - len(s)) + s

def P_(fmt, *a):
    print(fmt % a)

# (3) Dispersiya: MAXRAJNI HAR DOIM OSHKORA YOZAMIZ.
def varS(x):                        # maxraj n-1
    x = np.asarray(x, dtype=float)
    return float(((x - x.mean()) ** 2).sum() / (len(x) - 1))

def sdS(x):
    return math.sqrt(varS(x))

# (4) Korrelyatsiya — formulani oshkora yozamiz.
def korr(x, y):
    x = np.asarray(x, dtype=float); y = np.asarray(y, dtype=float)
    dx = x - x.mean(); dy = y - y.mean()
    return float((dx * dy).sum() / math.sqrt((dx ** 2).sum() * (dy ** 2).sum()))

# (5) ENG KICHIK KVADRATLAR — bitta funksiya, butun laboratoriya uchun.
#     F161: b1 = Sxy/Sxx · F163: b0 = ybar - b1*xbar
def ols(x, y):
    x = np.asarray(x, dtype=float); y = np.asarray(y, dtype=float)
    n = len(x); xb = x.mean(); yb = y.mean()
    sxx = float(((x - xb) ** 2).sum()); syy = float(((y - yb) ** 2).sum())
    sxy = float(((x - xb) * (y - yb)).sum())
    b1 = sxy / sxx; b0 = yb - b1 * xb
    yh = b0 + b1 * x; e = y - yh
    sse = float((e ** 2).sum()); ssr = syy - sse
    return {"n": n, "xb": float(xb), "yb": float(yb), "sxx": sxx,
            "syy": syy, "sxy": sxy, "b1": b1, "b0": b0, "yh": yh, "e": e,
            "sse": sse, "ssr": ssr, "sst": syy, "r2": ssr / syy,
            "se": math.sqrt(sse / (n - 2)),
            "r": sxy / math.sqrt(sxx * syy)}

# (6) Diagnostika: leveraj (F176), studentlashgan qoldiq (F177),
#     Kuk masofasi (F178).
def diagn(x, y):
    x = np.asarray(x, dtype=float)
    o = ols(x, y)
    h = 1 / o["n"] + (x - o["xb"]) ** 2 / o["sxx"]
    st = o["e"] / (o["se"] * np.sqrt(1 - h))
    return {"h": h, "stud": st, "cook": st ** 2 / 2 * h / (1 - h),
            "chegara": 4 / o["n"], "ort": 2 / o["n"]}

# (7) Darbin--Uotson (F180) va lag-1 korrelyatsiya.
def dw(e):
    e = np.asarray(e, dtype=float)
    return float((np.diff(e) ** 2).sum() / (e ** 2).sum())

def lag1(e):
    e = np.asarray(e, dtype=float)
    return korr(e[:-1], e[1:])

# (8) Qiyshiqlik — siljishi tuzatilgan (adjusted Fisher-Pearson).
def qiyshiq(x):
    x = np.asarray(x, dtype=float)
    n = len(x); d = x - x.mean()
    m2 = float((d ** 2).sum()) / n; m3 = float((d ** 3).sum()) / n
    return (m3 / m2 ** 1.5) * math.sqrt(n * (n - 1)) / (n - 2)

# (9) Chorak boʻyicha qoldiq SD lari — teng dispersiya sharti uchun.
#     Chegaralar: np.quantile (linear) — R dagi quantile(type = 7)
#     bilan AYNAN bir xil formula.
def chorakSD(xv, ev):
    xv = np.asarray(xv, dtype=float); ev = np.asarray(ev, dtype=float)
    q = np.quantile(xv, [0, 0.25, 0.5, 0.75, 1.0])
    out = []; nn = []
    for i in range(4):
        sel = (xv >= q[i]) & (xv <= q[i + 1]) if i == 3 else \
              (xv >= q[i]) & (xv < q[i + 1])
        out.append(sdS(ev[sel])); nn.append(int(sel.sum()))
    return {"q": q, "sd": out, "n": nn, "nisbat": max(out) / min(out)}

# (10) Nolga teng deb hisoblash: ikki tilda yigʻindi tartibi biroz
#      boshqacha boʻlishi mumkin, shuning uchun AYNIYATLAR raqam bilan
#      emas, CHEGARA bilan tekshiriladi.
def nolmi(v, eps=1e-9):
    return ha(abs(v) < eps)
```

# 1-topshiriq. Chiziqni **qoʻlda** qurish: yigʻindilardan formulaga {#t1}

> **Blum:** Qoʻllash · **Qiyinlik:** 3 (Amaliy) · **Ball:** 20 ·
> **Vaqt:** 18 daqiqa

**Savol.** Qozogʻistonda kartoshka hosildorligi $2010$–$2018$
yillarda yiliga oʻrtacha **necha t/ga** ga oʻzgargan?

**Nega qoʻlda?** Chunki `lm()` bir satrda javob beradi va siz
formulani hech qachon koʻrmaysiz. Bugun avval **beshta yigʻindi**
hisoblanadi ($n$, $\xbar$, $\ybar$, $S_{xx}$, $S_{xy}$), keyin
F161–F163 qoʻllanadi, va **faqat shundan keyin** dasturiy vosita
bilan tekshiriladi.

$$
b_1=\frac{S_{xy}}{S_{xx}} \tag{F161}\qquad
b_0=\ybar-b_1\xbar \tag{F163}\qquad
b_1=r\,\frac{s_y}{s_x} \tag{F162}
$$

```python
# --- Maʼlumot: Qozogʻiston, kartoshka, 2010-2018 ----------------------
kzp = kzq[(kzq["yil"] >= 2010) & (kzq["yil"] <= 2018)].reset_index(drop=True)
x1 = (kzp["yil"].values - 2010).astype(float)   # x = yil - 2010
y1 = kzp["kartoshka"].values.astype(float)      # t/ga

# --- Beshta yigʻindi: FORMULANI OSHKORA YOZAMIZ -----------------------
n1 = len(x1)
xb1 = x1.sum() / n1
yb1 = y1.sum() / n1
Sxx = float(((x1 - xb1) ** 2).sum())
Syy = float(((y1 - yb1) ** 2).sum())
Sxy = float(((x1 - xb1) * (y1 - yb1)).sum())

b1q = Sxy / Sxx                # F161
b0q = yb1 - b1q * xb1          # F163

P_("=== 1-TOPSHIRIQ: CHIZIQ QOʻLDA ===")
P_("n = %d   xbar = %.6f   ybar = %.6f", n1, xb1, yb1)
P_("Sxx = %.6f   Syy = %.6f   Sxy = %.6f", Sxx, Syy, Sxy)
P_("b1 = Sxy/Sxx      = %.6f  (t/ga har yilda)", b1q)
P_("b0 = ybar-b1*xbar = %.6f  (t/ga, 2010-yilda)", b0q)
P_("MODEL: yhat = %.6f + %.6f * x,   x = yil - 2010", b0q, b1q)
```

```python
# --- Dasturiy vosita bilan tekshiruv ----------------------------------
b1d, b0d = np.polyfit(x1, y1, 1)

P_("")
P_("-- Dasturiy vosita (lm) bilan taqqoslash --------------------------")
P_("qoʻlda   : b1 = %.6f   b0 = %.6f", b1q, b0q)
P_("dasturiy : b1 = %.6f   b0 = %.6f", b1d, b0d)
P_("ayirma nolga tengmi (< 1e-9): b1 %s · b0 %s",
   nolmi(b1q - b1d), nolmi(b0q - b0d))

# --- F162: qiyalik korrelyatsiyaning BIRLIKLI shakli ------------------
r1 = korr(x1, y1)
sx1 = sdS(x1); sy1 = sdS(y1)
P_("")
P_("-- F162: b1 = r * s_y/s_x -----------------------------------------")
P_("r = %.6f   s_x = %.6f   s_y = %.6f", r1, sx1, sy1)
P_("r*s_y/s_x = %.6f   b1 = %.6f   teng (< 1e-12): %s",
   r1 * sy1 / sx1, b1q, nolmi(r1 * sy1 / sx1 - b1q, 1e-12))
```

```python
# --- Qoldiqlar jadvali va IKKITA AYNIYAT ------------------------------
o1 = ols(x1, y1)
P_("")
P_("-- Qoldiqlar jadvali ----------------------------------------------")
P_("%s%s%s%s%s%s", pad("yil", 6), padl("x", 3), padl("y", 9),
   padl("yhat", 11), padl("e", 11), padl("e^2", 11))
for i in range(n1):
    P_("%s%s%s%s%s%s", pad("%d" % kzp["yil"].values[i], 6),
       padl("%d" % x1[i], 3), padl("%.3f" % y1[i], 9),
       padl("%.6f" % o1["yh"][i], 11),
       padl("%+.6f" % o1["e"][i], 11),
       padl("%.6f" % o1["e"][i] ** 2, 11))
P_("")
P_("F166: sum e_i     = 0 ga tengmi? %s", nolmi(o1["e"].sum()))
P_("F167: sum x_i e_i = 0 ga tengmi? %s", nolmi((x1 * o1["e"]).sum()))
P_("F181: chiziq (xbar, ybar) nuqtadan oʻtadimi? %s",
   nolmi(b0q + b1q * xb1 - yb1))
P_("Bu uchtasi TASODIF EMAS — ular eng kichik kvadratlar")
P_("tenglamalarining OʻZI. Ular har qanday maʼlumotda bajariladi,")
P_("model yaroqli boʻlsa ham, boʻlmasa ham. Demak ular MODEL")
P_("SIFATINING dalili emas.")
```

# 2-topshiriq. Talqin: qiyalik birliklar bilan, $R^2$ **va** $s_e$ {#t2}

> **Blum:** Tahlil · **Qiyinlik:** 3 (Amaliy) · **Ball:** 15 ·
> **Vaqt:** 14 daqiqa

**Savol.** Model qanchalik yaxshi? — Bu savol **ikkiga** boʻlinadi:
«model $x$ ni qanchalik ishlatgan?» ($R^2$) va «bashoratim
qanchalik aniq?» ($s_e$). **Bu ikki savol bir xil emas.**

$$
\text{SST}=\text{SSR}+\text{SSE}\tag{F171}\qquad
R^{2}=\frac{\text{SSR}}{\text{SST}}\tag{F172}\qquad
s_e=\sqrt{\frac{\text{SSE}}{n-2}}\tag{F174}
$$

```python
P_("=== 2-TOPSHIRIQ: R^2 VA s_e ===")
P_("SST = %.6f   SSR = %.6f   SSE = %.6f", o1["sst"], o1["ssr"], o1["sse"])
P_("F171: SST = SSR + SSE ayniyati bajariladimi? %s",
   nolmi(o1["sst"] - (o1["ssr"] + o1["sse"])))
P_("R^2 = SSR/SST = %.6f", o1["r2"])
P_("r^2           = %.6f   (F173: oddiy regressiyada teng: %s)",
   r1 ** 2, nolmi(r1 ** 2 - o1["r2"], 1e-12))
P_("s_e = sqrt(SSE/(n-2)) = sqrt(%.6f/%d) = %.6f t/ga",
   o1["sse"], n1 - 2, o1["se"])
P_("erkinlik darajasi = n - 2 = %d   (F175: ikkita parametr baholandi)",
   n1 - 2)
P_("s_y = %.6f t/ga  ->  model tipik xatoni %.2f foizga kamaytirdi",
   sy1, 100 * (1 - o1["se"] / sy1))
```

# 4-topshiriq. **L-I-N-E**: toʻrt shart, nomlangan toʻrt vosita {#t4}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 25 ·
> **Vaqt:** 24 daqiqa

| | Shart | Vosita | Muhimlik |
|---|---|---|---|
| **L** | Chiziqlilik | $e$ vs $\hat y$ — shakl boʻlmasin | **1 (eng muhim)** |
| **I** | Bogʻliqsizlik | $e$ vs tartib; lag-1; Darbin–Uotson | 2 |
| **N** | Normallik | QQ-diagramma (test emas!) | **4 (eng kam)** |
| **E** | Teng dispersiya | $e$ vs $\hat y$ — yelpigʻich boʻlmasin | 3 |

**Shartlar NIMA UCHUN kerak.** $b_0$ va $b_1$ ni hisoblash uchun
**hech qanday shart kerak emas** — eng kichik kvadratlar har doim
son beradi. Shartlar $s_e$, bashorat va **13-haftadagi xulosa**
uchun kerak.

```python
# --- L: chiziqlilik. Bashorat qiymati boʻyicha 5 guruh, har birida
#        oʻrtacha qoldiq. Chiziqli modelda ular NOL atrofida.
def Lshart(nom, o):
    n = o["n"]
    ordi = np.argsort(o["yh"], kind="stable")
    e = o["e"][ordi]
    m = []
    for i in range(1, 6):
        a = (i - 1) * n // 5; b = i * n // 5
        m.append(float(e[a:b].mean()))
    P_("%s%s", pad(nom, 26),
       "".join(padl("%+.4f" % v, 10) for v in m))
    seg = o["se"] / math.sqrt(n / 5)   # guruh oʻrtachasining standart xatosi
    return {"m": m, "seg": seg, "z": max(abs(v) for v in m) / seg}

P_("=== 4-TOPSHIRIQ: L-I-N-E ===")
P_("")
P_("-- L (chiziqlilik): yhat boʻyicha 5 guruhda oʻrtacha qoldiq -------")
P_("%s%s", pad("model", 26),
   "".join(padl("%d-guruh" % i, 10) for i in range(1, 6)))
kL1 = Lshart("Qozogʻiston kartoshka", o1)
kL2 = Lshart("Filmlar (tozalangan)", oC)
P_("guruh oʻrtachasining standart xatosi s_e/sqrt(n/5):")
P_("   kartoshka %.4f t/ga · filmlar %.4f mln $", kL1["seg"], kL2["seg"])
P_("eng katta |guruh oʻrtachasi| shu birlikda:")
P_("   kartoshka %.4f · filmlar %.4f", kL1["z"], kL2["z"])
P_("")
P_("TALQIN. Chiziqli modelda guruh oʻrtachalari nol atrofida TASODIFIY")
P_("taqsimlanadi. Filmlar modelida esa ular '+ + 0 - -' shaklida ---")
P_("bu TIZIMLI EGRILIK, yaʼni L SHUBHALI, va bu log-log modelga")
P_("oʻtishning birinchi sababi.")
P_("Kartoshka modelida n = 9 va har guruhda 1-2 kuzatuv bor, shuning")
P_("uchun bu sonli tekshiruv KUCHSIZ: kichik n da L uchun asosiy dalil")
P_("sonlar emas, DIAGRAMMA. Bu tekshiruvni katta n da ishlating.")
```

```python
# --- I: bogʻliqsizlik. AYNAN 11-haftadagi qator qaytadi. --------------
def Ishart(nom, d, ustun):
    o = ols(d["yil"].values, d[ustun].values)
    l1 = lag1(o["e"]); dws = dw(o["e"])
    cheg = 1.959964 / math.sqrt(o["n"])
    P_("%s n = %d  b1 = %.6f  R^2 = %.6f  s_e = %.6f",
       pad(nom, 22), o["n"], o["b1"], o["r2"], o["se"])
    P_("%s lag-1 = %+.6f   chegara = %.6f   DW = %.6f   -> %s",
       pad("", 22), l1, cheg, dws,
       "SHART BUZILGAN" if abs(l1) > cheg else "shart bajarilgan")
    return {"l1": l1, "dw": dws, "o": o}

P_("")
P_("-- I (bogʻliqsizlik): bir xil model, ikki mamlakat, ikki natija ---")
iu = Ishart("Oʻzbekiston bugʻdoy", uzq, "bugdoy")
ik = Ishart("Qozogʻiston bugʻdoy", kzq, "bugdoy")
P_("Nazariy bogʻliqlik: DW ~ 2(1 - lag1) = %.6f, kuzatilgan %.6f",
   2 * (1 - iu["l1"]), iu["dw"])
P_("11-HAFTADAN ESLATMA: AYNAN shu qator oʻshanda ham bogʻliqsizlik")
P_("shartini buzgan edi. Bir xil maʼlumot nuqsoni har bir usulda")
P_("qayta paydo boʻladi. Shart formuladan MUHIMROQ.")
```

```python
# --- I sharti: qoldiqlar TARTIB boʻyicha ------------------------------
fig, axs = plt.subplots(1, 2, figsize=(7, 3.2), sharex=True)
for ax, (nom, yil, e, c, mk) in zip(axs, [
        ("Uzbekistan wheat", uzq["yil"].values, iu["o"]["e"], oiBlue, "s"),
        ("Kazakhstan wheat", kzq["yil"].values, ik["o"]["e"], oiVerm, "o")]):
    ax.axhline(0, color=oiGrey, lw=0.8)
    ax.plot(yil, e, color=c, lw=1.0)
    ax.scatter(yil, e, color=c, s=14, marker=mk)
    ax.set_title(nom, fontsize=10)
    ax.set_xlabel("year")
axs[0].set_ylabel("residual (t/ha)")
fig.suptitle("Residuals in time order: independence check",
             fontweight="bold")
plt.tight_layout(); plt.show()
```

```python
# --- N: normallik. Test EMAS, QQ-diagramma qaror qiladi. --------------
def Nshart(nom, e):
    p = float(stats.shapiro(np.asarray(e, dtype=float)).pvalue)
    P_("%s n = %5d   qiyshiqlik = %+.4f   Shapiro-Wilk p = %.6f  -> %s",
       pad(nom, 26), len(e), qiyshiq(e), p,
       "H0 RAD ETILADI" if p < 0.05 else "H0 rad etilmaydi")

lx = np.log10(bx[mas]); ly = np.log10(by[mas])
oL = ols(lx, ly)
P_("")
P_("-- N (normallik): Shapiro-Wilk va qiyshiqlik ---------------------")
Nshart("Qozogʻiston kartoshka", o1["e"])
Nshart("Oʻzbekiston bugʻdoy", iu["o"]["e"])
Nshart("Filmlar (tozalangan)", oC["e"])
Nshart("Filmlar (log-log)", oL["e"])
P_("DIQQAT: n katta boʻlganda normallik testi ARZIMAS ogʻishni ham")
P_("rad etadi. Shuning uchun qaror TEST bilan emas, QQ-DIAGRAMMA")
P_("bilan qabul qilinadi. Va N --- eng KAM muhim shart: markaziy")
P_("limit teoremasi tufayli n katta boʻlsa u kerak emas.")
```

```python
# --- E: teng dispersiya. Yelpigʻich bormi? ----------------------------
cX = chorakSD(bx[mas], oC["e"])
cL = chorakSD(lx, oL["e"])
P_("")
P_("-- E (teng dispersiya): byudjet choragi boʻyicha qoldiq SD --------")
P_("%s%s%s%s%s%s", pad("model", 14), padl("1-chorak", 12),
   padl("2-chorak", 12), padl("3-chorak", 12), padl("4-chorak", 12),
   padl("nisbat", 10))
P_("%s%s%s%s%s%s", pad("xom", 14),
   "".join(padl("%.6f" % v, 12) for v in cX["sd"]),
   "", "", "", padl("%.4f" % cX["nisbat"], 10))
P_("%s%s%s%s%s%s", pad("log-log", 14),
   "".join(padl("%.6f" % v, 12) for v in cL["sd"]),
   "", "", "", padl("%.4f" % cL["nisbat"], 10))
P_("Qoʻpol mezon: nisbat < 2 boʻlsa shart bajarilgan.")
P_("xom     : %s   log-log : %s",
   "bajarilgan" if cX["nisbat"] < 2 else "BUZILGAN",
   "bajarilgan" if cL["nisbat"] < 2 else "BUZILGAN")
P_("")
P_("Log-log model: n = %d  b1 = %.6f  b0 = %.6f  R^2 = %.6f  s_e = %.6f",
   oL["n"], oL["b1"], oL["b0"], oL["r2"], oL["se"])
P_("F183 talqin: byudjet 1 foizga katta boʻlsa, daromad oʻrtacha")
P_("%.4f foizga katta (ELASTIKLIK). b1 > 1: daromad byudjetdan",
   oL["b1"])
P_("tezroq oʻsadi.")
P_("")
P_("BU KUTILGAN NATIJA EMAS EDI. Xom modelda SD byudjet bilan OʻSADI")
P_("(%.4f -> %.4f), log modelda esa KAMAYADI (%.4f -> %.4f).",
   cX["sd"][0], cX["sd"][3], cL["sd"][0], cL["sd"][3])
P_("Yaʼni logarifm yelpigʻichni YOʻQOTMADI --- uni TESKARI aylantirdi")
P_("va kuchini %.4f dan %.4f ga tushirdi. IKKALA modelda ham E sharti",
   cX["nisbat"], cL["nisbat"])
P_("BUZILGAN. QOIDA: aylantirish sehrli tayoqcha emas.")
```

```python
# --- E sharti: xom va log-log yonma-yon -------------------------------
fig, axs = plt.subplots(1, 2, figsize=(7, 3.2))
for ax, (nom, yh, e, c) in zip(axs, [
        ("raw", oC["yh"], oC["e"], oiGreen),
        ("log-log", oL["yh"], oL["e"], oiOrange)]):
    ax.axhline(0, color=oiGrey, lw=0.8)
    ax.scatter(yh, e, s=3, color=c, alpha=0.5)
    ax.set_title(nom, fontsize=10)
    ax.set_xlabel("fitted value")
axs[0].set_ylabel("residual")
fig.suptitle("Equal variance: the log transform reverses the fan",
             fontweight="bold")
plt.tight_layout(); plt.show()
```

# 5-topshiriq. Leveraj, studentlashgan qoldiq va Kuk masofasi {#t5}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 22 ·
> **Vaqt:** 21 daqiqa

$$
h_i=\frac1n+\frac{(x_i-\xbar)^2}{S_{xx}}\tag{F176}\qquad
\text{stud}_i=\frac{e_i}{s_e\sqrt{1-h_i}}\tag{F177}
$$
$$
D_i=\frac{\text{stud}_i^{2}}{2}\cdot\frac{h_i}{1-h_i}\tag{F178}
\qquad D_i>\frac4n\ \Rightarrow\ \text{qarab chiq}\tag{F179}
$$

**Uchta boshqa-boshqa tushuncha.** $h_i$ — nuqta $x$ boʻyicha
qanchalik chetda (**$y$ ga bogʻliq emas**). $e_i$ — nuqta chiziqdan
qanchalik uzoq. $D_i$ — nuqtani olib tashlasa chiziq qanchalik
siljiydi. **Faqat uchinchisi «taʼsirchanlik» ni oʻlchaydi.**

```python
# --- Tojikiston, kartoshka, 1992-2018 ---------------------------------
xt = tjq["yil"].values.astype(float); yt = tjq["kartoshka"].values.astype(float)
oT = ols(xt, yt); dT = diagn(xt, yt)

P_("=== 5-TOPSHIRIQ: TAʼSIRCHAN NUQTALAR ===")
P_("Tojikiston kartoshka 1992-2018 (toʻliq model):")
P_("b1 = %.6f t/ga har yilda   b0 = %.6f", oT["b1"], oT["b0"])
P_("R^2 = %.6f   s_e = %.6f t/ga", oT["r2"], oT["se"])
P_("")
P_("-- Uchta koʻrsatkich, uchta BOSHQA javob ------------------------")
iL = int(np.argmax(dT["h"])); iR = int(np.argmax(np.abs(dT["stud"])))
iC = int(np.argmax(dT["cook"]))
P_("Eng katta leveraj      : %d-yil  h = %.6f   (oʻrtacha 2/n = %.6f)",
   xt[iL], dT["h"][iL], dT["ort"])
P_("Eng katta |stud. qoldiq|: %d-yil  stud = %.6f", xt[iR], dT["stud"][iR])
P_("Eng katta Kuk masofasi : %d-yil  D = %.6f   (chegara 4/n = %.6f)",
   xt[iC], dT["cook"][iC], dT["chegara"])
P_("")
P_("Eng katta LEVERAJLI nuqtaning Kuk masofasi atigi %.6f.",
   dT["cook"][iL])
P_("Yaʼni KATTA LEVERAJ TAʼSIRCHANLIK DEGANI EMAS.")
P_("Eng taʼsirchan nuqtani taʼsirchan qilgan narsa leveraj emas,")
P_("STUDENTLASHGAN QOLDIQ: %.6f.", dT["stud"][iC])
```

```python
# --- Taʼsirchan nuqtasiz QAYTA baholash: IKKALA natija ham beriladi ---
kes = np.arange(len(xt)) != iC
oT2 = ols(xt[kes], yt[kes])
P_("")
P_("-- %d-yil qiymati va qoʻshnilari -------------------------------",
   xt[iC])
for yy in range(int(xt[iC]) - 1, int(xt[iC]) + 2):
    P_("   %d : %.3f t/ga", yy, float(yt[xt == yy][0]))
P_("")
P_("%s%s%s", pad("model", 22), padl("barchasi", 14),
   padl("%d-yilsiz" % xt[iC], 14))
P_("%s%s%s", pad("b1 (t/ga har yil)", 22),
   padl("%.6f" % oT["b1"], 14), padl("%.6f" % oT2["b1"], 14))
P_("%s%s%s", pad("b0", 22), padl("%.6f" % oT["b0"], 14),
   padl("%.6f" % oT2["b0"], 14))
P_("%s%s%s", pad("R^2", 22), padl("%.6f" % oT["r2"], 14),
   padl("%.6f" % oT2["r2"], 14))
P_("%s%s%s", pad("s_e (t/ga)", 22), padl("%.6f" % oT["se"], 14),
   padl("%.6f" % oT2["se"], 14))
P_("b1 oʻzgarishi: %+.4f foiz", 100 * (oT2["b1"] - oT["b1"]) / oT["b1"])
P_("")
P_("QAROR EMAS, DALIL: bu nuqtani oʻchirish yoki saqlash qarori")
P_("MAʼLUMOTNING KELIB CHIQISHIDAN chiqadi, natijadan emas.")
P_("R^2 ning %.4f dan %.4f ga koʻtarilishi --- SABAB emas,",
   oT["r2"], oT2["r2"])
P_("OGOHLANTIRISH. Ikkala natija ham hisobotda koʻrsatiladi.")
```

```python
# --- TAQQOSLASH: bir xil diagnostika, IKKI XIL toʻgʻri javob ----------
xr = kzq["yil"].values.astype(float); yr = kzq["sholi"].values.astype(float)
oR = ols(xr, yr); dR = diagn(xr, yr)
jC = int(np.argmax(dR["cook"]))
kes2 = np.arange(len(xr)) != jC
oR2 = ols(xr[kes2], yr[kes2])
P_("")
P_("-- Qozogʻiston sholi 1992-2018 -----------------------------------")
P_("b1 = %.6f   b0 = %.6f   R^2 = %.6f   s_e = %.6f",
   oR["b1"], oR["b0"], oR["r2"], oR["se"])
P_("Eng taʼsirchan: %d-yil  y = %.3f  h = %.6f  stud = %.6f  D = %.6f",
   xr[jC], yr[jC], dR["h"][jC], dR["stud"][jC], dR["cook"][jC])
P_("Shu nuqtasiz  : b1 = %.6f  R^2 = %.6f  s_e = %.6f ; %+.4f foiz",
   oR2["b1"], oR2["r2"], oR2["se"],
   100 * (oR2["b1"] - oR["b1"]) / oR["b1"])
P_("Keyingi yillar: %s",
   " · ".join("%d = %.3f" % (v, float(yr[xr == v][0]))
              for v in range(1992, 1996)))
P_("")
P_("IKKI NUQTA, BIR XIL DIAGNOSTIKA, IKKI XIL TOʻGʻRI JAVOB:")
P_("  Tojikiston 1995 --- qoʻshnilaridan ikki baravar yuqori, keyingi")
P_("    yil qaytib tushadi, mexanizm yoʻq -> ehtimol MANBADAGI XATO.")
P_("    Qaror: manbani tekshir; tuzat yoki chiqarib tashla va YOZIB QOʻY.")
P_("  Qozogʻiston 1992 --- pasayish 1995 gacha UZLUKSIZ davom etadi va")
P_("    mexanizmi maʼlum (sugʻorish tizimining qulashi) -> REAL TARIX.")
P_("    Qaror: nuqtani SAQLA, MODELNI oʻzgartir.")
P_("Chetlashgan kuzatuvni 'natija chiroyli boʻlsin' deb oʻchirish ---")
P_("bu 11-haftadagi p-hacking ning regressiyadagi shakli.")
```

# 6-topshiriq. Bashorat va **ekstrapolyatsiyani rad etish** {#t6}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 18 ·
> **Vaqt:** 18 daqiqa

**Savol.** 1-topshiriqdagi model bilan nimani bashorat qilish
mumkin — va nimani **rad etish** kerak?

```python
P_("=== 6-TOPSHIRIQ: BASHORAT VA EKSTRAPOLYATSIYA ===")
P_("Model: yhat = %.6f + %.6f x,  x = yil - 2010,  x in [%d; %d]",
   b0q, b1q, x1.min(), x1.max())
P_("s_e = %.6f t/ga", o1["se"])
P_("")
P_("%s%s%s%s", pad("yil", 7), padl("x", 5), padl("yhat", 11),
   padl("holat", 22))
for yy in [2010 + d for d in (4, 8, 9, 15, 40, 90)]:   # 2014 ... 2100
    xx = yy - 2010
    P_("%s%s%s%s", pad("%d" % yy, 7), padl("%d" % xx, 5),
       padl("%.4f" % (b0q + b1q * xx), 11),
       padl("maʼlumot ichida" if x1.min() <= xx <= x1.max()
            else "EKSTRAPOLYATSIYA", 22))
P_("")
P_("2100-yil uchun model %.4f t/ga deydi. Bugungi JAHON REKORDI",
   b0q + b1q * 90)
P_("taxminan 45-50 t/ga. Model MAʼNOSIZ javob berdi va hech qanday")
P_("ogohlantirish bermadi --- chunki u shunchaki toʻgʻri chiziq.")
P_("R^2 = %.6f ham buni koʻrsatmaydi.", o1["r2"])
```

```python
# --- Ekstrapolyatsiyaning IKKINCHI turi: notoʻgʻri OBYEKT -------------
uzp = uzq[(uzq["yil"] >= 2010) & (uzq["yil"] <= 2018)].reset_index(drop=True)
bash = b0q + b1q * (uzp["yil"].values - 2010)
kuz = uzp["kartoshka"].values.astype(float)
rmse = math.sqrt(float(((kuz - bash) ** 2).mean()))
P_("")
P_("-- Notoʻgʻri obyekt: Qozogʻiston modeli Oʻzbekistonga -----------")
P_("Kuzatilgan : %s", " ".join("%.3f" % v for v in kuz))
P_("Bashorat   : %s", " ".join("%.3f" % v for v in bash))
P_("RMSE = %.6f t/ga ; modelning oʻz s_e si = %.6f", rmse, o1["se"])
P_("Yaʼni %.4f marta yomonroq. x oraligʻi TOʻGʻRI edi ---",
   rmse / o1["se"])
P_("OBYEKT notoʻgʻri edi.")
P_("")
P_("XULOSA (yozib qoʻying): har bir model IKKI chegara bilan")
P_("chegaralangan --- x ning ORALIGʻI va modelning OBYEKTI.")
P_("Chiziq bu chegaralarni bilmaydi. Ularni siz aytasiz.")
```

```python
# --- Ekstrapolyatsiya rasmi -------------------------------------------
xg = np.arange(0, 91, dtype=float)
yg = b0q + b1q * xg
fig, ax = plt.subplots()
ax.fill_between(xg[xg > 8], 0, yg[xg > 8], color=oiVerm, alpha=0.07)
ax.plot(xg, yg, color=oiVerm, lw=1.6)
ax.axhline(48, color="#A8410F", ls=":")
ax.axvline(8, color=oiGrey, ls="--")
ax.scatter(x1, y1, color=oiBlue, s=28, zorder=3)
ax.text(30, 44, "world record ~ 48 t/ha", fontsize=8, color="#A8410F")
ax.text(10, 70, "data ends here", fontsize=8, color=oiGrey)
ax.set_ylim(10, 80)
ax.set_title("The line does not know where the data ended")
ax.set_xlabel("x = year - 2010"); ax.set_ylabel("potato yield (t/ha)")
plt.tight_layout(); plt.show()
```

bashorat soʻradi. **Rad etish xatini yozing** ($120$–$180$ soʻz):
*(1)* modelning oraligʻi qanday, *(2)* $2035$ uchun model qanday
son beradi va nega unga ishonib boʻlmaydi, *(3)* mijozga
buning oʻrniga nima taklif qilasiz.
