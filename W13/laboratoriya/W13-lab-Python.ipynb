```python
# --- Kerakli kutubxonalar ---------------------------------------------
import math
import numpy as np                  # sonli massivlar
import pandas as pd                 # jadval koʻrinishidagi maʼlumot
import matplotlib
import matplotlib.pyplot as plt     # grafiklar
from scipy import stats             # t, chi^2 va F taqsimotlari

# Okabe–Ito palitrasi: rang koʻrmaslikka moslashgan (2-haftadan beri).
oiBlue = "#0072B2"; oiOrange = "#E69F00"; oiGreen = "#009E73"
oiPink = "#CC79A7"; oiVerm   = "#D55E00"; oiSky   = "#56B4E9"
oiYell = "#F0E442"; oiGrey   = "#5A6470"

plt.rcParams.update({
    "figure.dpi": 110, "figure.figsize": (7, 4.2), "font.size": 10,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.alpha": 0.25, "axes.titleweight": "bold"})

# QOIDA (2-haftadan): grafik yorliqlarida faqat ASCII harflar.
print("MUHIT: matplotlib", matplotlib.__version__)
```

```python
# --- IKKI DAFTAR BIR XIL MATN CHIQARISHI UCHUN YORDAMCHILAR -----------
# (1) ha(): R «TRUE», Python «True» yozadi — mantiqiy qiymatni
#     OʻZIMIZ formatlaymiz.
def ha(b):
    return "HA" if bool(b) else "YOʻQ"

# (2) pad()/padl(): R BAYT boʻyicha, Python BELGI boʻyicha toʻldiradi.
def pad(s, n):
    return s + " " * max(0, n - len(s))

def padl(s, n):
    return " " * max(0, n - len(s)) + s

def P_(fmt, *a):
    print(fmt % a)

# (3) Dispersiya va korrelyatsiya: MAXRAJ HAR DOIM OSHKORA.
def varS(x):
    x = np.asarray(x, dtype=float)
    return float(((x - x.mean()) ** 2).sum() / (len(x) - 1))

def sdS(x):
    return math.sqrt(varS(x))

def korr(x, y):
    x = np.asarray(x, dtype=float); y = np.asarray(y, dtype=float)
    dx = x - x.mean(); dy = y - y.mean()
    return float((dx * dy).sum() / math.sqrt((dx ** 2).sum() * (dy ** 2).sum()))

# (4) ENG KICHIK KVADRATLAR + XULOSA. 12-haftadagi ols() ga bu hafta
#     TOʻRTTA yangi maydon qoʻshildi: seb1 (F184), seb0 (F187),
#     t1 (F185) va p1. QOIDA: qiyalik hech qachon SE va oraliqsiz
#     keltirilmaydi.
def ols(x, y):
    x = np.asarray(x, dtype=float); y = np.asarray(y, dtype=float)
    n = len(x); xb = x.mean(); yb = y.mean()
    sxx = float(((x - xb) ** 2).sum()); syy = float(((y - yb) ** 2).sum())
    sxy = float(((x - xb) * (y - yb)).sum())
    b1 = sxy / sxx; b0 = yb - b1 * xb
    yh = b0 + b1 * x; e = y - yh
    sse = float((e ** 2).sum()); ssr = syy - sse
    se = math.sqrt(sse / (n - 2))
    seb1 = se / math.sqrt(sxx)                    # F184
    seb0 = se * math.sqrt(1 / n + xb ** 2 / sxx)  # F187
    t1 = b1 / seb1                                # F185
    return {"n": n, "xb": float(xb), "yb": float(yb), "sxx": sxx,
            "syy": syy, "sxy": sxy, "b1": b1, "b0": b0, "yh": yh, "e": e,
            "sse": sse, "ssr": ssr, "sst": syy, "r2": ssr / syy, "se": se,
            "r": sxy / math.sqrt(sxx * syy), "seb1": seb1, "seb0": seb0,
            "t1": t1, "p1": 2 * float(stats.t.cdf(-abs(t1), n - 2))}

# (5) t-taqsimot: kvantil va ikki tomonlama p-qiymat.
def tq(pp, df):
    return float(stats.t.ppf(pp, df))

def tp2(tt, df):
    return 2 * float(stats.t.cdf(-abs(tt), df))

# (6) IKKI ORALIQ (F190--F193). MAJBURIY ARGUMENT `tur`: hech qachon
#     «oraliq» deb yozilmaydi, har doim QAYSI oraliq ekani aytiladi.
def oraliq(o, xs, tur, konf=0.95):
    yh = o["b0"] + o["b1"] * xs
    ich = 1 / o["n"] + (xs - o["xb"]) ** 2 / o["sxx"]
    sef = o["se"] * math.sqrt(ich) if tur == "ortacha" \
        else o["se"] * math.sqrt(1 + ich)
    tc = tq(1 - (1 - konf) / 2, o["n"] - 2)
    return {"yh": yh, "se": sef, "lo": yh - tc * sef, "hi": yh + tc * sef,
            "kenglik": 2 * tc * sef, "tur": tur}

# (7) MINSTD LCG: x <- 48271 x mod 2147483647. Ikki tilda AYNAN bir xil
#     ketma-ketlik beradi — shuning uchun butstrep natijasi ham bir xil.
class LCG:
    def __init__(self, seed):
        self.x = seed % 2147483647

    def uv(self, n):
        o = np.empty(n); x = self.x
        for i in range(n):
            x = (48271 * x) % 2147483647
            o[i] = x / 2147483647
        self.x = x
        return o

    def idx(self, n, hi):
        return (self.uv(n) * hi).astype(int)   # 0 dan boshlanadi

# (8) Ayniyatlar raqam bilan emas, CHEGARA bilan tekshiriladi.
def nolmi(v, eps=1e-9):
    return ha(abs(v) < eps)

# (9) Kvantil: np.quantile (linear) --- R quantile(type = 7) bilan bir xil.
def kv(x, pp):
    return np.quantile(np.asarray(x, dtype=float), pp)

# (10) chi^2 bogʻliqsizlik testi (11-haftadan) --- formulani oshkora yozamiz.
def chi2(a, b):
    a = np.asarray(a); b = np.asarray(b)
    ta = sorted(set(a.tolist())); tb = sorted(set(b.tolist()))
    O = np.zeros((len(ta), len(tb)))
    for i, va in enumerate(ta):
        for j, vb in enumerate(tb):
            O[i, j] = float(((a == va) & (b == vb)).sum())
    nn = O.sum()
    E = np.outer(O.sum(axis=1), O.sum(axis=0)) / nn
    st = float(((O - E) ** 2 / E).sum())
    df = (O.shape[0] - 1) * (O.shape[1] - 1)
    return {"stat": st, "df": df, "p": float(stats.chi2.sf(st, df))}
```

```python
# --- Ikkita HAQIQIY maʼlumot toʻplami ---------------------------------
# A. markaziy-osiyo-hosildorlik.csv — 5 mamlakat x 27 yil (1992–2018),
#    hosildorlik t/ga. Manba: FAOSTAT / Our World in Data (OWID).
# B. kinolar-byudjet.csv — 2 467 film. Manba: The Numbers / Kaggle.
DATA = "data"
hosil = pd.read_csv(DATA + "/markaziy-osiyo-hosildorlik.csv", encoding="utf-8")
kino = pd.read_csv(DATA + "/kinolar-byudjet.csv", encoding="utf-8")

# QOIDA: oʻzbekcha mamlakat nomini LITERAL yozmaymiz — fayldan olamiz.
mamlar = list(pd.unique(hosil["mamlakat"]))
UZ = mamlar[0]; KG = mamlar[1]; KZ = mamlar[2]
TJ = mamlar[3]; TM = mamlar[4]
uzq = hosil[hosil["mamlakat"] == UZ].reset_index(drop=True)
kzq = hosil[hosil["mamlakat"] == KZ].reset_index(drop=True)

# 12-haftadagi model AYNAN qaytadi: Qozogʻiston kartoshka 2010-2018.
kzp = kzq[(kzq["yil"] >= 2010) & (kzq["yil"] <= 2018)].reset_index(drop=True)
x1 = (kzp["yil"].values - 2010).astype(float)   # x = yil - 2010
y1 = kzp["kartoshka"].values.astype(float)      # t/ga
o1 = ols(x1, y1)

P_("")
P_("=== 0-BOSQICH: MAʼLUMOT ===")
P_("A. hosildorlik fayli: %d qator, %d mamlakat, %d-%d yillar",
   len(hosil), len(mamlar), hosil["yil"].min(), hosil["yil"].max())
P_("   pozitsiyalar: 1 = %s · 3 = %s", UZ, KZ)
P_("B. kinolar fayli: %d film", len(kino))
P_("")
P_("Bugungi asosiy model (12-haftadan): kartoshka ~ (yil - 2010),")
P_("n = %d, x oraligʻi [%d; %d]", o1["n"], x1.min(), x1.max())
P_("MANBA: OWID/FAOSTAT (hosildorlik) · The Numbers/Kaggle (kinolar).")
P_("Ikkalasi ham HAQIQIY maʼlumot --- hech qanday simulyatsiya yoʻq.")
```

# 1-topshiriq. Qiyalik uchun **toʻliq** xulosa {#t1}

> **Blum:** Qoʻllash · **Qiyinlik:** 3 (Amaliy) · **Ball:** 20 ·
> **Vaqt:** 18 daqiqa

**Savol.** $b_1=0{,}588033$ t/ga — bu *tanlanma* bahosi. Boshqa toʻqqiz
yil olsak, boshqa son chiqardi. **Qanchalik boshqa?**

$$
\SE(b_1)=\frac{s_e}{\sqrt{S_{xx}}}\tag{F184}\qquad
t=\frac{b_1-\beta_1^{(0)}}{\SE(b_1)},\ \mathrm{df}=n-2\tag{F185}
$$
$$
b_1\pm t^{*}\SE(b_1)\tag{F186}\qquad
\SE(b_0)=s_e\sqrt{\tfrac1n+\tfrac{\xbar^{2}}{S_{xx}}}\tag{F187}
$$

```python
# --- Qiyalik uchun xulosa: SE, t, df, p, ishonch oraligʻi -------------
df1 = o1["n"] - 2
tc1 = tq(0.975, df1)
lo1 = o1["b1"] - tc1 * o1["seb1"]
hi1 = o1["b1"] + tc1 * o1["seb1"]

P_("=== 1-TOPSHIRIQ: QIYALIK UCHUN TOʻLIQ XULOSA ===")
P_("Model: kartoshka ~ (yil - 2010), n = %d", o1["n"])
P_("b1 = %.6f t/ga har yilda   b0 = %.6f t/ga", o1["b1"], o1["b0"])
P_("R^2 = %.6f   s_e = %.6f t/ga   S_xx = %.6f", o1["r2"], o1["se"], o1["sxx"])
P_("")
P_("F184: SE(b1) = s_e/sqrt(S_xx) = %.6f/sqrt(%.6f) = %.6f",
   o1["se"], o1["sxx"], o1["seb1"])
P_("F187: SE(b0) = s_e*sqrt(1/n + xbar^2/S_xx)      = %.6f", o1["seb0"])
P_("F185: t = b1/SE(b1) = %.6f/%.6f = %.6f   df = n-2 = %d",
   o1["b1"], o1["seb1"], o1["t1"], df1)
P_("      p = %.8f  ->  %s", o1["p1"],
   "H0 RAD ETILADI" if o1["p1"] < 0.05 else "H0 rad etilmaydi")
P_("      t_(0,975)(%d) = %.6f", df1, tc1)
P_("F186: 95 foiz IO = %.6f +- %.6f*%.6f = [%.6f; %.6f]",
   o1["b1"], tc1, o1["seb1"], lo1, hi1)
P_("      nol oraliq ICHIDAMI? %s  -> test va oraliq %s",
   ha(lo1 <= 0 <= hi1),
   "MOS" if (o1["p1"] < 0.05) == (not (lo1 <= 0 <= hi1)) else "ZID")
P_("")
P_("TALQIN (birliklar bilan). 2010-2018 yillarda Qozogʻistonda kartoshka")
P_("hosildorligi har yili oʻrtacha %.4f t/ga oshgan; aniqlik %.4f t/ga.",
   o1["b1"], tc1 * o1["seb1"])
```

```python
# --- UCHTA TEST, BITTA SAVOL: qiyalik = korrelyatsiya = ANOVA F -------
tr1 = o1["r"] * math.sqrt(o1["n"] - 2) / math.sqrt(1 - o1["r"] ** 2)   # F188
MSR = o1["ssr"] / 1
MSE = o1["sse"] / (o1["n"] - 2)
Fst = MSR / MSE                                                        # F189
pF = float(stats.f.sf(Fst, 1, o1["n"] - 2))

P_("")
P_("-- Uchta test, bitta savol ----------------------------------------")
P_("1) Qiyalik testi : t = %.8f   p = %.8f", o1["t1"], o1["p1"])
P_("2) Korrelyatsiya : t = r*sqrt(n-2)/sqrt(1-r^2) = %.8f   p = %.8f",
   tr1, tp2(tr1, o1["n"] - 2))
P_("3) ANOVA         : F = MSR/MSE = %.6f/%.6f = %.8f   p = %.8f",
   MSR, MSE, Fst, pF)
P_("t^2 = %.8f ; F ga tengmi (< 1e-9)? %s", o1["t1"] ** 2,
   nolmi(o1["t1"] ** 2 - Fst))
P_("t(qiyalik) = t(korrelyatsiya) (< 1e-9)? %s", nolmi(o1["t1"] - tr1))
P_("")
P_("-- ANOVA jadvali --------------------------------------------------")
P_("%s%s%s%s%s", pad("Manba", 12), padl("SS", 14), padl("df", 5),
   padl("MS", 14), padl("F", 12))
P_("%s%s%s%s%s", pad("Model", 12), padl("%.6f" % o1["ssr"], 14),
   padl("1", 5), padl("%.6f" % MSR, 14), padl("%.6f" % Fst, 12))
P_("%s%s%s%s", pad("Qoldiq", 12), padl("%.6f" % o1["sse"], 14),
   padl("%d" % (o1["n"] - 2), 5), padl("%.6f" % MSE, 14))
P_("%s%s%s", pad("Jami", 12), padl("%.6f" % o1["sst"], 14),
   padl("%d" % (o1["n"] - 1), 5))
P_("Tekshiruv: SSR/SST = %.6f = R^2 = %.6f", o1["ssr"] / o1["sst"], o1["r2"])
P_("")
P_("DIQQAT: bu ayniyat faqat BITTA tushuntiruvchi oʻzgaruvchili modelda")
P_("oʻrinli. Koʻp omilli regressiyada F butun modelni, t esa bitta")
P_("koeffitsiyentni tekshiradi va ular zid natija berishi mumkin.")
```

```python
# --- SE(b1) UCHTA narsaga bogʻliq: s_e, n va x ning TARQOQLIGI --------
# Uchinchisi deyarli har doim unutiladi. Uni oʻlchaymiz: bir xil n = 9,
# bir xil s_e, faqat x ning joylashuvi oʻzgaradi.
xdiz = [("0..8 (bizning misol)", np.arange(9.0)),
        ("faqat 3..5 (siqilgan)", np.array([3, 3, 4, 4, 4, 4, 4, 5, 5.])),
        ("faqat 0 va 8 (chekkalar)", np.array([0, 0, 0, 0, 4, 8, 8, 8, 8.]))]
P_("")
P_("-- Uchinchi omil: bir xil n, bir xil s_e, turli x tarqoqligi ------")
P_("%s%s%s", pad("x qiymatlari", 28), padl("S_xx", 12),
   padl("SE(b1) nisbati", 16))
baza = None
for nm, xs in xdiz:
    sx = float(((xs - xs.mean()) ** 2).sum()); rel = math.sqrt(1 / sx)
    if baza is None:
        baza = rel
    P_("%s%s%s", pad(nm, 28), padl("%.4f" % sx, 12),
       padl("%.4f" % (rel / baza), 16))
P_("")
P_("x ni SIQISH standart xatoni bir necha marta OSHIRADI; chekkalarga")
P_("toʻplash esa KAMAYTIRADI --- bir xil n va bir xil s_e bilan.")
P_("Bu DIZAYN masalasi, hisoblash masalasi emas. Ikkita chegara bor:")
P_("(a) x ni kengaytirish ekstrapolyatsiyaga olib bormasin (12-hafta);")
P_("(b) faqat chekkalardan olsangiz, chiziqlilikni TEKSHIRIB boʻlmaydi.")
P_("")
P_("SHART TEKSHIRUVI (12-haftadan). Bu model uchun L bajarilgan,")
P_("N bajarilgan (Shapiro-Wilk p = 0.741843), E da yelpigʻich yoʻq;")
P_("I --- n = 9 juda kichik, xulosa ehtiyotkor. Demak yuqoridagi")
P_("IO va p-qiymat OʻRINLI. 5-topshiriqda shart BUZILGAN holatni")
P_("koʻramiz va oʻsha yerda xulosa oʻzgaradi.")
```

# 2-topshiriq. **Ikki oraliq**: chiziq qayerda va keyingi kuzatuv qayerda {#t2}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 20 ·
> **Vaqt:** 18 daqiqa

$$
\SE_{\text{oʻrtacha}}=s_e\sqrt{\tfrac1n+\tfrac{(x^{*}-\xbar)^2}{S_{xx}}}
\tag{F190}\qquad
\SE_{\text{bashorat}}=s_e\sqrt{1+\tfrac1n+\tfrac{(x^{*}-\xbar)^2}{S_{xx}}}
\tag{F192}
$$

# 4-topshiriq. Butstrep qiyaligi va $t$-oraliq {#t4}

**Savol.** $t$-oraliq **formuladan** keladi va formula shartlarga
tayanadi. Butstrep esa maʼlumotning oʻzidan keladi. **Ikkalasi bir
xil javob beradimi?**

```python
# --- Butstrep: n ta juftlikni QAYTARIB tanlaymiz (F194) ---------------
# MINSTD LCG, urugʻ 20261301, 20000 takror. Ikki tilda AYNAN bir xil.
M = 20000
g = LCG(20261301)
bs = np.empty(M); yaroqsiz = 0
for i in range(M):
    ii = g.idx(o1["n"], o1["n"])
    xx = x1[ii]; yy = y1[ii]
    sx = float(((xx - xx.mean()) ** 2).sum())
    if sx < 1e-12:
        bs[i] = np.nan; yaroqsiz += 1
    else:
        bs[i] = float(((xx - xx.mean()) * (yy - yy.mean())).sum()) / sx
bsv = bs[~np.isnan(bs)]
qb = kv(bsv, [0.025, 0.975])

P_("=== 4-TOPSHIRIQ: BUTSTREP ===")
P_("Takrorlar: %d · urugʻ 20261301 · MINSTD LCG (48271, 2147483647)", M)
P_("Yaroqsiz takrorlar (barcha x teng): %d", yaroqsiz)
P_("Butstrep oʻrtachasi = %.6f   (kuzatilgan b1 = %.6f)",
   bsv.mean(), o1["b1"])
P_("Butstrep SE         = %.6f   (formula SE   = %.6f)",
   sdS(bsv), o1["seb1"])
P_("Butstrep SE / formula SE = %.4f", sdS(bsv) / o1["seb1"])
P_("")
P_("%s%s%s", pad("oraliq", 22), padl("chegaralar", 24), padl("kenglik", 11))
P_("%s%s%s", pad("persentil (butstrep)", 22),
   padl("[%.6f; %.6f]" % (qb[0], qb[1]), 24),
   padl("%.6f" % (qb[1] - qb[0]), 11))
P_("%s%s%s", pad("t-oraliq (formula)", 22),
   padl("[%.6f; %.6f]" % (lo1, hi1), 24),
   padl("%.6f" % (hi1 - lo1), 11))
P_("Kenglik nisbati (butstrep / t) = %.4f",
   (qb[1] - qb[0]) / (hi1 - lo1))
```

# 5-topshiriq. **I sharti buzilgan**: samarali tanlanma hajmi {#t5}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 17 ·
> **Vaqt:** 16 daqiqa

$$
n_{\text{eff}}=n\,\frac{1-\rho}{1+\rho}\tag{F199}\qquad
\SE_{\text{tuzatilgan}}\approx\SE\cdot\sqrt{\frac{1+\rho}{1-\rho}}
$$

```python
# --- Oʻzbekiston bugʻdoy ~ yil: 12-haftada I sharti BUZILGAN edi ------
oE = ols(uzq["yil"].values.astype(float), uzq["bugdoy"].values.astype(float))
er = oE["e"]
rho = korr(er[:-1], er[1:])                    # lag-1 korrelyatsiya
lim = 1.959964 / math.sqrt(oE["n"])            # z_(0,975)/sqrt(n)
dfE = oE["n"] - 2
tcE = tq(0.975, dfE)

P_("=== 5-TOPSHIRIQ: SHART BUZILGANDA XULOSA ===")
P_("Model: bugʻdoy ~ yil, n = %d", oE["n"])
P_("b1 = %.6f t/ga har yilda   s_e = %.6f t/ga   S_xx = %.6f",
   oE["b1"], oE["se"], oE["sxx"])
P_("")
P_("-- SODDA (shart bajarilgan deb faraz qilib) -----------------------")
P_("SE(b1) = %.6f   t = %.6f   df = %d   p = %.3e",
   oE["seb1"], oE["t1"], dfE, oE["p1"])
P_("95 foiz IO: [%.6f; %.6f]  ->  %s",
   oE["b1"] - tcE * oE["seb1"], oE["b1"] + tcE * oE["seb1"],
   "H0 RAD ETILADI" if oE["p1"] < 0.05 else "H0 rad etilmaydi")
P_("")
P_("-- SHART TEKSHIRUVI: I (bogʻliqsizlik) ---------------------------")
P_("Qoldiqlar lag-1 = %.6f   chegara z_(0,975)/sqrt(n) = %.6f", rho, lim)
P_("|lag-1| > chegara?  %s  ->  %s", ha(abs(rho) > lim),
   "I SHARTI BUZILGAN" if abs(rho) > lim else "shart bajarilgan")
```

```python
# --- Samarali tanlanma hajmi va TUZATILGAN xulosa (F199) -------------
neff = oE["n"] * (1 - rho) / (1 + rho)
infl = math.sqrt((1 + rho) / (1 - rho))
seT = oE["seb1"] * infl
tT = oE["b1"] / seT
dfT = max(neff - 2, 1)
pT = tp2(tT, dfT)

P_("")
P_("-- TUZATILGAN xulosa ---------------------------------------------")
P_("1-rho = %.6f   1+rho = %.6f", 1 - rho, 1 + rho)
P_("F199: n_eff = n(1-rho)/(1+rho) = %d*%.6f/%.6f = %.6f",
   oE["n"], 1 - rho, 1 + rho, neff)
P_("Yaʼni %d ta yil taxminan %.2f ta bogʻliqsiz kuzatuvga teng.",
   oE["n"], neff)
P_("SE tuzatish koeffitsiyenti sqrt((1+rho)/(1-rho)) = %.6f", infl)
P_("Tuzatilgan SE = %.6f * %.6f = %.6f", oE["seb1"], infl, seT)
P_("Tuzatilgan t  = %.6f   df ~ %.4f   p ~ %.6f", tT, dfT, pT)
P_("")
P_("%s%s%s", pad("usul", 16), padl("SE(b1)", 12), padl("p", 14))
P_("%s%s%s", pad("sodda", 16), padl("%.6f" % oE["seb1"], 12),
   padl("%.3e" % oE["p1"], 14))
P_("%s%s%s", pad("tuzatilgan", 16), padl("%.6f" % seT, 12),
   padl("%.6f" % pT, 14))
P_("Xulosa oʻzgardimi? %s", ha((oE["p1"] < 0.05) != (pT < 0.05)))
P_("")
P_("QOIDA. Buzilgan shart b1 ni EMAS, unga boʻlgan ISHONCHNI buzadi.")
P_("b1 = %.6f baribir toʻgʻri hisoblangan; oʻzgargani --- uning",
   oE["b1"])
P_("atrofidagi oraliq va p-qiymat.")
P_("")
P_("NIMA ISBOTLANMADI. n_eff --- QOʻPOL tuzatish: u AR(1) tuzilmasini")
P_("faraz qiladi va qoldiqlardagi toʻlqin CHIZIQLILIK buzilishidan")
P_("kelib chiqmaganini TEKSHIRMAYDI. x = yil boʻlgani uchun L va I ni")
P_("bu maʼlumot bilan ajratib boʻlmaydi (12-hafta). Shuning uchun")
P_("hisobotda IKKALA talqin ham koʻrsatiladi.")
```

# 6-topshiriq. **Chalgʻituvchi omil**: uchta shart, ikkita nomzod {#t6}

**Savol.** PG-13 va R reytingli filmlar **foydalilik** boʻyicha farq

**Uchinchi oʻzgaruvchi $z$ chalgʻituvchi omil boʻlishi uchun (F196):**

1. $z$ **$x$ bilan** bogʻliq boʻlishi kerak;
2. $z$ **$y$ bilan** bogʻliq boʻlishi kerak;
3. $z$ **$x$ ning oqibati boʻlmasligi** kerak.

Bugun **ikkita ishonarli nomzod** tekshiriladi. **Ulardan biri
shartlarni bajaradi, ikkinchisi bajarmaydi** — va buni faqat
**tekshirish** orqali bilish mumkin.

```python
# --- Xom farq ---------------------------------------------------------
kk = kino[(kino["byudjet"] > 0) & (kino["jahon_daromadi"] > 0)].copy()
kk["b"] = kk["byudjet"] / 1e6
kk["gr"] = kk["jahon_daromadi"] / 1e6
kk["foyda"] = kk["gr"] > 2.5 * kk["b"]     # IKKALA oʻzgaruvchining oqibati!
sel = kk[kk["reyting"].isin(["PG-13", "R"])].reset_index(drop=True)
reytlar = ["PG-13", "R"]

P_("=== 6-TOPSHIRIQ: CHALGʻITUVCHI OMIL ===")
P_("Savol: PG-13 va R filmlari foydalilik boʻyicha farq qiladimi?")
P_("'Foydali' = jahon daromadi > 2,5 x byudjet")
P_("")
P_("%s%s%s%s", pad("reyting", 10), padl("foydali", 10), padl("jami", 8),
   padl("ulush", 12))
ul = [0.0, 0.0]
for i in range(2):
    d = sel[sel["reyting"] == reytlar[i]]
    ul[i] = float(d["foyda"].mean())
    P_("%s%s%s%s", pad(reytlar[i], 10),
       padl("%d" % int(d["foyda"].sum()), 10), padl("%d" % len(d), 8),
       padl("%.6f" % ul[i], 12))
dXom = ul[0] - ul[1]
P_("Xom farq = %.6f (PG-13 foydasiga)", dXom)
P_("")
P_("SHART TEKSHIRUVI. Bu regressiya emas, ikkita ulushning farqi;")
P_("shuning uchun L-I-N-E emas, 11-haftadagi shartlar amal qiladi:")
P_("kuzatuvlar bogʻliqsiz (davomli filmlar --- eng zaif joyi),")
P_("har bir katakda kutilgan son 10 dan katta (n = %d).", len(sel))
```

```python
# --- NOMZOD 1: byudjet ------------------------------------------------
# Ishonarli koʻrinadi: PG-13 filmlari qimmatroq, qimmat filmlar
# koʻproq daromad keltiradi. Shartlarni TEKSHIRAMIZ.
pg = (sel["reyting"] == "PG-13").values.astype(float)
fy = sel["foyda"].values.astype(float)
bb = sel["b"].values.astype(float)
nS = len(sel)
r_bx = korr(bb, pg)          # (1) z ~ x
r_by = korr(bb, fy)          # (2) z ~ y
t_bx = r_bx * math.sqrt(nS - 2) / math.sqrt(1 - r_bx ** 2)
t_by = r_by * math.sqrt(nS - 2) / math.sqrt(1 - r_by ** 2)
p_bx = tp2(t_bx, nS - 2); p_by = tp2(t_by, nS - 2)
bm1 = float(bb[pg == 1].mean()); bm2 = float(bb[pg == 0].mean())

P_("")
P_("-- NOMZOD 1: BYUDJET ---------------------------------------------")
P_("Oʻrtacha byudjet: PG-13 %.6f mln $ · R %.6f mln $", bm1, bm2)
P_("farq = %.6f mln $ (%.4f foiz)", bm1 - bm2, (bm1 - bm2) / bm2 * 100)
P_("(1) byudjet ~ reyting   : r = %.6f  p = %.4e  -> %s",
   r_bx, p_bx, "BOGʻLIQ" if p_bx < 0.05 else "BOGʻLIQ EMAS")
P_("(2) byudjet ~ foydalilik: r = %.6f  p = %.4e  -> %s",
   r_by, p_by, "BOGʻLIQ" if p_by < 0.05 else "BOGʻLIQ EMAS")
P_("(3) byudjet reytingning oqibatimi? Yoʻq --- byudjet ssenariy")
P_("    bosqichida belgilanadi, reyting film tayyor boʻlgach beriladi.")
P_("XULOSA: byudjet chalgʻituvchi omil %s --- %d-shart bajarilmadi.",
   "HISOBLANADI" if (p_bx < 0.05 and p_by < 0.05) else "HISOBLANMAYDI", 2)
P_("")
P_("Baribir qatlamlab koʻramiz (byudjet boʻyicha uchdan bir qatlamlar):")
qb3 = kv(bb, [1 / 3, 2 / 3])
qat = np.where(bb <= qb3[0], 1, np.where(bb <= qb3[1], 2, 3))
nomq = ["past", "oʻrta", "yuqori"]
P_("%s%s%s%s%s%s", pad("qatlam", 10), padl("PG-13", 10), padl("n", 6),
   padl("R", 10), padl("n", 6), padl("farq", 11))
dw = 0.0; ww = 0
for k in (1, 2, 3):
    a = fy[(qat == k) & (pg == 1)]; b = fy[(qat == k) & (pg == 0)]
    dw += (a.mean() - b.mean()) * (len(a) + len(b))
    ww += len(a) + len(b)
    P_("%s%s%s%s%s%s", pad(nomq[k - 1], 10),
       padl("%.6f" % a.mean(), 10), padl("%d" % len(a), 6),
       padl("%.6f" % b.mean(), 10), padl("%d" % len(b), 6),
       padl("%+.6f" % (a.mean() - b.mean()), 11))
dBud = dw / ww
P_("Vaznlangan farq = %.6f (xom %.6f, oʻzgarish %+.2f foiz)",
   dBud, dXom, (dBud - dXom) / abs(dXom) * 100)
P_("KUTILGANIDEK: shart bajarilmagan boʻlsa, tuzatish ham kam narsani")
P_("oʻzgartiradi.")
```

```python
# --- NOMZOD 2: janr ---------------------------------------------------
c1 = chi2(sel["janr"].values, sel["reyting"].values)   # (1) z ~ x
c2 = chi2(sel["janr"].values, sel["foyda"].values)     # (2) z ~ y
janrlar = sorted(set(sel["janr"].tolist()))
jn = sel["janr"].values

P_("")
P_("-- NOMZOD 2: JANR ------------------------------------------------")
P_("(1) janr ~ reyting   : chi^2 = %.4f  df = %d  p = %.4e  -> %s",
   c1["stat"], c1["df"], c1["p"],
   "BOGʻLIQ" if c1["p"] < 0.05 else "BOGʻLIQ EMAS")
P_("(2) janr ~ foydalilik: chi^2 = %.4f  df = %d  p = %.4e  -> %s",
   c2["stat"], c2["df"], c2["p"],
   "BOGʻLIQ" if c2["p"] < 0.05 else "BOGʻLIQ EMAS")
P_("(3) janr reytingning oqibatimi? Yoʻq --- janr ssenariy bosqichida,")
P_("    reyting esa film tayyor boʻlgach beriladi. Vaqt tartibi toʻgʻri.")
P_("XULOSA: janr chalgʻituvchi omil %s.",
   "HISOBLANADI" if (c1["p"] < 0.05 and c2["p"] < 0.05) else "hisoblanmaydi")
P_("")
P_("-- Janr boʻyicha qatlamlash (F197) -------------------------------")
P_("%s%s%s%s%s%s", pad("janr", 12), padl("PG-13", 10), padl("n", 6),
   padl("R", 10), padl("n", 6), padl("farq", 11))
dg = 0.0; wg = 0; ishoralar = []
for gn in janrlar:
    a = fy[(jn == gn) & (pg == 1)]; b = fy[(jn == gn) & (pg == 0)]
    if len(a) < 10 or len(b) < 10:
        continue
    dd = a.mean() - b.mean(); ishoralar.append(dd)
    dg += dd * (len(a) + len(b)); wg += len(a) + len(b)
    P_("%s%s%s%s%s%s", pad(gn, 12), padl("%.6f" % a.mean(), 10),
       padl("%d" % len(a), 6), padl("%.6f" % b.mean(), 10),
       padl("%d" % len(b), 6), padl("%+.6f" % dd, 11))
dAdj = dg / wg
P_("Vaznlangan (janrga tuzatilgan) farq = %.6f", dAdj)
P_("Xom farq                            = %.6f", dXom)
P_("Oʻzgarish = %+.6f (%+.2f foiz)", dAdj - dXom,
   (dAdj - dXom) / abs(dXom) * 100)
P_("Barcha janrlarda ishora bir xilmi? %s",
   ha(all(v > 0 for v in ishoralar) or all(v < 0 for v in ishoralar)))
P_("Ishora oʻzgardimi? %s ; kattaligi oʻsdimi? %s",
   ha((dAdj > 0) != (dXom > 0)), ha(abs(dAdj) > abs(dXom)))
P_("")
P_("DIQQAT --- KUTILMAGAN YOʻNALISH. Odatda chalgʻituvchi omilni")
P_("tuzatish farqni KICHRAYTIRADI. Bu yerda esa u farqni OSHIRDI.")
P_("Sabab: R filmlari foydaliligi PAST janrda koʻproq, PG-13 esa")
P_("foydaliligi YUQORI janrlarda. Janr farqni YASHIRAYOTGAN edi ---")
P_("bu 'salbiy chalgʻitish' deyiladi va u xuddi shunday real.")
P_("")
P_("NIMA ISBOTLANMADI (F201). Janrga tuzatilgan farq 'PG-13 reytingi")
P_("foydalilikni oshiradi' degani EMAS. Beshta izohdan faqat BITTASI")
P_("(tasodif) yoʻqotildi. Ochiq qolganlari: (a) teskari sababiyat ---")
P_("studiya kutilgan daromadga qarab montajni reytingga moslashi;")
P_("(b) OʻLCHANMAGAN chalgʻituvchi omillar --- rejissyor, yulduz,")
P_("chiqish sanasi, davomli film ekani; (c) tanlash siljishi --- faylda")
P_("faqat prokatga chiqqan filmlar bor. Bularni yoʻqotish uchun")
P_("DIZAYN kerak, hisoblash emas --- va reytingni tasodifiy tayinlab")
P_("boʻlmaydi.")
```

```python
# --- Qatlamlash rasmi: janrlar boʻyicha farq --------------------------
nomlar = []; upg = []; ur = []
for gn in janrlar:
    a = fy[(jn == gn) & (pg == 1)]; b = fy[(jn == gn) & (pg == 0)]
    if len(a) < 10 or len(b) < 10:
        continue
    nomlar.append(gn); upg.append(a.mean()); ur.append(b.mean())
ix = np.arange(len(nomlar))
fig, ax = plt.subplots()
ax.bar(ix - 0.19, upg, width=0.34, color=oiBlue, label="PG-13")
ax.bar(ix + 0.19, ur, width=0.34, color=oiOrange, label="R")
ax.axhline(ul[0], color=oiBlue, ls="--")
ax.axhline(ul[1], color=oiOrange, ls=":")
ax.set_xticks(ix); ax.set_xticklabels(nomlar)
ax.legend(frameon=False)
ax.set_title("Profitable share by genre: the gap is wider inside strata")
ax.set_xlabel("genre"); ax.set_ylabel("share profitable")
plt.tight_layout(); plt.show()
```
