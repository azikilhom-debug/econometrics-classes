```python
# --- IKKI DAFTAR BIR XIL MATN CHIQARISHI UCHUN YORDAMCHILAR ------------
# (1) tf(): R «TRUE», Python esa «True» yozadi. Mantiqiy qiymatni
#     OʻZIMIZ formatlaymiz.
def tf(b):
    return "TRUE" if bool(b) else "FALSE"

# (2) pad()/padl(): R ning sprintf("%-16s") formati satrni BAYT boʻyicha
#     toʻldiradi, Python esa BELGI boʻyicha. «Oʻzbekiston» da 11 ta belgi,
#     ammo 12 ta bayt bor — shu sababli oʻzbekcha yorliqlarni formatga
#     bermay, kenglikni OʻZIMIZ hisoblaymiz.
def pad(s, n):                      # chapga tekislash (oʻngdan boʻshliq)
    return s + " " * max(0, n - len(s))

def padl(s, n):                     # oʻngga tekislash (chapdan boʻshliq)
    return " " * max(0, n - len(s)) + s

# (3) P_(): R dagi sprintf + cat ekvivalenti. Har bir son QOʻLDA
#     formatlanadi, tilning avtomatik chiqarish mexanizmiga tayanmaymiz.
def P_(fmt, *a):
    print(fmt % a)

# (4) Bosh toʻplam koʻrsatkichlari — MAXRAJI N (n-1 EMAS!).
#     Python da np.var(x) maxraji N (bosh toʻplam), R da esa var(x)
#     maxraji n-1 (tanlanma). Ikki til ikki xil standartni tanlagan,
#     shuning uchun ikkalasini ham OʻZIMIZ yozamiz.
def dispersiya_bosh(x):
    x = np.asarray(x, dtype=float)
    return float(np.mean((x - x.mean()) ** 2))

def ogish_bosh(x):
    return math.sqrt(dispersiya_bosh(x))

# (5) Qiyshiqlik va ekssess kurtoz — moment taʼrifi boʻyicha
#     (scipy.stats.skew / kurtosis bilan bir xil, bias=True).
def qiyshiqlik(x):
    x = np.asarray(x, dtype=float); d = x - x.mean()
    m2 = np.mean(d ** 2); m3 = np.mean(d ** 3)
    return float(m3 / m2 ** 1.5)

def kurtoz(x):
    x = np.asarray(x, dtype=float); d = x - x.mean()
    m2 = np.mean(d ** 2); m4 = np.mean(d ** 4)
    return float(m4 / m2 ** 2 - 3)

# (6) Kvantil: numpy ning standart ("linear") usuli R ning type=7 usuli
#     bilan AYNAN bir xil. Boshqa usul ISHLATMANG.
def kv(x, p):
    return float(np.quantile(np.asarray(x, dtype=float), p, method="linear"))
```

```python
# --- MINSTD generatori (2-haftadan beri oʻzgarmagan) -------------------
# Nima uchun oʻzimizning generatorimiz kerak?
#   R:      set.seed(2026); runif(3)
#   Python: np.random.default_rng(2026).random(3)
# Bu ikkisi BOSHQA-BOSHQA sonlar beradi. Ikkalasi ham toʻgʻri, lekin
# ikkalasi ham bir-biriga mos emas. Shu sababli kursda tasodifiylik
# kerak boʻlganda HAR DOIM quyidagi 4 qatorli generator ishlatiladi:
#     x <- (48271 * x) mod 2147483647,    u <- x / 2147483647
# Bu — Lehmer (1951) generatori, «MINSTD» nomi bilan mashhur.
A_ = 48271; M_ = 2147483647

def minstd(urug):
    holat = {"x": urug % M_}
    def gen(n):                        # n ta u ~ U(0,1) qaytaradi
        o = np.empty(int(n)); x = holat["x"]
        for i in range(int(n)):
            x = (A_ * x) % M_; o[i] = x
        holat["x"] = x
        return o / M_
    return gen

# Indeks tanlash: 0 dan hi-1 gacha butun son (R bilan bir xil).
# Python da vektor 0 dan boshlanadi, shuning uchun +1 KERAK EMAS.
def indeks(gen, n, hi):
    return (gen(n) * hi).astype(np.int64)

# Nazorat qiymati: urugʻ 2026 dan keyingi uchta son.
g = minstd(2026); u3 = g(3)
P_("=== 0-BOSQICH: MINSTD NAZORATI ===")
P_("urugʻ 2026 -> u1 = %.10f  u2 = %.10f  u3 = %.10f", u3[0], u3[1], u3[2])
P_("bu uchta son ikkinchi tildagi daftarda ham AYNAN shunday chiqishi shart")

# Solishtirish uchun: tilning OʻZ generatori (bu satr ikki tilda FARQ QILADI)
print("MUHIT: Python ning default_rng(2026).random() = %.10f  "
      "(R da boshqa son chiqadi)"
      % np.random.default_rng(2026).random())
```

# 1-topshiriq. Haqiqiy ustun — chekli bosh toʻplam {#t1}

> **Blum:** Qoʻllash · **Qiyinlik:** 2 (Konseptual) · **Ball:** 15 ·
> **Vaqt:** 12 daqiqa

**Statistik tushuncha.** Butun hafta bitta ajratishga tayanadi:

| | Bosh toʻplam (barcha birliklar) | Tanlanma ($n$ ta birlik) |
|---|---|---|
| Belgi | $\mu$, $\sigma$, $p$ — **parametr** | $\xbar$, $s$, $\phat$ — **statistika** |
| Alifbo | yunon | lotin |
| Qiymati | **qatʼiy son**, noaniqligi yoʻq | **tasodifiy miqdor** |
| Maxraj | $\sigma^{2}=\frac{1}{N}\sum(x_i-\mu)^{2}$ | $s^{2}=\frac{1}{n-1}\sum(x_i-\xbar)^{2}$ |

Bugun biz maʼlumot faylini **butun bosh toʻplam** deb qabul qilamiz.
Bu sunʼiy holat, ammo u bizga *haqiqatni bilish* imkonini beradi: $\mu$ va
$\sigma$ ni **hisoblab qoʻyamiz**, keyin tanlanmalar ularga qanchalik
yaqin tushishini oʻlchaymiz. Haqiqiy tadqiqotda $\mu$ hech qachon
maʼlum boʻlmaydi — shuning uchun bugungi mashq **kalitni koʻrib turib**
oʻynaladigan oʻyin.

> **Tuzoq.** R da `var(x)` va `sd(x)` **$n-1$** ga boʻladi — bu
> *tanlanma* koʻrsatkichi. Python da `np.var(x)` **$N$** ga boʻladi —
> bu *bosh toʻplam* koʻrsatkichi. Ikki til ikki xil standartni tanlagan.
> Shuning uchun 0-bosqichda `dispersiya_bosh()` va `ogish_bosh()`
> funksiyalarini **oʻzimiz** yozdik.

**Maʼlumot.** (a) 2018-yilgi bugʻdoy hosildorligi, 5 ta Markaziy Osiyo
mamlakati, t/ga — $N=5$ lik *kichkina* bosh toʻplam.
(b) 2 467 ta film byudjeti (mln $) — $N=2467$ lik *katta* bosh toʻplam.

**Kod.**

```python
# --- 1A: N = 5 lik bosh toʻplam ---------------------------------------
# Qator POZITSIYA boʻyicha tanlanadi (mamlakat nomini literal yozmaymiz).
h18 = hosil[hosil["yil"] == 2018].copy()
h18["_t"] = [mamlakatlar.index(m) for m in h18["mamlakat"]]
h18 = h18.sort_values("_t")
X = h18["bugdoy"].to_numpy(dtype=float)
N = len(X)

P_("")
P_("=== 1-TOPSHIRIQ (A): N = 5 LIK BOSH TOʻPLAM ===")
P_("2018-yil bugʻdoy hosildorligi, t/ga  [manba: OWID / FAOSTAT]")
for i in range(N):
    P_("  %s %s %8.3f", kod5[i], pad(mamlakatlar[i], 13), X[i])

mu   = float(X.mean())
sig2 = dispersiya_bosh(X)            # maxraj N — PARAMETR
sig  = ogish_bosh(X)
s2   = float(np.var(X, ddof=1))      # maxraj n-1 — STATISTIKA
P_("")
P_("PARAMETR  mu       = %.4f/%d = %.6f", X.sum(), N, mu)
P_("PARAMETR  sigma^2  = %.6f/%d = %.6f  (maxraj N)",
   ((X - mu) ** 2).sum(), N, sig2)
P_("PARAMETR  sigma    = %.6f", sig)
P_("STATISTIKA s^2 (R var()) = %.6f  (maxraj n-1) -- BOSHQA SON!", s2)
P_("nisbat s^2/sigma^2 = %.6f = N/(N-1) = %.6f", s2 / sig2, N / (N - 1))

P_("")
P_("Ogʻishlar (x - mu) va ularning yigʻindisi:")
P_("  %s", "  ".join("%+.4f" % v for v in (X - mu)))
P_("  |yigʻindi| = %.10f  -> nolga teng: %s",
   abs((X - mu).sum()), tf(abs((X - mu).sum()) < 1e-12))
```

```python
# --- 1B: N = 2467 lik haqiqiy bosh toʻplam ----------------------------
b   = kino["byudjet"].to_numpy(dtype=float) / 1e6    # mln $ ga oʻtkazamiz
Nb  = len(b)
mub = float(b.mean()); sigb = ogish_bosh(b)

P_("")
P_("=== 1-TOPSHIRIQ (B): N = 2467 LIK BOSH TOʻPLAM ===")
P_("Kino byudjetlari, mln $  [manba: The Numbers / Kaggle, 2000-2018]")
P_("N        = %d", Nb)
P_("mu       = %.6f", mub)
P_("sigma    = %.6f", sigb)
P_("median   = %.4f", np.median(b))
P_("min; max = %.4f; %.4f", b.min(), b.max())
P_("qiyshiqlik gamma1 = %.6f", qiyshiqlik(b))
P_("ekssess kurtoz    = %.6f", kurtoz(b))
P_("")
P_("mu > median (%.4f > %.4f): %s  -> oʻngga qiyshiq",
   mub, np.median(b), tf(mub > np.median(b)))
P_("")
P_("5%% qoidasi: chekli bosh toʻplam tuzatmasini n < 0.05*N boʻlsa eʼtiborsiz")
P_("qoldirsa boʻladi.  0.05 * %d = %.2f", Nb, 0.05 * Nb)
P_("Demak n <= %d gacha sigma/sqrt(n) formulasi tuzatmasiz ishlaydi.",
   int(math.floor(0.05 * Nb)))
```

# 2-topshiriq. Barcha tanlanmalarni sanab chiqish {#t2}

> **Blum:** Qoʻllash · **Qiyinlik:** 3 (Amaliy) · **Ball:** 20 ·
> **Vaqt:** 18 daqiqa

**Statistik tushuncha.**

> **Taʼrif.** $\bar{X}$ ning **tanlanma taqsimoti** — bu bosh toʻplamdan
> olinishi mumkin boʻlgan *barcha* $n$ hajmli tanlanmalar boʻyicha
> $\bar{X}$ qiymatlarining taqsimoti.

Bu taʼrif odatda mavhum tuyuladi, chunki «barcha tanlanmalar» cheksiz
koʻp boʻladi. $N=5$, $n=2$ da esa ular **atigi 10 ta** (qaytarishsiz)
yoki **25 ta** (qaytarish bilan). Shuning uchun bugun biz taʼrifni
*taxminan* emas, **toʻliq** bajaramiz — va $\E[\Xbar]=\mu$ hamda
$\Var(\Xbar)=\sigma^{2}/n$ ni jadvaldan **oʻqib olamiz**.

Isbot (6-haftadan): $X_1,\dots,X_n$ bogʻliqsiz va bir xil taqsimlangan boʻlsa,
$$\E[\Xbar]=\frac{1}{n}\sum\E[X_i]=\mu,\qquad
  \Var(\Xbar)=\frac{1}{n^{2}}\sum\Var(X_i)=\frac{\sigma^{2}}{n}.$$
Ikkinchi tenglikda $\Cov(X_i,X_j)=0$ **ishlatildi** — bogʻliqsizlik aynan
shu bitta qatorda kerak boʻladi. Qaytarishsiz tanlashda tanlanmalar
bogʻliqsiz **emas**, shuning uchun natija $\frac{\sigma^{2}}{n}\cdot
\frac{N-n}{N-1}$ ga oʻzgaradi (**chekli bosh toʻplam tuzatmasi**, fpc).

**Kod.**

```python
# --- 2A: qaytarishSIZ — C(5,2) = 10 ta tanlanma -----------------------
import itertools
juft = list(itertools.combinations(range(N), 2))     # 10 ta juftlik
mwor = np.array([(X[i] + X[j]) / 2 for i, j in juft])

P_("")
P_("=== 2-TOPSHIRIQ (A): QAYTARISHSIZ, n = 2, BARCHA 10 TANLANMA ===")
P_("%s %s %s", pad("tanlanma", 16), padl("qiymatlar", 20), padl("Xbar", 10))
for k, (i, j) in enumerate(juft):
    P_("%s %s %s",
       pad("{%s, %s}" % (mamlakatlar[i][:4], mamlakatlar[j][:4]), 16),
       padl("(%.3f + %.3f)/2" % (X[i], X[j]), 20),
       padl("%.4f" % mwor[k], 10))
E_wor = float(mwor.mean()); V_wor = dispersiya_bosh(mwor)
fpc   = (N - 2) / (N - 1)
P_("")
P_("E[Xbar]   = %.6f   mu = %.6f   -> siljimagan: %s",
   E_wor, mu, tf(abs(E_wor - mu) < 1e-12))
P_("Var(Xbar) = %.6f", V_wor)
P_("sigma^2/n = %.6f   -> BU YERDA MOS EMAS", sig2 / 2)
P_("sigma^2/n * (N-n)/(N-1) = %.6f * %.6f = %.6f -> mos: %s",
   sig2 / 2, fpc, sig2 / 2 * fpc, tf(abs(V_wor - sig2 / 2 * fpc) < 1e-12))
P_("fpc koeffitsiyenti sqrt((N-n)/(N-1)) = %.6f", math.sqrt(fpc))
P_("SE = %.6f", math.sqrt(V_wor))
```

```python
# --- 2B: qaytarish BILAN — 5^2 = 25 ta tanlanma -----------------------
# Sikl tartibi aniq yozilgan: i — tashqi, j — ichki. Shunda R va Python
# bir xil TARTIBDAGI 25 ta tanlanmani hosil qiladi.
i1 = np.repeat(np.arange(N), N)          # i tashqi sikl
i2 = np.tile(np.arange(N), N)            # j ichki sikl
mwr = (X[i1] + X[i2]) / 2

P_("")
P_("=== 2-TOPSHIRIQ (B): QAYTARISH BILAN, 25 TA TANLANMA ===")
P_("5x5 jadval, Xbar qiymatlari:")
P_("%s%s", pad("", 8), "".join(padl(kod5[j], 9) for j in range(N)))
for i in range(N):
    satr = "".join(padl("%.4f" % ((X[i] + X[j]) / 2), 9) for j in range(N))
    P_("%s%s", pad(kod5[i], 8), satr)
E_wr = float(mwr.mean()); V_wr = dispersiya_bosh(mwr)
P_("")
P_("E[Xbar]   = %.6f = mu : %s", E_wr, tf(abs(E_wr - mu) < 1e-12))
P_("Var(Xbar) = %.6f   sigma^2/n = %.6f  -> mos: %s",
   V_wr, sig2 / 2, tf(abs(V_wr - sig2 / 2) < 1e-12))
P_("SE(Xbar)  = %.6f   sigma/sqrt(2) = %.6f",
   math.sqrt(V_wr), sig / math.sqrt(2))
P_("Turli xil Xbar qiymatlari: %d ta (25 tanlanmadan)",
   len(np.unique(np.round(mwr, 4))))
P_("Kenglik: bosh toʻplam [%.3f; %.3f] -> Xbar [%.4f; %.4f]  -> torayadi",
   X.min(), X.max(), mwr.min(), mwr.max())
P_("")
P_("SE solishtirish: qaytarishsiz %.6f  <  qaytarish bilan %.6f : %s",
   math.sqrt(V_wor), math.sqrt(V_wr), tf(math.sqrt(V_wor) < math.sqrt(V_wr)))
```

```python
# --- 2C: NIMA UCHUN n-1? Javob 25 ta tanlanmada -----------------------
# Har bir tanlanma uchun IKKI xil dispersiya hisoblaymiz va ularning
# 25 ta tanlanma boʻyicha OʻRTACHASINI sigma^2 bilan solishtiramiz.
s2_n1 = np.empty(N * N); s2_n = np.empty(N * N)
for k in range(N * N):
    tan_ = np.array([X[i1[k]], X[i2[k]]])
    s2_n1[k] = ((tan_ - tan_.mean()) ** 2).sum() / (2 - 1)   # maxraj n-1
    s2_n[k]  = ((tan_ - tan_.mean()) ** 2).sum() / 2         # maxraj n
P_("")
P_("=== 2-TOPSHIRIQ (C): NIMA UCHUN n-1 GA BOʻLAMIZ? ===")
P_("sigma^2 (haqiqat)          = %.6f", sig2)
P_("E[s^2]   maxraj n-1 boʻlsa = %.6f  -> siljimagan: %s",
   s2_n1.mean(), tf(abs(s2_n1.mean() - sig2) < 1e-12))
P_("E[s^2_n] maxraj n   boʻlsa = %.6f  -> SILJIGAN (past baholaydi)",
   s2_n.mean())
P_("siljish nisbati = %.6f = (n-1)/n = %.6f", s2_n.mean() / sig2, (2 - 1) / 2)
P_("")
P_("Xulosa: n-1 maxraji «tuzatish» emas, u siljishni AYNAN yoʻqotadi.")
P_("Sabab: tanlanma oʻrtachasi tanlanmaning OʻZIDAN olingan, shuning uchun")
P_("kuzatuvlar oʻz oʻrtachasiga mu ga qaraganda YAQINROQ turadi.")
```

# 5-topshiriq. MLT panjarasi va Kolmogorov masofasi {#t5}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 20 ·
> **Vaqt:** 22 daqiqa

**Statistik tushuncha.**

> **Markaziy limit teoremasi (MLT).** $X_1,\dots,X_n$ bogʻliqsiz, bir xil
> taqsimlangan, $\E[X]=\mu$, $0<\Var(X)=\sigma^{2}<\infty$ boʻlsa,
> $$Z_n=\frac{\Xbar-\mu}{\sigma/\sqrt{n}}\ \xrightarrow{\ d\ }\ \mathcal{N}(0,1).$$

Darsliklarning koʻpchiligida «$n\ge30$ boʻlsa MLT ishlaydi» deb
yoziladi. **Bu qoida emas.** MLT — *limit* teoremasi; u $n=30$ da
qanchalik yaqinlashganini **aytmaydi**. Yaqinlashish tezligi bosh
toʻplamning **qiyshiqligiga** bogʻliq.

Bugun biz «yetarlimi?» degan savolni **koʻz bilan emas, son bilan**
hal qilamiz. Oʻlchov —

> **Kolmogorov masofasi.**
> $D_n=\sup_x\bigl|F_{Z_n}(x)-\Phi(x)\bigr|$ — empirik taqsimot
> funksiyasi bilan standart normal taqsimot funksiyasi orasidagi eng
> katta farq. $D_n=0{,}05$ degani: har qanday $x$ da ehtimollik xatosi
> $5$ foiz punktdan oshmaydi.

**Maʼlumot.** Toʻrtta bosh toʻplam, qiyshiqligi oʻsish tartibida:
tekis $U(0,1)$ ($\gamma_1=0$) · haqiqiy kino byudjetlari ($1{,}617$) ·
koʻrsatkichli $\Expo(1)$ ($2{,}000$) · $\Bern(0{,}10)$ ($2{,}667$).

**Kod.**

```python
# --- 5A: toʻrtta bosh toʻplamni tayyorlash -----------------------------
# Uchtasi generatsiya qilinadi, toʻrtinchisi HAQIQIY maʼlumot.
# Urugʻlar maʼruza slaydlaridagi bilan bir xil (gen/01_verify_w07.py).
gp1 = minstd(20260101); POP_tekis = gp1(400000)
gp2 = minstd(20260102); POP_exp   = -np.log(gp2(400000))
gp3 = minstd(20260103); POP_bern  = (gp3(400000) < 0.10).astype(float)

# Har bir bosh toʻplam uchun: nomi, massiv, mu, sigma, tanlash urugʻi.
# Urugʻ = 20260200 + n + 7*(nom uzunligi) — maʼruza skriptidagi qoida;
# quyida u tayyor son sifatida yozilgan (nom uzunligi tildan qatʼi nazar
# bir xil boʻlishi uchun).
nomlar = ["Tekis U(0,1)", "Kino byudjeti", "Koʻrsatkichli Exp(1)",
          "Bernulli p=0,10"]
urugla = [20260284, 20260361, 20260340, 20260305]
poplar = [POP_tekis, b, POP_exp, POP_bern]
mu_lar = [0.5, mub, 1.0, 0.10]
sg_lar = [math.sqrt(1 / 12), sigb, 1.0, math.sqrt(0.09)]
g1_lar = [0.0, g1b, 2.0, (1 - 2 * 0.1) / math.sqrt(0.09)]

P_("")
P_("=== 5-TOPSHIRIQ (A): TOʻRTTA BOSH TOʻPLAM ===")
P_("%s%s%s%s%s", pad("bosh toʻplam", 24), padl("N", 9), padl("mu", 12),
   padl("sigma", 12), padl("gamma1", 10))
for i in range(4):
    P_("%s%s%s%s%s", pad(nomlar[i], 24), padl("%d" % len(poplar[i]), 9),
       padl("%.4f" % mu_lar[i], 12), padl("%.4f" % sg_lar[i], 12),
       padl("%.4f" % g1_lar[i], 10))
```

```python
# --- 5B: Kolmogorov masofasi -------------------------------------------
# sup|F_n(x) - Phi(x)| — empirik taqsimot funksiyasini saralab hisoblaymiz.
def kolmogorov(z):
    zs = np.sort(z); Fn = np.arange(1, len(zs) + 1) / len(zs)
    return float(np.max(np.abs(Fn - stats.norm.cdf(zs))))

n_grid = [1, 2, 5, 30, 100]
D_mat  = np.zeros((4, len(n_grid)))
Q_mat  = np.zeros((4, len(n_grid)))

P_("")
P_("=== 5-TOPSHIRIQ (B): MLT PANJARASI (m = 20 000) ===")
P_("Har katakda: qiyshiqlik(Xbar) | Kolmogorov masofasi sup|F-Phi|")
P_("%s%s", pad("bosh toʻplam", 24),
   "".join(padl("n=%d" % n_, 17) for n_ in n_grid))
for i in range(4):
    arr = poplar[i]; satr = ""
    for j, n_ in enumerate(n_grid):
        gl  = minstd(urugla[i] + n_)
        idx = indeks(gl, 20000 * n_, len(arr))
        S   = arr[idx].reshape(20000, n_)
        xb  = S.mean(axis=1)
        z   = (xb - mu_lar[i]) / (sg_lar[i] / math.sqrt(n_))
        Q_mat[i, j] = qiyshiqlik(xb); D_mat[i, j] = kolmogorov(z)
        satr += padl("%.3f" % Q_mat[i, j], 9) + padl("%.4f" % D_mat[i, j], 8)
    P_("%s%s", pad(nomlar[i], 24), satr)
```

```python
# --- 5C: «n >= 30» qoidasini SINASH ------------------------------------
j30 = n_grid.index(30)
P_("")
P_("=== 5-TOPSHIRIQ (C): «n >= 30» QOIDASI SINOVDAN OʻTDIMI? ===")
P_("n = 30 dagi Kolmogorov masofasi:")
for i in range(4):
    P_("  %s gamma1 = %s  D_30 = %s", pad(nomlar[i], 24),
       padl("%.3f" % g1_lar[i], 7), padl("%.4f" % D_mat[i, j30], 8))
P_("")
P_("Eng yomon / eng yaxshi nisbat = %.4f / %.4f = %.1f barobar",
   D_mat[3, j30], D_mat[0, j30], D_mat[3, j30] / D_mat[0, j30])
P_("Ayni bir n da, ayni bir m da. Farqni tugʻdirgan narsa — QIYSHIQLIK.")
P_("")
P_("Qiyshiqlik qonuni gamma1(Xbar) = gamma1(X)/sqrt(n) tekshiruvi:")
P_("%s%s%s%s", padl("n", 6), padl("oʻlchangan", 14), padl("nazariy", 12),
   padl("farq", 10))
for j, n_ in enumerate(n_grid):
    P_("%s%s%s%s", padl("%d" % n_, 6),
       padl("%.4f" % Q_mat[1, j], 14),
       padl("%.4f" % (g1b / math.sqrt(n_)), 12),
       padl("%.4f" % (Q_mat[1, j] - g1b / math.sqrt(n_)), 10))
P_("(2-qator: haqiqiy kino byudjetlari bosh toʻplami)")
P_("")
P_("XULOSA: «n >= 30» — qoida emas, qoʻpol tavsiya. Kerakli n bosh")
P_("toʻplamning qiyshiqligiga bogʻliq: D ~ gamma1/sqrt(n) tartibida.")
```

# 6-topshiriq. Ulush $\phat$ va muvaffaqiyat–muvaffaqiyatsizlik sharti {#t6}

> **Blum:** Qoʻllash · **Qiyinlik:** 3 (Amaliy) · **Ball:** 15 ·
> **Vaqt:** 15 daqiqa

**Statistik tushuncha.** Ulush — oʻrtachaning **maxsus holi**. Agar
$X_i\in\{0,1\}$ boʻlsa, $\phat=\Xbar$. Demak barcha natijalar avtomatik
koʻchadi:
$$\E[\phat]=p,\qquad \Var(\phat)=\frac{p(1-p)}{n},\qquad
  \SE(\phat)=\sqrt{\frac{p(1-p)}{n}}.$$

Ammo bitta farq bor: $\Bern(p)$ ning qiyshiqligi
$\gamma_1=\frac{1-2p}{\sqrt{p(1-p)}}$ — $p$ nolga yoki birga yaqin
boʻlsa **juda katta**. Shuning uchun ulush uchun MLT ni qoʻllashdan
oldin **shart tekshiriladi**:

> **Muvaffaqiyat–muvaffaqiyatsizlik sharti:** $np\ge10$ **va**
> $n(1-p)\ge10$.

**Maʼlumot.** 2 467 ta filmning reytingi. Bosh toʻplamdagi PG-13
ulushi — haqiqiy $p$.

**Kod.**

```python
# --- 6A: haqiqiy ulush -------------------------------------------------
# Reyting nomlarini FAYLDAN olamiz (literal yozmaymiz) — 2-hafta qoidasi.
# DIQQAT: ustunda 9 ta boʻsh katak (NaN) bor. Python da NaN bilan
# taqqoslash False beradi, R da esa NA beradi va butun hisob jimgina NA
# ga aylanadi. Ikki til bir xil ishlashi uchun boʻsh kataklarni OʻZIMIZ
# aniq boshqaramiz: ular «muvaffaqiyatsizlik» (0) deb sanaladi.
reytinglar = [r for r in kino["reyting"].unique() if isinstance(r, str)]
P_("")
P_("=== 6-TOPSHIRIQ (A): ULUSHNING TANLANMA TAQSIMOTI ===")
P_("Fayldagi reytinglar: %s", ", ".join(reytinglar))
P_("boʻsh (NA) kataklar: %d ta -> 0 deb sanaladi",
   int(kino["reyting"].isna().sum()))
# PG-13 — roʻyxatdagi qaysi pozitsiyada? Uni POZITSIYA boʻyicha olamiz.
poz  = 2                                # roʻyxatdagi ikkinchi toifa
pg13 = (kino["reyting"] == reytinglar[poz - 1]).to_numpy().astype(np.int64)
p0   = float(pg13.mean())
P_("tanlangan toifa: %s (roʻyxatda %d-oʻrin)", reytinglar[poz - 1], poz)
P_("p = %d/%d = %.6f     sqrt(p(1-p)) = %.6f",
   pg13.sum(), len(pg13), p0, math.sqrt(p0 * (1 - p0)))

P_("")
P_("%s%s%s%s%s%s%s", padl("n", 6), padl("E[phat]", 12), padl("p", 12),
   padl("SD(phat)", 12), padl("sqrt(p(1-p)/n)", 17), padl("np", 8),
   padl("n(1-p)", 9))
for n_ in [5, 10, 25, 50, 100, 400]:
    gl  = minstd(20260301 + n_)
    idx = indeks(gl, 20000 * n_, len(pg13))
    ph  = pg13[idx].reshape(20000, n_).mean(axis=1)
    P_("%s%s%s%s%s%s%s", padl("%d" % n_, 6),
       padl("%.5f" % ph.mean(), 12), padl("%.5f" % p0, 12),
       padl("%.5f" % ogish_bosh(ph), 12),
       padl("%.5f" % math.sqrt(p0 * (1 - p0) / n_), 17),
       padl("%.1f" % (n_ * p0), 8), padl("%.1f" % (n_ * (1 - p0)), 9))
P_("")
P_("Muvaffaqiyat-muvaffaqiyatsizlik sharti np >= 10 birinchi marta")
P_("n = %d da bajariladi (np = %.2f).",
   math.ceil(10 / p0), math.ceil(10 / p0) * p0)
P_("n = 25 da np = %.1f < 10 -> normal yaqinlashishga hali ishonib boʻlmaydi.",
   25 * p0)
```

# 7-topshiriq. Koshi: MLT ishlamaydigan hol {#t7}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 15 ·
> **Vaqt:** 15 daqiqa

**Statistik tushuncha.** MLT ning shartlari uchta:
(i) bogʻliqsizlik, (ii) bir xil taqsimot, (iii) **chekli dispersiya**
$0<\sigma^{2}<\infty$. Birinchi ikkitasi odatda muhokama qilinadi,
uchinchisi esa deyarli hech qachon. Bugun biz uni **buzamiz**.

> **Koshi taqsimoti.** Zichligi $f(x)=\dfrac{1}{\pi(1+x^{2})}$.
> Dumlari shu qadar ogʻirki, $\E[X]$ **mavjud emas**, $\Var(X)$ ham
> **mavjud emas**. Nazariy natija: $n$ ta Koshi kuzatuvining oʻrtachasi
> — **yana aynan oʻsha Koshi taqsimoti**. Yaʼni $n$ ni oshirish
> **hech narsani** yaxshilamaydi.

**Bu topshiriqning maqsadi — kodni ishlatib, uning ishlamasligini
koʻrish.** 3-topshiriqdagi ayni oʻsha sxema, faqat bosh toʻplam boshqa.

Koshi sonini teng taqsimlangan $u$ dan olish:
$X=\tan\!\bigl(\pi(u-\tfrac12)\bigr)$ (teskari funksiya usuli, 5-hafta).

**Kod.**

```python
# --- 7A: 3-topshiriqdagi AYNI sxema, faqat Koshi bosh toʻplamida -------
# DIQQAT: oʻrtacha va standart ogʻish oʻrniga MEDIAN va IQR ishlatamiz.
# Sabab: Koshi da E[X] mavjud emas, shuning uchun mean() ning oʻzi
# maʼnosiz — u hech narsaga yaqinlashmaydi.
P_("")
P_("=== 7-TOPSHIRIQ: KOSHI TAQSIMOTI — MLT SHARTI BUZILGANDA ===")
P_("Nazariy: standart Koshi taqsimotining IQR = 2 (AYNAN).")
P_("%s%s%s%s%s", padl("n", 8), padl("median(Xbar)", 15), padl("IQR(Xbar)", 13),
   padl("nazariy IQR", 14), padl("P(|Xbar|>10)", 15))
koshi_iqr = []; koshi_n = []
for n_ in [1, 5, 30, 100, 1000]:
    gl = minstd(20260401 + n_)
    u_ = gl(20000 * n_)
    C_ = np.tan(math.pi * (u_ - 0.5)).reshape(20000, n_)
    xb = C_.mean(axis=1)
    iq = kv(xb, 0.75) - kv(xb, 0.25)
    koshi_iqr.append(iq); koshi_n.append(n_)
    P_("%s%s%s%s%s", padl("%d" % n_, 8),
       padl("%.4f" % np.median(xb), 15), padl("%.4f" % iq, 13),
       padl("%.1f" % 2.0, 14),
       padl("%.5f" % (np.abs(xb) > 10).mean(), 15))
    del u_, C_, xb
P_("")
P_("IQR n oshgani sayin KICHRAYMAYDI — u 2 atrofida qotib qoladi.")
P_("P(|Xbar| > 10) ham kamaymaydi: n = 1000 da ham ~6.7 %%.")
```

```python
# --- 7B: solishtirish uchun NORMAL bosh toʻplam ------------------------
# Bir xil kod, bir xil m; farq faqat bosh toʻplamda.
# Normal sonlar: Boks-Myuller (5-hafta), MINSTD ustida.
def boks_myuller(gen, n):
    m2 = n + (n % 2)
    us = gen(m2)
    u1 = us[0::2]; u2 = us[1::2]
    r  = np.sqrt(-2 * np.log(u1)); th = 2 * math.pi * u2
    return np.concatenate([r * np.cos(th), r * np.sin(th)])[:n]

P_("")
P_("Solishtirish: NORMAL bosh toʻplam (nazariy IQR(Xbar) = 1.349/sqrt(n))")
P_("%s%s%s%s", padl("n", 8), padl("IQR(Xbar)", 14), padl("1.349/sqrt(n)", 16),
   padl("Koshi IQR", 13))
norm_iqr = []
for k, n_ in enumerate(koshi_n):
    gl  = minstd(20260501 + n_)
    Z_  = boks_myuller(gl, 20000 * n_).reshape(20000, n_)
    zb  = Z_.mean(axis=1)
    iqn = kv(zb, 0.75) - kv(zb, 0.25)
    norm_iqr.append(iqn)
    P_("%s%s%s%s", padl("%d" % n_, 8), padl("%.4f" % iqn, 14),
       padl("%.4f" % (1.3490 / math.sqrt(n_)), 16),
       padl("%.4f" % koshi_iqr[k], 13))
    del Z_, zb
P_("")
P_("Normal ustunda IQR 1/sqrt(n) tezligida kichrayadi; Koshi ustunida —")
P_("umuman kichraymaydi. MLT ning uchinchi sharti (chekli dispersiya)")
P_("aynan shu farqni tugʻdiradi.")
```

```python
# --- 7C: ikkita taqsimotning «torayishi» yonma-yon ---------------------
fig, ax = plt.subplots()
ax.plot(koshi_n, koshi_iqr, "x-", color=oiVerm, ms=8, linewidth=1.4,
        label="Cauchy (no variance): IQR stays at 2")
ax.plot(koshi_n, norm_iqr, "o-", color=oiBlue, ms=6, linewidth=1.4,
        label="Normal: IQR = 1.349/sqrt(n)")
ax.set_xscale("log"); ax.set_yscale("log"); ax.set_xticks(koshi_n)
ax.get_xaxis().set_major_formatter(matplotlib.ticker.ScalarFormatter())
ax.set_xlabel("n (log scale)"); ax.set_ylabel("IQR of Xbar (log scale)")
ax.set_title("Where the CLT fails: the spread of Xbar refuses to shrink")
ax.legend(frameon=False, fontsize=9, loc="lower left")
fig.text(0.99, -0.02, "Same code, same m = 20000, only the population changed"
         "  |  MINSTD, seeds 20260401+n / 20260501+n",
         ha="right", fontsize=8, color=oiGrey)
plt.show()
```

**Talqin (diagnostika — bu topshiriqning asosiy qismi).**

1. Kod **xato bermadi**, `NaN` chiqarmadi, ogohlantirish yozmadi.
   Ammo natija **notoʻgʻri**. MLT ning qaysi sharti buzildi va
   **qaysi qatorda** buzilishi kerak edi?
2. Nima uchun bu topshiriqda `mean()` va `sd()` emas, `median()` va
   `IQR` ishlatildi? Agar `sd(xb)` ni hisoblasangiz nima chiqadi va
   uni takrorlasangiz nima oʻzgaradi? **Sinab koʻring.**
3. $\SE=\sigma/\sqrt{n}$ formulasini Koshi uchun yozib boʻladimi?
   $\sigma$ ni nimaga tenglashtirasiz?
4. **Iqtisodiyotda qayerda uchraydi?** Moliyaviy daromadlar, sugʻurta
   zararlari, shahar aholisi, firma hajmi — «ogʻir dumli» kattaliklar.
   Ulardan **bittasini** tanlang va $n$ ni oshirish nima uchun
   yetarli boʻlmasligini bir xatboshida yozing.

**Qaror.** Analitik: «Men $n=1000$ ta kuzatuv oldim, MLT boʻyicha
$\Xbar$ normal taqsimlangan». Uning **birinchi** tekshirishi kerak
boʻlgan narsa nima? Bir jumla.

# Refleksiya savollari

Har biriga **3–4 jumla** bilan yozma javob bering.

1. **Taʼrifdan tajribaga.** Maʼruzada «tanlanma taqsimoti» taʼrifini
   eshitdingiz. 2-topshiriqda uni **oʻzingiz sanab chiqdingiz**.
   Taʼrifni tushunishingiz oʻzgardimi? Qaysi qadam eng koʻp yordam berdi?
2. **Ikki ustun.** $|S_n-n/2|$ oʻsib, $|\phat-0{,}5|$ kamayishi sizga
   avval qanday tuyulgan edi? Endi buni bir jumlada qanday
   tushuntirasiz?
3. **Qoidaning narxi.** «$n\ge30$» ni ilgari qoida deb bilganmisiz?
   $0{,}1462$ va $0{,}0109$ sonlari sizning amaliyotingizni qanday
   oʻzgartiradi?
4. **Ishlamagan kod.** 7-topshiriqda kod xato bermadi, ammo javob
   notoʻgʻri edi. Kelajakdagi ishingizda **qanday tekshiruv** shu
   turdagi jimgina xatolarni tutadi?
5. **Tasodifiy xato va siljish.** $n$ ni oshirish $\SE$ ni kamaytiradi.
   U **nimani kamaytirmaydi**? Oʻz sohangizdan bitta misol keltiring.
6. **Ikki til.** Sonlaringiz boshqa tildagi kursdoshingiznikiga mos
   tushdimi? Mos tushmagan boʻlsa, farq **qaysi qadamda** paydo boʻldi?
   Bu **takrorlanuvchanlik** haqida nima oʻrgatdi?
