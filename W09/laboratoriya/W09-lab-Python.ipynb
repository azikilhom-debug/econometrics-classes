```python
# --- Kerakli kutubxonalar ---------------------------------------------
import math
import numpy as np                  # sonli massivlar
import pandas as pd                 # jadval koʻrinishidagi maʼlumot
import matplotlib
import matplotlib.pyplot as plt     # grafiklar
from scipy import stats             # qt / dbinom ekvivalentlari

# Okabe–Ito palitrasi: rang koʻrmaslikka moslashgan.
oiBlue = "#0072B2"; oiOrange = "#E69F00"; oiGreen = "#009E73"
oiPink = "#CC79A7"; oiVerm   = "#D55E00"; oiSky   = "#56B4E9"
oiYell = "#F0E442"; oiGrey   = "#5A6470"

plt.rcParams.update({
    "figure.dpi": 110, "figure.figsize": (7, 4.2), "font.size": 10,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.alpha": 0.25, "axes.titleweight": "bold"})

# QOIDA (2-haftadan): grafik yorliqlarida faqat ASCII harflar —
# shunda daftar har qanday shrift sozlamasida bir xil koʻrinadi.
print("MUHIT: matplotlib", matplotlib.__version__)
```

```python
# --- IKKI DAFTAR BIR XIL MATN CHIQARISHI UCHUN YORDAMCHILAR ------------
# (1) ha(): R «TRUE», Python esa «True» yozadi. Mantiqiy qiymatni
#     OʻZIMIZ formatlaymiz.
def ha(b):
    return "HA" if bool(b) else "YOʻQ"

# (2) pad()/padl(): R ning sprintf("%-16s") formati satrni BAYT boʻyicha
#     toʻldiradi, Python esa BELGI boʻyicha. «YOʻQ» da 4 ta belgi, ammo
#     5 ta bayt bor — shu sababli oʻzbekcha yorliqlarni formatga bermay,
#     kenglikni OʻZIMIZ hisoblaymiz.
def pad(s, n):                      # chapga tekislash (oʻngdan boʻshliq)
    return s + " " * max(0, n - len(s))

def padl(s, n):                     # oʻngga tekislash (chapdan boʻshliq)
    return " " * max(0, n - len(s)) + s

# (3) P_(): R dagi sprintf + cat ekvivalenti. Har bir son QOʻLDA
#     formatlanadi, tilning avtomatik chiqarish mexanizmiga tayanmaymiz.
def P_(fmt, *a):
    print(fmt % a)

# (4) Qator boʻyicha koʻrsatkichlar. Matritsaning HAR BIR QATORI —
#     bitta tanlanma.
def qatorOrtacha(M):
    return M.mean(axis=1)

def qatorVarS(M):                   # maxraj n-1 (STATISTIKA)
    return M.var(axis=1, ddof=1)

def qatorVarP(M):                   # maxraj n   (s^2_n)
    return M.var(axis=1, ddof=0)

def qatorMedian(M):
    return np.median(M, axis=1)

# (5) Bosh toʻplam koʻrsatkichlari — MAXRAJI N (n-1 EMAS!).
def dispersiya_bosh(x):
    x = np.asarray(x, dtype=float)
    return float(np.mean((x - x.mean()) ** 2))

def ogish_bosh(x):
    return math.sqrt(dispersiya_bosh(x))

def qiyshiqlik(x):
    x = np.asarray(x, dtype=float); d = x - x.mean()
    m2 = np.mean(d ** 2); m3 = np.mean(d ** 3)
    return float(m3 / m2 ** 1.5)

# (6) Kvantil: numpy ning standart ("linear") usuli R ning type=7 usuli
#     bilan AYNAN bir xil. Boshqa usul ISHLATMANG.
def kv(x, p):
    return float(np.quantile(np.asarray(x, dtype=float), p, method="linear"))

# (7) Kritik qiymatlar. z ni LITERAL yozamiz (ikki tilda bir xil bit).
Z975 = 1.959964

def tkr(daraja, df):
    return float(stats.t.ppf(daraja, df))

# R ning sd()/var() maxraji n-1; Python da buni OSHKORA yozamiz.
def sd_(x):
    return float(np.std(np.asarray(x, dtype=float), ddof=1))

def var_(x):
    return float(np.var(np.asarray(x, dtype=float), ddof=1))
```

```python
# --- MINSTD generatori (2-haftadan beri oʻzgarmagan) -------------------
# Nima uchun oʻzimizning generatorimiz kerak?
#   R:      set.seed(2026); runif(3)
#   Python: np.random.default_rng(2026).random(3)
# Bu ikkisi BOSHQA-BOSHQA sonlar beradi. Ikkalasi ham toʻgʻri, lekin
# ikkalasi ham bir-biriga mos emas. Shu sababli kursda tasodifiylik
# kerak boʻlganda HAR DOIM quyidagi 4 qatorli generator ishlatiladi:
#     x <- (48271 * x) mod 2147483647,    u <- x / 2147483647
# Bu — Lehmer (1951) generatori, «MINSTD» nomi bilan mashhur.
A_ = 48271; M_ = 2147483647

def minstd(urug):
    holat = {"x": urug % M_}
    def gen(n):                        # n ta u ~ U(0,1) qaytaradi
        n = int(n); o = np.empty(n); x = holat["x"]
        for i in range(n):
            x = (A_ * x) % M_; o[i] = x
        holat["x"] = x
        return o / M_
    return gen

# Indeks tanlash: 0 dan hi-1 gacha butun son (R bilan bir xil).
# Python da vektor 0 dan boshlanadi, shuning uchun +1 KERAK EMAS.
def indeks(gen, n, hi):
    return (gen(n) * hi).astype(np.int64)

# Box-Muller: tekis sonlardan normal sonlar. IKKI TILDA BIR XIL TARTIB:
# avval barcha kosinuslar, keyin barcha sinuslar.
def minstd_norm(gen, n):
    n = int(n); m2 = n + (n % 2)
    us = gen(m2).reshape(-1, 2)
    r = np.sqrt(-2 * np.log(us[:, 0])); th = 2 * np.pi * us[:, 1]
    return np.concatenate([r * np.cos(th), r * np.sin(th)])[:n]

# Nazorat qiymati: urugʻ 2026 dan keyingi uchta son.
g = minstd(2026); u3 = g(3)
P_("=== 0-BOSQICH: MINSTD NAZORATI ===")
P_("urugʻ 2026 -> u1 = %.10f  u2 = %.10f  u3 = %.10f", u3[0], u3[1], u3[2])
P_("bu uchta son ikkinchi tildagi daftarda ham AYNAN shunday chiqishi shart")

# Solishtirish uchun: tilning OʻZ generatori (bu satr ikki tilda FARQ QILADI)
print("MUHIT: Python ning default_rng(2026).random(1) = %.10f  "
      "(R da boshqa son chiqadi)"
      % np.random.default_rng(2026).random(1)[0])
```

```python
# --- Ikkita haqiqiy maʼlumot toʻplami ---------------------------------
# A. kinolar-byudjet.csv — 2 467 film (2000–2018), byudjet AQSh
#    dollarida. Manba: The Numbers / Kaggle «movie industry».
#    BUGUN BU FAYL — BUTUN BOSH TOʻPLAM.
# B. markaziy-osiyo-hosildorlik.csv — 5 mamlakat x 27 yil (1992–2018),
#    hosildorlik t/ga. Manba: FAOSTAT / Our World in Data (OWID).
DATA = "data"
kino = pd.read_csv(DATA + "/kinolar-byudjet.csv", encoding="utf-8")
hosil = pd.read_csv(DATA + "/markaziy-osiyo-hosildorlik.csv", encoding="utf-8")

b = kino["byudjet"].to_numpy(dtype=float) / 1e6      # mln dollarda
Nb = len(b)
MU = float(b.mean())                                 # PARAMETR
SIG = ogish_bosh(b)                                  # PARAMETR (maxraj N)
SIG2 = SIG ** 2
MEDB = float(np.median(b))

P_("")
P_("=== 0-BOSQICH: BOSH TOʻPLAM (BUGUN U BIZGA MAʼLUM) ===")
P_("manba: The Numbers / Kaggle, 2000-2018; birlik: mln $")
P_("N        = %d", Nb)
P_("mu       = %.6f   <- PARAMETR, qatʼiy son", MU)
P_("sigma    = %.6f   <- PARAMETR (maxraj N)", SIG)
P_("sigma^2  = %.4f", SIG2)
P_("median   = %.4f", MEDB)
P_("min; max = %.2f; %.2f", b.min(), b.max())
P_("qiyshiqlik gamma1 = %.6f  -> kuchli OʻNGGA qiyshiq", qiyshiqlik(b))
P_("")
P_("DIQQAT: mu (%.2f) va median (%.2f) — IKKI BOSHQA parametr.", MU, MEDB)
P_("Ular bir-birining oʻrnini bosmaydi.")
P_("hosildorlik fayli: %d qator, %d-%d yillar",
   len(hosil), int(hosil["yil"].min()), int(hosil["yil"].max()))
```

# 1-topshiriq. Siljimaganlik — oʻlchangan, aytilmagan {#t1}

> **Blum:** Qoʻllash · **Qiyinlik:** 2 (Konseptual) · **Ball:** 14 ·
> **Vaqt:** 11 daqiqa

**Statistik tushuncha.** **Baholovchi** — bu *formula* (tasodifiy
miqdor, taqsimoti bor). **Baho** — bu *son* (bitta tanlanmadan chiqqan
natija). Xossalar **formulaga** tegishli, songa emas:

$$\text{siljish}(\hat\theta)=\E[\hat\theta]-\theta,
\qquad \hat\theta \text{ siljimagan} \iff \E[\hat\theta]=\theta .$$

$\E[\hat\theta]$ ni biz bilmaymiz — uni **20 000 ta tanlanma boʻyicha
oʻrtacha** bilan baholaymiz. Demak bizning «$\E[\hat\theta]$» imiz ham
tasodifiy: uning oʻz standart xatosi bor, uni **Monte-Karlo xatosi**
deymiz:

$$\text{MK xatosi} \approx 1{,}96\cdot\frac{\SD(\hat\theta)}{\sqrt{m}},
\qquad m = 20\,000 .$$

> **Qoida.** «Siljish nolga yaqin» — bu **tekshiriladigan daʼvo**, umid
> emas. Agar $|\text{siljish}| < \text{MK xatosi}$ boʻlsa, kuzatilgan
> farq shovqin bilan izohlanadi. Aks holda siljish **haqiqiy**.

**Maʼlumot.** 2 467 ta film byudjeti — butun bosh toʻplam. Undan
qaytarish bilan $m=20\,000$ ta tanlanma olamiz, $n\in\{5,10,30\}$.

**Kod.**

```python
# --- 1-topshiriq: toʻrtta baholovchini bosh toʻplam bilan solishtirish -
m = 20000

# Bitta baholovchi uchun bitta qator chiqaradigan yordamchi.
# v — 20 000 ta baho; nishon — haqiqiy parametr qiymati.
def bahol_qator(n_, nom, v, nishon):
    ort = float(np.mean(v)); silj = ort - nishon
    # Monte-Karlo xatosi: 20 000 ta tanlanma boʻyicha oʻrtachaning oʻzi
    # ham tasodifiy. 1,96 ATAYLAB yaxlitlangan holda ishlatiladi.
    mk = 1.96 * sd_(v) / math.sqrt(len(v))
    P_("%s%s%s%s%s%s%s",
       padl("%d" % n_, 4), padl(nom, 11),
       padl("%.4f" % ort, 13), padl("%.4f" % nishon, 12),
       padl("%+.4f" % silj, 12), padl("+-%.4f" % mk, 13),
       padl(ha(abs(silj) < mk), 16))

P_("=== 1-TOPSHIRIQ: SILJIMAGANLIK ===")
P_("m = %d ta tanlanma; generator MINSTD, urugʻ 20260901+n", m)
P_("nishon: mu = %.4f, sigma^2 = %.4f, bosh toʻplam medianasi = %.4f",
   MU, SIG2, MEDB)
P_("")
P_("%s%s%s%s%s%s%s", padl("n", 4), padl("baholovchi", 11),
   padl("E[baho]", 13), padl("nishon", 12), padl("siljish", 12),
   padl("MK xatosi", 13), padl("nol bilan mos?", 16))
for n_ in [5, 10, 30]:
    g = minstd(20260901 + n_)
    S = b[indeks(g, m * n_, Nb)].reshape(m, n_)
    bahol_qator(n_, "xbar",   qatorOrtacha(S), MU)
    bahol_qator(n_, "s^2",    qatorVarS(S),    SIG2)
    bahol_qator(n_, "s^2_n",  qatorVarP(S),    SIG2)
    bahol_qator(n_, "median", qatorMedian(S),  MEDB)
    del S
P_("")
P_("s^2_n ning siljishi TASODIFIY EMAS — u nazariy -sigma^2/n ga teng:")
for n_ in [5, 10, 30]:
    P_("  n=%3d:  nazariy -sigma^2/n = %10.4f", n_, -SIG2 / n_)
P_("")
P_("Nima uchun? E[s^2_n] = (n-1)/n * sigma^2, demak siljish = -sigma^2/n.")
P_("Shu sababli 1-haftadagi n-1 maxraji SILJISHNI YOʻQOTISH uchun kerak.")
```

**Diqqat — bu topshiriqning eng muhim qadami.** Jadvalda $n=10$ satrida
$\Xbar$ va $s^{2}$ uchun ham «nol bilan mos?» ustunida **YOʻQ**
chiqadi. Bu **siljish borligini isbotlamaydi.** $95\,\%$ chegara har
$20$ sinovdan taxminan bittasida tabiiy ravishda buziladi; bundan
tashqari $\Xbar$ va $s^{2}$ **bir xil** 20 000 ta tanlanmadan
hisoblangan, shuning uchun ular birga «yiqiladi». Yagona toʻgʻri yoʻl —
**takrorlash**.

```python
# --- 1B: bitta 'YOʻQ' ni tekshirish — beshta boshqa urugʻ -------------
# Agar siljish HAQIQIY boʻlsa, u har bir urugʻda BIR XIL ISHORA bilan
# va bir xil kattalikda takrorlanadi (s^2_n dagi kabi). Agar u shovqin
# boʻlsa, ishorasi tasodifiy oʻzgaradi va oʻrtachasi nolga intiladi.
P_("")
P_("--- 1B: n=10 ni beshta boshqa urugʻ bilan takrorlaymiz ---")
P_("%s%s%s%s%s", padl("urugʻ", 12), padl("E[xbar]", 12),
   padl("siljish", 12), padl("MK xatosi", 13), padl("nol bilan mos?", 16))
siljlar = []
for u in [20260911, 20261911, 20262911, 20263911, 20264911]:
    g = minstd(u)
    S = b[indeks(g, m * 10, Nb)].reshape(m, 10)
    xb = qatorOrtacha(S); silj = float(xb.mean()) - MU
    mk = 1.96 * sd_(xb) / math.sqrt(m)
    siljlar.append(silj)
    P_("%s%s%s%s%s", padl("%d" % u, 12),
       padl("%.4f" % xb.mean(), 12), padl("%+.4f" % silj, 12),
       padl("+-%.4f" % mk, 13), padl(ha(abs(silj) < mk), 16))
    del S, xb
siljlar = np.array(siljlar)
P_("")
P_("beshta siljishning oʻrtachasi = %+.4f", siljlar.mean())
P_("musbat: %d ta, manfiy: %d ta -> ishora TASODIFIY",
   int((siljlar > 0).sum()), int((siljlar < 0).sum()))
P_("Solishtiring: s^2_n ning siljishi HAR DOIM manfiy va HAR DOIM")
P_("-sigma^2/n ga teng. MANA SHU — haqiqiy siljish. Yuqoridagi 'YOʻQ'")
P_("esa Monte-Karlo shovqini edi.")
P_("")
P_("QOIDA: bitta simulyatsiyaga tayanib 'siljish bor' deb xulosa")
P_("chiqarmang. m ni oshiring yoki urugʻni oʻzgartirib takrorlang.")
```

**Natija va talqin.** Yozma javob bering:

# 2-topshiriq. Samaradorlik: oʻrtacha yoki median? {#t2}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 14 ·
> **Vaqt:** 11 daqiqa

**Statistik tushuncha.** Ikkala baholovchi ham siljimagan boʻlsa, tanlov
**tarqoqlik** boʻyicha qilinadi. **Samaradorlik** —
$\Var(\hat\theta_1)/\Var(\hat\theta_2)$ nisbati. Simmetrik normal bosh
toʻplamda nazariya aniq javob beradi:

$$\frac{\Var(\text{median})}{\Var(\Xbar)}\ \longrightarrow\ \frac{\pi}{2}=1{,}5708 .$$

Yaʼni normal bosh toʻplamda median **$57\,\%$ koʻproq tarqaladi** —
oʻrtacha yutadi. Ammo bu **bosh toʻplam shakliga** bogʻliq. Ogʻir dumli
$t_3$ taqsimotida nima boʻlishini **oldindan yozib qoʻying**, keyin
tekshiring.

**Maʼlumot.** Ikkita *simulyatsiya qilingan* bosh toʻplam (haqiqiy emas
— generator va urugʻ koʻrsatiladi): (a) $\Norm(0,1)$;
(b) $t_3$ — dispersiyasi mavjud, ammo dumlari ogʻir.

**Kod.**

```python
# --- 2-topshiriq: bir xil kod, ikki xil bosh toʻplam -------------------
P_("=== 2-TOPSHIRIQ: SAMARADORLIK ===")
P_("m = %d ta tanlanma; MINSTD, urugʻ 20260950+n va 20260980+n", m)
P_("")
P_("(a) NORMAL bosh toʻplam N(0,1) — nazariy nisbat pi/2 = %.6f",
   math.pi / 2)
P_("%s%s%s%s%s", padl("n", 5), padl("Var(xbar)", 14),
   padl("Var(median)", 14), padl("nisbat", 10), padl("nazariy", 10))
nis_norm = []
for n_ in [5, 11, 31, 101]:
    g = minstd(20260950 + n_)
    Zm = minstd_norm(g, m * n_).reshape(m, n_)
    vx = var_(qatorOrtacha(Zm)); vm = var_(qatorMedian(Zm))
    nis_norm.append(vm / vx)
    P_("%s%s%s%s%s", padl("%d" % n_, 5),
       padl("%.6f" % vx, 14), padl("%.6f" % vm, 14),
       padl("%.4f" % (vm / vx), 10), padl("%.4f" % (math.pi / 2), 10))
    del Zm
P_("nisbat > 1 -> OʻRTACHA yutadi.")
P_("")
P_("(b) OGʻIR DUMLI bosh toʻplam t_3 (dispersiyasi bor, dumlari ogʻir)")
P_("%s%s%s%s", padl("n", 5), padl("Var(xbar)", 14),
   padl("Var(median)", 14), padl("nisbat", 10))
nis_t3 = []
for n_ in [5, 11, 31, 101]:
    g = minstd(20260980 + n_)
    U = g(m * n_)
    T3 = stats.t.ppf(np.clip(U, 1e-12, 1 - 1e-12), df=3).reshape(m, n_)
    vx = var_(qatorOrtacha(T3)); vm = var_(qatorMedian(T3))
    nis_t3.append(vm / vx)
    P_("%s%s%s%s", padl("%d" % n_, 5),
       padl("%.6f" % vx, 14), padl("%.6f" % vm, 14),
       padl("%.4f" % (vm / vx), 10))
    del U, T3
P_("nisbat < 1 -> MEDIAN yutadi. TARTIB TESKARISIGA AYLANDI.")
P_("")
P_("XULOSA: 'eng yaxshi baholovchi' degan narsa YOʻQ. Javob bosh")
P_("toʻplam SHAKLIGA bogʻliq — bu 5-haftadagi barqarorlik gʻoyasi.")
```

```python
# --- Grafik: nisbat n boʻyicha, ikki bosh toʻplam uchun ----------------
n_lar = [5, 11, 31, 101]
fig, ax = plt.subplots()
ax.axhline(1.0, ls=":", color=oiGrey)
ax.axhline(math.pi / 2, ls="--", color=oiBlue)
ax.plot(n_lar, nis_norm, "-o", color=oiBlue, label="Normal N(0,1)")
ax.plot(n_lar, nis_t3, "-s", color=oiVerm, label="Heavy-tailed t_3")
ax.text(70, math.pi / 2 + 0.05, "pi/2 = 1.5708", color=oiBlue, fontsize=8)
ax.text(70, 1.06, "equal variance", color=oiGrey, fontsize=8)
ax.set_xticks(n_lar); ax.set_xticklabels([str(v) for v in n_lar])
ax.set_xlabel("sample size n"); ax.set_ylabel("Var(median) / Var(mean)")
ax.set_title("Var(median) / Var(mean): the ranking reverses")
ax.legend(frameon=False, ncol=2, loc="lower right", fontsize=9)
fig.suptitle("Same code, same m = 20000; only the population changed",
             y=0.94, fontsize=9, color="grey")
fig.text(0.99, -0.02, "MINSTD, seeds 20260950+n and 20260980+n",
         ha="right", fontsize=8, color=oiGrey)
plt.show()
```

# 3-topshiriq. MSE: siljigan, ammo aniqroq {#t3}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 14 ·
> **Vaqt:** 11 daqiqa

**Statistik tushuncha.** Siljish va tarqoqlikni bitta koʻrsatkichga
birlashtiramiz:

$$\text{MSE}(\hat\theta)=\E\bigl[(\hat\theta-\theta)^{2}\bigr]
=\text{siljish}(\hat\theta)^{2}+\Var(\hat\theta).$$

Normal bosh toʻplamda $\sigma^{2}=1$ ni baholash uchun nazariya aniq
javob beradi:

$$\text{MSE}(s^{2})=\frac{2}{n-1},
\qquad \text{MSE}(s^{2}_{n})=\frac{2n-1}{n^{2}} .$$

Har bir $n\ge2$ uchun ikkinchisi **kichikroq**. Yaʼni **siljimagan
baholovchi eng aniq baholovchi emas.**

**Maʼlumot.** $\Norm(0,1)$ dan $m=20\,000$ ta tanlanma,
$n\in\{3,5,10,30\}$; nishon $\sigma^{2}=1$.

**Kod.**

```python
# --- 3-topshiriq: MSE ni ikki yoʻl bilan hisoblash ---------------------
P_("=== 3-TOPSHIRIQ: MSE = SILJISH^2 + DISPERSIYA ===")
P_("Normal bosh toʻplam, nishon sigma^2 = 1; MINSTD, urugʻ 20261000+n")
P_("")
P_("%s%s%s%s%s%s%s", padl("n", 4), padl("silj(s^2)", 12),
   padl("Var(s^2)", 11), padl("MSE(s^2)", 11), padl("silj(s^2_n)", 13),
   padl("Var(s^2_n)", 12), padl("MSE(s^2_n)", 12))
mse_kuz = np.zeros((4, 2)); i_ = -1
for n_ in [3, 5, 10, 30]:
    i_ += 1
    g = minstd(20261000 + n_)
    Zm = minstd_norm(g, m * n_).reshape(m, n_)
    a = qatorVarS(Zm)            # s^2   (maxraj n-1)
    cc = qatorVarP(Zm)           # s^2_n (maxraj n)
    mseA = float(((a - 1) ** 2).mean()); mseC = float(((cc - 1) ** 2).mean())
    mse_kuz[i_, :] = [mseA, mseC]
    P_("%s%s%s%s%s%s%s", padl("%d" % n_, 4),
       padl("%.5f" % (a.mean() - 1), 12),
       padl("%.5f" % dispersiya_bosh(a), 11),
       padl("%.5f" % mseA, 11),
       padl("%.5f" % (cc.mean() - 1), 13),
       padl("%.5f" % dispersiya_bosh(cc), 12),
       padl("%.5f" % mseC, 12))
    del Zm, a, cc
P_("")
P_("Endi NAZARIY qiymatlar bilan solishtiramiz:")
P_("%s%s%s%s%s", padl("n", 4), padl("MSE(s^2) kuz.", 15),
   padl("2/(n-1)", 11), padl("MSE(s^2_n) kuz.", 17), padl("(2n-1)/n^2", 13))
i_ = -1
for n_ in [3, 5, 10, 30]:
    i_ += 1
    P_("%s%s%s%s%s", padl("%d" % n_, 4),
       padl("%.5f" % mse_kuz[i_, 0], 15),
       padl("%.5f" % (2 / (n_ - 1)), 11),
       padl("%.5f" % mse_kuz[i_, 1], 17),
       padl("%.5f" % ((2 * n_ - 1) / n_ ** 2), 13))
P_("")
P_("Har bir n da MSE(s^2_n) < MSE(s^2). SILJIMAGAN != ENG YAXSHI.")
P_("Nisbat n oshgani sayin 1 ga intiladi -> farq kichik n da muhim.")
```

**Talqin.** Yozma javob bering:

1. Kuzatilgan MSE lar nazariy qiymatlarga qanchalik yaqin? Farqni
   **Monte-Karlo xatosi** bilan izohlash mumkinmi?
2. $n=3$ da MSE nisbati qancha, $n=30$ da-chi? Qaysi $n$ dan boshlab
   farq amalda ahamiyatsiz boʻladi?
3. Nima uchun **kurs va butun statistika amaliyoti** shunga qaramay
   $s^{2}$ ni ($n-1$ bilan) ishlatadi? Uchta sabab keltiring.
4. $\text{MSE}=\text{siljish}^{2}+\Var$ ayniyati qaysi topshiriqdagi
   qaysi ikkita ustunni bogʻlaydi? Jadvaldan sonlar bilan koʻrsating.

**Qaror.** Sizning tashkilotingiz $\sigma^{2}$ ni baholaydi va xato

# 4-topshiriq. $t$-statistikaning tanlanma taqsimoti {#t4}

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 14 ·
> **Vaqt:** 12 daqiqa

**Statistik tushuncha.** 7-haftada quyidagi natijadan foydalandik:

$$Z=\frac{\Xbar-\mu}{\sigma/\sqrt{n}}\ \sim\ \Norm(0,1)
\qquad(\sigma \text{ MAʼLUM}).$$

Endi $\sigma$ nomaʼlum va uni $s$ bilan almashtiramiz:

$$T=\frac{\Xbar-\mu}{s/\sqrt{n}} .$$

**Maxraj endi tasodifiy.** Ikkita tasodifiy miqdorning nisbati
$\Norm(0,1)$ boʻlishi shart emas — va u emas ham. $T$ ning taqsimoti —
**$n-1$ erkinlik darajali $t$-taqsimot**; uning nazariy standart
ogʻishi $\sqrt{\mathrm{df}/(\mathrm{df}-2)}$ ($\mathrm{df}>2$ da).

> **Diqqat.** $s$ oʻrtacha hisobda $\sigma$ ga teng emas: kichik $n$ da
> u tez-tez $\sigma$ dan **kichik** chiqadi, va aynan shunday
> hollarda $T$ kattalashadi. Shu sababli dumlar ogʻirlashadi.

**Kod.**

```python
# --- 4-topshiriq: T ning taqsimotini QURAMIZ ---------------------------
mm = 40000
P_("=== 4-TOPSHIRIQ: NIMA UCHUN t KERAK ===")
P_("Normal bosh toʻplam N(0,1); mm = %d ta tanlanma; urugʻ 20261100+n", mm)
P_("")
P_("%s%s%s%s%s%s", padl("n", 5), padl("df", 5), padl("SD(z-stat)", 13),
   padl("SD(t-stat)", 13), padl("nazariy sqrt(df/(df-2))", 26),
   padl("P(|T|>1,96)", 14))
T5 = None
for n_ in [3, 5, 10, 30, 100]:
    g = minstd(20261100 + n_)
    Zm = minstd_norm(g, mm * n_).reshape(mm, n_)
    xb = qatorOrtacha(Zm); s = np.sqrt(qatorVarS(Zm))
    tstat = xb / (s / math.sqrt(n_))          # sigma NOMAʼLUM -> s
    zstat = xb / (1 / math.sqrt(n_))          # sigma MAʼLUM   -> 1
    # Oxirgi ustunda ATAYLAB yaxlitlangan 1,96 ishlatiladi — chunki savol
    # aynan shu: amaliyotda hamma yozadigan 1,96 qanchalik xato qiladi?
    if n_ == 5:
        T5 = tstat
    df = n_ - 1
    naz = ("%.4f" % math.sqrt(df / (df - 2))) if df > 2 else "inf"
    P_("%s%s%s%s%s%s", padl("%d" % n_, 5), padl("%d" % df, 5),
       padl("%.4f" % sd_(zstat), 13),
       padl("%.4f" % sd_(tstat), 13),
       padl(naz, 26), padl("%.5f" % np.mean(np.abs(tstat) > 1.96), 14))
    del Zm
P_("")
P_("SD(z) har doim 1 — sigma maʼlum boʻlgani uchun.")
P_("SD(t) har doim KATTAROQ va n bilan 1 ga tushadi.")
P_("P(|T|>1,96) 0,05 dan KATTA -> 1,96 ni ishlatish oraliqni JUDA TOR qiladi.")
P_("")
P_("--- t ning kritik qiymatlari (95%% daraja, ikki tomonlama) ---")
P_("%s%s%s%s", padl("df", 6), padl("t_0,975", 10),
   padl("z=1,96 dan farq", 18), padl("ortiqcha kenglik %", 20))
for df in [1, 2, 3, 4, 5, 9, 10, 14, 19, 24, 29, 39, 59, 99, 199, 999]:
    tc = tkr(0.975, df)
    P_("%s%s%s%s", padl("%d" % df, 6), padl("%.4f" % tc, 10),
       padl("%.4f" % (tc - Z975), 18),
       padl("%.2f" % ((tc / Z975 - 1) * 100), 20))
P_("df -> cheksiz: t_0,975 -> %.6f (aynan z)", Z975)
P_("Amaliy xulosa: df>=30 da farq 4%% dan kam; df=2 da 4,3 barobar.")
```

```python
# --- Grafik: n=5 dagi T ning taqsimoti, z va t egri chiziqlari ---------
T5c = T5[np.abs(T5) <= 6]                  # dumlar 6 dan tashqarida ham bor
xs = np.linspace(-6, 6, 501)
fig, ax = plt.subplots()
ax.hist(T5c, bins=90, density=True, color=oiSky, alpha=0.55)
ax.plot(xs, stats.norm.pdf(xs), "--", color=oiBlue, lw=1.4)
ax.plot(xs, stats.t.pdf(xs, df=4), "-", color=oiVerm, lw=1.6)
for v in (-1.96, 1.96):
    ax.plot([v, v], [0, 0.30], ":", color=oiGrey)
ax.text(3.5, 0.33, "solid line: t with 4 df", color=oiVerm, fontsize=8)
ax.text(3.5, 0.29, "dashed line: N(0,1)", color=oiBlue, fontsize=8)
ax.text(-5.9, 0.33, "dotted: +/- 1.96", color=oiGrey, fontsize=8)
ax.set_xlim(-6, 6)
ax.set_xlabel("T"); ax.set_ylabel("density")
ax.set_title("Simulated distribution of T = (xbar - mu)/(s/sqrt(n)), n = 5")
fig.suptitle("40000 samples; the histogram follows t(4), not N(0,1)",
             y=0.94, fontsize=9, color="grey")
fig.text(0.99, -0.02, "MINSTD, seed 20261105 | tails beyond |T|=6 are cut "
         "for display", ha="right", fontsize=8, color=oiGrey)
plt.show()
```

**Talqin.** Yozma javob bering:

1. $n=3$ da $\SD(T)=4{,}61$, nazariy qiymat esa **cheksiz**. Bu
   qarama-qarshilikmi? $t_2$ taqsimotining dispersiyasi haqida nima

2. $\Prob(|T|>1{,}96)$ ustunini oʻqing. $n=5$ da u $0{,}05$ oʻrniga

   Bashoratingizni yozib qoʻying — 5-topshiriqda tekshirasiz.

## 5A. Bitta oraliq — Oʻzbekiston bugʻdoy hosildorligi

**Statistik tushuncha.** $\sigma$ nomaʼlum boʻlganda oʻrtacha uchun
$100(1-\alpha)\,\%$ ishonch oraligʻi:

$$\xbar \pm t_{1-\alpha/2}(n-1)\cdot\frac{s}{\sqrt{n}},
\qquad \SE=\frac{s}{\sqrt{n}} .$$

**Shartlar:** (i) tasodifiy tanlanma; (ii) kuzatuvlar bogʻliqsiz;
(iii) bosh toʻplam normalga yaqin **yoki** $n$ yetarlicha katta.

**Maʼlumot.** Oʻzbekistonda bugʻdoy hosildorligi, 2009–2018,
$n=10$ yil, t/ga. **Haqiqiy maʼlumot**, manba: OWID / FAOSTAT.

> **Ehtiyot boʻling.** Bu — *vaqt qatori*, tasodifiy tanlanma emas.
> Ketma-ket yillar bir-biriga bogʻliq (ob-havo, texnologiya, siyosat).
> Biz oraliqni **mashq sifatida** quramiz va (iii) shart buzilishini
> talqin bosqichida ochiq aytamiz. Buni yashirish — eng jiddiy xato.

**Kod.**

```python
# --- 5A: bitta oraliqni bosqichma-bosqich qurish -----------------------
# QOIDA: mamlakat nomini kod ichida literal YOZMAYMIZ — fayldan olamiz
# va POZITSIYA boʻyicha tanlaymiz (C lokalda solishtirish ishlamaydi).
mamlakatlar = list(pd.unique(hosil["mamlakat"]))
uz_nom = mamlakatlar[0]                        # birinchi qator = Oʻzbekiston
sat = ((hosil["mamlakat"] == uz_nom) & (hosil["yil"] >= 2009)
       & (hosil["yil"] <= 2018))
x = hosil.loc[sat, "bugdoy"].to_numpy(dtype=float)
yil = hosil.loc[sat, "yil"].to_numpy(dtype=int)

P_("=== 5A-TOPSHIRIQ: BITTA ISHONCH ORALIGʻI (haqiqiy maʼlumot) ===")
P_("manba: OWID / FAOSTAT; bugʻdoy hosildorligi, t/ga; %s, %d-%d",
   pad(uz_nom, 11), yil.min(), yil.max())
P_("qiymatlar: %s", ", ".join("%.3f" % v for v in x))
n5 = len(x); xb5 = float(x.mean()); ss5 = math.sqrt(var_(x))
se5 = ss5 / math.sqrt(n5)
P_("")
P_("1-qadam  n            = %d", n5)
P_("2-qadam  yigʻindi     = %.4f", x.sum())
P_("3-qadam  xbar         = %.6f", xb5)
P_("4-qadam  sum(x-xbar)^2= %.6f", ((x - xb5) ** 2).sum())
P_("5-qadam  s^2          = %.6f   (maxraj n-1 = %d)", var_(x), n5 - 1)
P_("6-qadam  s            = %.6f", ss5)
P_("7-qadam  SE = s/sqrt(n) = %.6f", se5)
P_("")
P_("%s%s%s%s%s%s", padl("daraja", 8), padl("t kritik", 11),
   padl("xatolik chegarasi", 20), padl("quyi", 10), padl("yuqori", 10),
   padl("eni", 10))
enlar = []
for par in [(0.90, 0.95), (0.95, 0.975), (0.99, 0.995)]:
    tc = tkr(par[1], n5 - 1); E = tc * se5
    enlar.append(2 * E)
    P_("%s%s%s%s%s%s", padl("%.0f%%" % (par[0] * 100), 8),
       padl("%.4f" % tc, 11), padl("%.4f" % E, 20),
       padl("%.4f" % (xb5 - E), 10), padl("%.4f" % (xb5 + E), 10),
       padl("%.4f" % (2 * E), 10))
P_("")
P_("Ishonch darajasi oshsa oraliq KENGAYADI: %.4f -> %.4f -> %.4f",
   enlar[0], enlar[1], enlar[2])
P_("99%% oraliq 95%% dan %.4f barobar keng.", enlar[2] / enlar[1])
P_("")
P_("XATO QILSAK: z=1,96 ni t oʻrniga ishlatsak")
P_("  [%.4f; %.4f]  eni %.4f", xb5 - Z975 * se5, xb5 + Z975 * se5,
   2 * Z975 * se5)
P_("  bu 95%% t-oraligʻidan %.2f%% TOR -> ORTIQCHA ishonch.",
   (1 - 2 * Z975 * se5 / enlar[1]) * 100)
```

## 5B. Qoplash tajribasi — 20 000 ta oraliq

**Statistik tushuncha.** Ishonch oraligʻining taʼrifi **chastotaviy**:
usulni koʻp marta takrorlasak, oraliqlarning $95\,\%$ i haqiqiy
parametrni qoplaydi. Bosh toʻplam bizga maʼlum boʻlgani uchun buni
**sanash** mumkin. Uchta usulni yonma-yon qoʻyamiz:

| Usul | Formula | Nimani bilishni talab qiladi |
|---|---|---|
| $z$ va $s$ | $\xbar\pm1{,}96\,s/\sqrt n$ | hech narsa (lekin notoʻgʻri) |
| $t$ va $s$ | $\xbar\pm t_{0{,}975}(n-1)\,s/\sqrt n$ | hech narsa |
| $z$ va $\sigma$ | $\xbar\pm1{,}96\,\sigma/\sqrt n$ | $\sigma$ ni (biz bilamiz!) |

Uchinchi ustun — **diagnostika asbobi**. U $\sigma$ nomaʼlumligini
umuman yoʻqotadi. Agar qoplash **shunda ham** $0{,}95$ dan past boʻlsa,

# 7-topshiriq. Ulush uchun oraliq: Wald, Vilson, Agresti–Coull {#t7}

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 15 ·
> **Vaqt:** 19 daqiqa

**Statistik tushuncha.** Ulush uchun eng koʻp oʻrgatiladigan formula —
**Wald**:

$$\phat \pm z_{1-\alpha/2}\sqrt{\frac{\phat(1-\phat)}{n}} .$$

U ikki joyda taxminiy: (i) binomial taqsimotni normal bilan
almashtiradi; (ii) $\SE$ da nomaʼlum $p$ oʻrniga $\phat$ ni qoʻyadi.
Ikkinchisi $\phat$ chetga yaqin boʻlganda halokatli. **Vilson** va
**Agresti–Coull** shu nuqsonni tuzatadi.

> **Muhim uslubiy nuqta.** Ulushda qoplashni **simulyatsiya qilish
> shart emas**: $X\sim\Bin(n,p)$ diskret, shuning uchun barcha
> mumkin boʻlgan $x=0,1,\dots,n$ boʻyicha **yigʻib** aniq
> hisoblash mumkin:
> $$\text{qoplash}(p,n)=\sum_{x=0}^{n}\binom{n}{x}p^{x}(1-p)^{n-x}\,
>   \mathbf{1}\{L(x,n)\le p\le U(x,n)\} .$$
> Bu yerda Monte-Karlo xatosi **umuman yoʻq**.

**Maʼlumot.** Kino reytingi ustuni. **Diqqat: 9 ta filmda reyting
yoʻq.** Bu — laboratoriyaning ikkinchi tuzogʻi.

**Kod.**

```python
# --- 7-topshiriq (0): YOʻQ QIYMATLARNI OCHIQ QARSHI OLISH --------------
# TUZOQ: R da NA == "PG-13" -> NA (TRUE ham, FALSE ham emas), va
# sum(...) natijasi NA ga aylanadi — XATO XABARI CHIQMAYDI.
# Python (pandas) da esa NaN == "PG-13" -> False, yaʼni jimgina
# «PG-13 emas» deb hisoblanadi. IKKI TIL IKKI XIL JAVOB BERADI.
# Yechim: NA ni HAR DOIM oshkora boshqarish.
reyt = kino["reyting"]
yoq = reyt.isna().to_numpy()
P_("=== 7-TOPSHIRIQ (0): YOʻQ QIYMATLAR ===")
P_("jami film            = %d", len(reyt))
P_("reytingi yoʻq (NA)   = %d", int(yoq.sum()))
P_("reytingi bor         = %d", int((~yoq).sum()))
P_("")
pg13 = ((~yoq) & (reyt.to_numpy() == "PG-13")).astype(int)  # NA -> 0, OSHKORA
p0 = float(pg13.mean())                        # maxraj: BARCHA 2467 film
p_bor = float(pg13.sum() / (~yoq).sum())       # maxraj: faqat reytingi borlar
P_("PG-13 filmlar soni   = %d", int(pg13.sum()))
P_("p (barcha filmlar)   = %d/%d = %.6f", int(pg13.sum()), len(reyt), p0)
P_("p (reytingi borlar)  = %d/%d = %.6f", int(pg13.sum()),
   int((~yoq).sum()), p_bor)
P_("farq = %.6f — kichik, AMMO uni JIM oʻtkazib yuborish mumkin emas.",
   p_bor - p0)
P_("Bugun bosh toʻplam parametri sifatida p = %.6f ni olamiz", p0)
P_("(taʼrif: 'barcha 2467 filmdan PG-13 boʻlganlari ulushi').")
P_("Har qanday hisobotda MAXRAJ nima ekanini YOZISH SHART.")
```

```python
# --- 7-topshiriq (1): uchta formula ------------------------------------
# Har biri (quyi, yuqori) juftligini qaytaradi; x vektor boʻlishi mumkin.
def wald(x, n, z=Z975):
    ph = x / n; se = np.sqrt(ph * (1 - ph) / n)
    return {"lo": ph - z * se, "hi": ph + z * se}

def vilson(x, n, z=Z975):
    ph = x / n; d = 1 + z * z / n
    c_ = (ph + z * z / (2 * n)) / d
    hw = z * np.sqrt(ph * (1 - ph) / n + z * z / (4 * n * n)) / d
    return {"lo": c_ - hw, "hi": c_ + hw}

def agresti(x, n, z=Z975):
    nt = n + z * z; pt = (x + z * z / 2) / nt
    se = np.sqrt(pt * (1 - pt) / nt)
    return {"lo": pt - z * se, "hi": pt + z * se}

# ANIQ qoplash: barcha x boʻyicha yigʻindi (simulyatsiya EMAS).
def aniq_qoplash(p, n, usul):
    x = np.arange(n + 1)
    pr = stats.binom.pmf(x, n, p)
    iv = usul(x, n)
    return float(pr[(iv["lo"] <= p) & (p <= iv["hi"])].sum())

P_("")
P_("=== 7-TOPSHIRIQ (1): ANIQ QOPLASH (barcha x boʻyicha yigʻindi) ===")
P_("%s%s%s%s%s", padl("p", 6), padl("n", 6), padl("Wald", 10),
   padl("Vilson", 10), padl("Agresti-Coull", 16))
for pp in [0.05, 0.10, 0.30, 0.50]:
    for n_ in [20, 30, 50, 100]:
        P_("%s%s%s%s%s", padl("%.2f" % pp, 6), padl("%d" % n_, 6),
           padl("%.4f" % aniq_qoplash(pp, n_, wald), 10),
           padl("%.4f" % aniq_qoplash(pp, n_, vilson), 10),
           padl("%.4f" % aniq_qoplash(pp, n_, agresti), 16))
P_("")
P_("DIQQAT: jadvaldagi 'sakrashlar' XATO EMAS. X diskret boʻlgani uchun")
P_("qoplash n bilan MONOTON emas — TEBRANADI (Braun-Say-Dasgupta, 2001).")
P_("Shuning uchun bitta n ga qarab hukm chiqarish notoʻgʻri; n=10..200")
P_("boʻyicha OʻRTACHA xulq-atvorga qaraymiz:")
P_("%s%s%s%s", padl("p", 6), padl("Wald", 12), padl("Vilson", 12),
   padl("Agresti-C.", 12))
for pp in [0.05, 0.10, 0.30, 0.50]:
    ns = range(10, 201)
    aw = float(np.mean([aniq_qoplash(pp, nn, wald) for nn in ns]))
    av = float(np.mean([aniq_qoplash(pp, nn, vilson) for nn in ns]))
    aa = float(np.mean([aniq_qoplash(pp, nn, agresti) for nn in ns]))
    P_("%s%s%s%s", padl("%.2f" % pp, 6), padl("%.4f" % aw, 12),
       padl("%.4f" % av, 12), padl("%.4f" % aa, 12))
P_("")
P_("Tartib aniq: Wald < Vilson ~ Agresti-Coull.")
P_("Faqat Vilson va Agresti-Coull 0,95 atrofida turadi.")
P_("Wald ni oʻrganamiz, chunki u HAMMA JOYDA uchraydi — va uning")
P_("QAYERDA buzilishini bilishimiz kerak.")
```

```python
# --- Grafik: qoplashning n boʻyicha TEBRANISHI ------------------------
ns7 = list(range(10, 101))
qw = [aniq_qoplash(0.05, nn, wald) for nn in ns7]
qv = [aniq_qoplash(0.05, nn, vilson) for nn in ns7]
qa = [aniq_qoplash(0.05, nn, agresti) for nn in ns7]
fig, ax = plt.subplots()
ax.axhline(0.95, color=oiGrey, lw=1.0)
ax.plot(ns7, qw, "-", color=oiVerm, lw=1.0, label="Wald")
ax.plot(ns7, qv, "--", color=oiBlue, lw=1.0, label="Wilson")
ax.plot(ns7, qa, "-.", color=oiGreen, lw=1.0, label="Agresti-Coull")
ax.set_ylim(0.55, 1.0)
ax.set_xlabel("sample size n")
ax.set_ylabel("exact coverage (summed over all x)")
ax.set_title("Exact coverage of nominal 95% intervals, p = 0.05")
ax.legend(frameon=False, fontsize=9, loc="lower right")
fig.suptitle("Coverage oscillates with n because X is discrete "
             "(Brown-Cai-DasGupta 2001)", y=0.94, fontsize=9, color="grey")
fig.text(0.99, -0.02, "No simulation: computed by summing the binomial pmf",
         ha="right", fontsize=8, color=oiGrey)
plt.show()
```

```python
# --- 7-topshiriq (2): qoʻlda hisoblash --------------------------------
xx = 34; n7 = 100; ph = xx / n7
P_("")
P_("=== 7-TOPSHIRIQ (2): QOʻLDA HISOB — n=100 tanlanmada 34 ta PG-13 ===")
P_("phat = %d/%d = %.4f", xx, n7, ph)
P_("shart: n*phat = %.1f >= 10 va n*(1-phat) = %.1f >= 10 -> %s",
   n7 * ph, n7 * (1 - ph), ha(n7 * ph >= 10 and n7 * (1 - ph) >= 10))
P_("SE = sqrt(%.2f*%.2f/%d) = sqrt(%.8f) = %.6f",
   ph, 1 - ph, n7, ph * (1 - ph) / n7, math.sqrt(ph * (1 - ph) / n7))
P_("")
P_("%s%s%s%s%s", padl("usul", 16), padl("quyi", 12), padl("yuqori", 12),
   padl("eni", 12), padl("p ni qopladimi?", 16))
for nom_, f_ in [("Wald", wald), ("Vilson", vilson),
                 ("Agresti-Coull", agresti)]:
    iv = f_(xx, n7)
    P_("%s%s%s%s%s", padl(nom_, 16), padl("%.6f" % iv["lo"], 12),
       padl("%.6f" % iv["hi"], 12), padl("%.6f" % (iv["hi"] - iv["lo"]), 12),
       padl(ha(iv["lo"] <= p0 <= iv["hi"]), 16))
P_("haqiqiy p = %.6f", p0)
P_("")
P_("--- Tanlanma hajmini rejalashtirish: n >= z^2 p(1-p)/E^2 ---")
P_("%s%s%s", padl("E", 8), padl("p=0,50 (eng yomon)", 20),
   padl("p=0,36 (maʼlum)", 18))
for E in [0.05, 0.03, 0.02, 0.01]:
    P_("%s%s%s", padl("%.2f" % E, 8),
       padl("%d" % math.ceil(Z975 ** 2 * 0.25 / E ** 2), 20),
       padl("%d" % math.ceil(Z975 ** 2 * p0 * (1 - p0) / E ** 2), 18))
P_("p nomaʼlum boʻlsa HAR DOIM p=0,5 ni oling — bu eng yomon hol.")
```

# 8-topshiriq. Butstrep: ishlaydi va ishlamaydi {#t8}

> **Blum:** Yaratish · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 15 ·
> **Vaqt:** 20 daqiqa

**Statistik tushuncha.** Butstrep (Efron, 1979) bitta gʻoyaga tayanadi:

> Bosh toʻplamdan qayta tanlab ololmaymiz. Ammo **tanlanmaning
> oʻzidan** qaytarish bilan qayta tanlash mumkin. Tanlanma — bosh
> toʻplamning eng yaxshi mavjud surati.

Algoritm: asl tanlanmadan $n$ ta kuzatuvni **qaytarish bilan** tanlab,
statistikani hisoblaymiz; buni $B$ marta takrorlaymiz; hosil boʻlgan
$B$ ta qiymatning standart ogʻishi — **butstrep $\SE$**; ularning
$2{,}5$ va $97{,}5$ protsentillari — **protsentil butstrep oraligʻi**.

Bugun butstrepni **uchta joyda** sinaymiz:

| Qayerda | Klassik formula bormi | Kutish |
|---|---|---|
| Oʻrtacha | **Bor** ($s/\sqrt n$) | tekshirish imkoni |
| Median | **Yoʻq** | butstrepning asosiy qiymati |
| Maksimum | Yoʻq | **butstrep buziladi** |

**Kod.**

```python
# --- 8A: bitta haqiqiy tanlanma (n=40) --------------------------------
g = minstd(20261400)
samp = b[indeks(g, 40, Nb)]
sxb = float(samp.mean()); ss = math.sqrt(var_(samp)); sse = ss / math.sqrt(40)
smd = float(np.median(samp))
tc8 = tkr(0.975, 39)

P_("=== 8-TOPSHIRIQ: BUTSTREP ===")
P_("asl tanlanma n=40 (urugʻ 20261400), saralangan, mln $:")
P_("  %s", ", ".join("%.2f" % v for v in np.sort(samp)))
P_("xbar = %.6f   s = %.6f   SE = %.6f   median = %.4f",
   sxb, ss, sse, smd)
P_("klassik t-oraliq 95%%: %.4f +- %.4f*%.4f = [%.4f; %.4f]  eni %.4f",
   sxb, tc8, sse, sxb - tc8 * sse, sxb + tc8 * sse, 2 * tc8 * sse)
P_("haqiqiy mu = %.4f -> qopladimi: %s", MU,
   ha(sxb - tc8 * sse <= MU <= sxb + tc8 * sse))
```

```python
# --- 8B: B = 10 000 ta qayta tanlanma ---------------------------------
# TEZLIK QOIDASI: for-sikl ISHLATMANG. Barcha indekslarni bir marta
# generatsiya qilib, B x n matritsa yasang va qator boʻyicha hisoblang.
B = 10000
g2 = minstd(20261401)
bi = indeks(g2, B * 40, 40).reshape(B, 40)
bs = samp[bi]                             # B x 40 qayta tanlanmalar
bmean = qatorOrtacha(bs)
bmed = qatorMedian(bs)

P_("")
P_("B = %d ta qayta tanlanma (urugʻ 20261401)", B)
P_("--- OʻRTACHA uchun (bu yerda KLASSIK FORMULA BOR — tekshiramiz) ---")
P_("butstrep taqsimotining oʻrtachasi = %.6f  (asl xbar = %.6f)",
   bmean.mean(), sxb)
P_("butstrep SE = SD(butstrep oʻrtachalari) = %.6f", sd_(bmean))
P_("klassik  SE = s/sqrt(n)                 = %.6f", sse)
P_("farq = %.2f%% — deyarli bir xil. BUTSTREP FORMULANI TASDIQLADI.",
   abs(sd_(bmean) - sse) / sse * 100)
lo_p = kv(bmean, 0.025); hi_p = kv(bmean, 0.975)
P_("protsentil butstrep 95%% oraliq: [%.4f; %.4f]  eni %.4f",
   lo_p, hi_p, hi_p - lo_p)
P_("klassik t-oraliq:               [%.4f; %.4f]  eni %.4f",
   sxb - tc8 * sse, sxb + tc8 * sse, 2 * tc8 * sse)
P_("butstrep oraligʻi %.4f barobar (t biroz kengroq — u qoʻshimcha",
   (hi_p - lo_p) / (2 * tc8 * sse))
P_("ehtiyot zaxirasini oʻz ichiga oladi).")
P_("")
P_("--- MEDIAN uchun (bu yerda KLASSIK FORMULA YOʻQ) ---")
P_("asl median  = %.4f", smd)
P_("butstrep SE = %.6f", sd_(bmed))
lo_m = kv(bmed, 0.025); hi_m = kv(bmed, 0.975)
P_("protsentil butstrep 95%% oraliq: [%.4f; %.4f]", lo_m, hi_m)
P_("haqiqiy bosh toʻplam medianasi = %.4f -> qopladimi: %s",
   MEDB, ha(lo_m <= MEDB <= hi_m))
P_("MANA BUTSTREPNING ASOSIY QIYMATI: formula boʻlmagan joyda ishlaydi.")
P_("")
P_("DIQQAT: median uchun oraliq CHEKKALARI — asl tanlanmadagi sonlar")
P_("(%.2f va %.2f). Butstrep tanlanmada YOʻQ qiymatni yarata olmaydi.",
   lo_m, hi_m)
```

```python
# --- 8C: butstrep ISHLAMAYDIGAN hol — maksimum ------------------------
g3 = minstd(20261500)
sm2 = b[indeks(g3, 50, Nb)]
g4 = minstd(20261501)
bmx = sm2[indeks(g4, 4000 * 50, 50).reshape(4000, 50)].max(axis=1)
vals, cnts = np.unique(bmx, return_counts=True)

P_("")
P_("--- BUTSTREP ISHLAMAYDIGAN HOL: MAKSIMUM ---")
P_("n=50 tanlanma (urugʻ 20261500), tanlanma maksimumi = %.4f", sm2.max())
P_("haqiqiy bosh toʻplam maksimumi = %.4f", b.max())
P_("4000 ta butstrep maksimumi faqat %d ta TURLI qiymat oladi:", len(vals))
P_("  %s", ", ".join("%.1f" % v for v in vals))
P_("ularning %.1f%% i asl maksimumning AYNAN oʻzi.", cnts.max() / 4000 * 100)
P_("butstrep taqsimoti maksimumdan KATTA qiymat HECH QACHON bermaydi.")
P_("-> oraliq siljigan; u haqiqiy maksimumni (%.1f) qoplay OLMAYDI.",
   b.max())
P_("SABAB: maksimum — 'silliq' statistika emas. Butstrepning nazariy")
P_("asosi silliq (differensiallanuvchi) statistikalar uchun ishlaydi.")
```

```python
# --- Grafik: uchta butstrep taqsimoti ---------------------------------
fig, axs = plt.subplots(1, 3, figsize=(9.2, 3.2))
for axx, dat, nom_ in zip(
        axs, [bmean, bmed, bmx],
        ["mean (formula exists)", "median (no formula)",
         "maximum (bootstrap FAILS)"]):
    axx.hist(dat, bins=45, color=oiSky)
    axx.set_title(nom_, fontsize=9)
    axx.set_xlabel("bootstrap statistic, million $", fontsize=8)
axs[0].set_ylabel("count")
fig.suptitle("Three bootstrap distributions from the same idea\n"
             "The maximum takes only 6 distinct values and never exceeds "
             "the sample max", fontsize=9)
fig.text(0.99, -0.03, "MINSTD, seeds 20261401 (B=10000) and 20261501 (B=4000)",
         ha="right", fontsize=8, color=oiGrey)
fig.tight_layout()
plt.show()
```

```python
# --- 8D: butstrep sehrli tayoqcha EMAS — qoplashni sanaymiz -----------
# 500 ta haqiqiy tanlanma, har birida B=2000 butstrep. Bu qism eng
# ogʻiri (~45 s); u butstrepning HAQIQIY qoplashini oʻlchaydi.
P_("")
P_("--- BUTSTREP ORALIGʻINING QOPLASHI (haqiqiy tekshiruv) ---")
P_("500 ta haqiqiy tanlanma, har birida B=2000 butstrep; urugʻ 20261450+n")
P_("%s%s%s", padl("n", 5), padl("butstrep protsentil", 22),
   padl("klassik t", 12))
for n_ in [10, 30, 100]:
    gg = minstd(20261450 + n_)
    reps = 500; okB = 0; okT = 0
    idxs = indeks(gg, reps * n_, Nb).reshape(reps, n_)
    gb = minstd(20261460 + n_)
    tcc = tkr(0.975, n_ - 1)
    for r in range(reps):
        sm = b[idxs[r]]
        bidx = indeks(gb, 2000 * n_, n_).reshape(2000, n_)
        bm = qatorOrtacha(sm[bidx])
        if kv(bm, 0.025) <= MU <= kv(bm, 0.975):
            okB += 1
        if abs(sm.mean() - MU) < tcc * math.sqrt(var_(sm)) / math.sqrt(n_):
            okT += 1
    P_("%s%s%s", padl("%d" % n_, 5),
       padl("%.4f" % (okB / reps), 22),
       padl("%.4f" % (okT / reps), 12))
P_("Ikkalasi ham 0,95 dan PAST (bosh toʻplam qiyshiq).")
P_("BUTSTREP SEHRLI TAYOQCHA EMAS: u FORMULANI almashtiradi,")
P_("MAʼLUMOTNI emas. Qiyshiqlik muammosi joyida qoladi.")
```
