```python
# --- IKKI DAFTAR BIR XIL MATN CHIQARISHI UCHUN YORDAMCHILAR -----------
# (1) ha(): R «TRUE», Python «True» yozadi --- mantiqiy qiymatni
#     OʻZIMIZ formatlaymiz.
def ha(b):
    return "HA" if bool(b) else "YOʻQ"


# (2) pad()/padl(): R BAYT boʻyicha, Python BELGI boʻyicha toʻldiradi.
def pad(s, n):
    s = str(s)
    return s + " " * max(0, n - len(s))


def padl(s, n):
    s = str(s)
    return " " * max(0, n - len(s)) + s


# (3) Dispersiya, SD va korrelyatsiya: MAXRAJ HAR DOIM OSHKORA.
def varS(x):
    x = np.asarray(x, float)
    return float(((x - x.mean()) ** 2).sum()) / (len(x) - 1)


def sdS(x):
    return math.sqrt(varS(x))


def korr(x, y):
    x = np.asarray(x, float); y = np.asarray(y, float)
    dx = x - x.mean(); dy = y - y.mean()
    return float((dx * dy).sum()) / math.sqrt(float((dx ** 2).sum())
                                              * float((dy ** 2).sum()))


# (4) ENG KICHIK KVADRATLAR + XULOSA (12--13-haftalardan oʻzgarmadi).
def ols(x, y):
    x = np.asarray(x, float); y = np.asarray(y, float)
    n = len(x); xb = float(x.mean()); yb = float(y.mean())
    sxx = float(((x - xb) ** 2).sum()); syy = float(((y - yb) ** 2).sum())
    sxy = float(((x - xb) * (y - yb)).sum())
    b1 = sxy / sxx; b0 = yb - b1 * xb
    e = y - (b0 + b1 * x)
    sse = float((e ** 2).sum()); ssr = syy - sse
    se = math.sqrt(sse / (n - 2)); seb1 = se / math.sqrt(sxx); t1 = b1 / seb1
    return {"n": n, "xb": xb, "sxx": sxx, "b1": b1, "b0": b0, "e": e,
            "se": se, "seb1": seb1, "t1": t1,
            "p1": 2 * float(stats.t.sf(abs(t1), n - 2)), "r2": ssr / syy}


# (5) t va normal taqsimot.
def tq(pp, df):
    return float(stats.t.ppf(pp, df))


def tp2(tt, df):
    return 2 * float(stats.t.sf(abs(tt), df))


def np2(zz):
    return 2 * float(stats.norm.sf(abs(zz)))


# (6) Kvantil (type 7) --- OSHKORA yozilgan. Formulani oʻzimiz yozsak,
#     ikki tilda bir xil ekani KOʻRINIB turadi.
def kv7(x, pp):
    s = np.sort(np.asarray(x, float)); n = len(s); out = []
    for p in pp:
        h = (n - 1) * p; lo = int(math.floor(h)); fr = h - lo
        hi = min(lo + 1, n - 1)
        out.append(float(s[lo] + fr * (s[hi] - s[lo])))
    return out


# (7) MINSTD LCG: x <- 48271 x mod 2147483647. Ikki tilda AYNAN bir xil
#     ketma-ketlik. numpy.random ISHLATILMAYDI (7-topshiriqqa qarang).
class LCG:
    def __init__(self, urug):
        self.x = urug % 2147483647

    def uv(self, n):
        o = np.empty(n); x = self.x
        for i in range(n):
            x = (48271 * x) % 2147483647
            o[i] = x / 2147483647
        self.x = x
        return o

    def idx(self, n, hi):
        return (self.uv(n) * hi).astype(int)


# (8) F203: standart xatodan maʼnoli xonalar soni va chop etish.
def xona(se):
    return max(0, int(math.ceil(-math.log10(se))) + 1)


def chop(v, d):
    dek = max(0, d - 1 - int(math.floor(math.log10(abs(v)))))
    return "%.*f" % (dek, v)
```

## 1-topshiriq. **48 ta halol spetsifikatsiya**

> **Blum:** Qoʻllash · **Qiyinlik:** 3 (Amaliy) · **Ball:** 14 ·
> **Vaqt:** 14 daqiqa

**Savol.** *PG-13 filmlari R filmlaridan koʻra koʻproq foyda
keltiradimi?*

Bu savolga javob berish uchun tahlilchi **hech kimni aldamasdan**
beshta qaror qabul qilishi kerak. Har birining ikki yoki uchta
**asosli** varianti bor:

| # | Qaror | Variantlar |
|---|---|---|
| 1 | «Foydali» chegarasi | $2{,}0$ · $2{,}5$ · $3{,}0$ karra byudjet |
| 2 | Daromadi nol boʻlgan qatorlar | chiqarilsin / qolsin |
| 3 | Janrga tuzatish (13-hafta, F197) | ha / yoʻq |
| 4 | Eng yuqori 1 foiz byudjet | chiqarilsin / qolsin |
| 5 | PG ni PG-13 ga qoʻshish | ha / yoʻq |

$3\times2\times2\times2\times2=48$. **Hech biri notoʻgʻri emas.**
Bu — **tarmoqlanuvchi yoʻllar bogʻi** (*garden of forking paths*):
har bir tugunda halol qaror, oxirida esa 48 ta halol javob.

$$\text{Koʻpolam} = \{\hat\theta(s):\ s\in S\},\quad |S|=48 \tag{F202}$$

```python
# --- BITTA SPETSIFIKATSIYA --------------------------------------------
# chegara: 2.0 / 2.5 / 3.0     nol: 0/1 (daromadi nol qatorlar chiqsinmi)
# tuz: 0/1 (janrga tuzatish)   top: 0/1 (eng yuqori 1 foiz byudjet chiqsinmi)
# grp: 0 = PG-13 vs R          1 = (PG + PG-13) vs R
def spek(chegara, nol, tuz, top, grp):
    m = byu > 0                                    # byudjeti nolga teng emas
    if nol == 1:
        m = m & (dar > 0)                          # qaror 2
    if top == 1:
        q = kv7(byu[m], [0.99])[0]; m = m & (byu <= q)      # qaror 4
    f = (dar > chegara * byu).astype(float)        # qaror 1: «foydali»
    if grp == 0:                                   # qaror 5
        m = m & ((rey == "PG-13") | (rey == "R")); nomA = "PG-13"; rr = rey
    else:
        m = m & ((rey == "PG") | (rey == "PG-13") | (rey == "R")); nomA = "PG+"
        rr = np.where((rey == "PG") | (rey == "PG-13"), "PG+", rey)
    if tuz == 1:                                   # qaror 3: janrga tuzatish
        num = 0.0; den = 0.0; vr = 0.0             # F197 (13-hafta)
        for g in JAN:                              # ASCII tartib --- MUHIM
            ma = m & (jan == g) & (rr == nomA)
            mb = m & (jan == g) & (rr == "R")
            la = int(ma.sum()); lb = int(mb.sum())
            if la < 10 or lb < 10:
                continue                           # kichik qatlam tashlanadi
            pa = float(f[ma].sum()) / la; pb = float(f[mb].sum()) / lb
            w = la + lb
            num += (pa - pb) * w; den += w
            vr += (w ** 2) * (pa * (1 - pa) / la + pb * (1 - pb) / lb)
        dif = num / den; se = math.sqrt(vr) / den
    else:
        ma = m & (rr == nomA); mb = m & (rr == "R")
        n1 = int(ma.sum()); n2 = int(mb.sum())
        s1 = float(f[ma].sum()); s2 = float(f[mb].sum())
        pu = (s1 + s2) / (n1 + n2)                 # birlashgan ulush
        se = math.sqrt(pu * (1 - pu) * (1 / n1 + 1 / n2))
        dif = s1 / n1 - s2 / n2
    z = dif / se
    return {"dif": dif, "se": se, "z": z, "p": np2(z)}


# --- BARCHA 48 SPETSIFIKATSIYA ----------------------------------------
MV = []
for chegara in (2.0, 2.5, 3.0):
    for nol in (0, 1):
        for tuz in (0, 1):
            for top in (0, 1):
                for grp in (0, 1):
                    r = spek(chegara, nol, tuz, top, grp)
                    MV.append({"chegara": chegara, "nol": nol, "tuz": tuz,
                               "top": top, "grp": grp, "dif": r["dif"],
                               "se": r["se"], "p": r["p"]})
MVp = np.array([r["p"] for r in MV])
MVd = np.array([r["dif"] for r in MV])
nsig = int((MVp < 0.05).sum()); nns = int((MVp >= 0.05).sum())

och("w14-t1a")
P_("=== 1-TOPSHIRIQ: 48 TA HALOL SPETSIFIKATSIYA (F202) ===")
P_("Savol: PG-13 filmlari R filmlaridan koʻra koʻproq foyda keltiradimi?")
P_("Besh qaror: 3 x 2 x 2 x 2 x 2 = %d spetsifikatsiya. Hech biri notoʻgʻri emas.",
   len(MV))
P_("")
P_("Spetsifikatsiyalar soni:   %d", len(MV))
P_("p <  0.05 (ahamiyatli):    %d", nsig)
P_("p >= 0.05 (ahamiyatsiz):   %d", nns)
P_("Ulush: %.6f  ---  deyarli tanga tashlash.", nsig / len(MV))
P_("")
P_("Effekt bahosi diapazoni: %.6f .. %.6f  (%.4f marta)",
   MVd.min(), MVd.max(), MVd.max() / MVd.min())
P_("p-qiymat diapazoni:      %.6f .. %.6f", MVp.min(), MVp.max())
P_("Median effekt: %.6f   median p: %.6f", np.median(MVd), np.median(MVp))
P_("")
P_("TEKSHIRUV: %d + %d = %d ?  %s",
   nsig, nns, len(MV), ha(nsig + nns == len(MV)))
P_("TEKSHIRUV: ham ahamiyatli, ham ahamiyatsiz bormi?  %s",
   ha(nsig > 0 and nns > 0))
yop()
```

## 2-topshiriq. Spetsifikatsiya egri chizigʻi

> **Blum:** Tahlil · **Qiyinlik:** 4 (Analitik) · **Ball:** 10 ·

**Savol.** 48 ta sonni **bitta rasmda** koʻrsatish mumkinmi — va
qaysi qaror natijani **hal qiladi**?

**Spetsifikatsiya egri chizigʻi** (*specification curve*): effekt
bahosi oʻsish tartibida chiziladi, ahamiyatlilik **rang bilan emas,
shakl bilan ham** koʻrsatiladi (rang koʻrmaslik uchun).

## 3-topshiriq. **Urugʻsiz butstrep**

> **Blum:** Baholash · **Qiyinlik:** 5 (Ilgʻor) · **Ball:** 12 ·
> **Vaqt:** 12 daqiqa

**Savol.** Kod bir xil, maʼlumot bir xil, formula bir xil. Faqat
**urugʻ yozilmagan**. Javob har safar bir xil boʻladimi?

$$\text{Takrorlanuvchanlik} = \text{maʼlumot} + \text{kod} +
  \text{urugʻ} + \text{versiyalar} \tag{F204}$$

```python
# --- 12-haftadagi model AYNAN qaytadi ---------------------------------
kz = hosil[(hosil["mamlakat"] == mam[2]) & (hosil["yil"] >= 2010)
           & (hosil["yil"] <= 2018)].sort_values("yil")
x1 = (kz["yil"].values - 2010).astype(float)
y1 = kz["kartoshka"].values.astype(float)
o1 = ols(x1, y1)


# --- Bitta butstrep ishga tushirish (berilgan urugʻ bilan) ------------
def butstrep(urug, M=4000):
    g = LCG(urug); bs = np.empty(M); k = 0
    for j in range(M):
        ii = g.idx(o1["n"], o1["n"])       # JUFTLIKLARNI qaytarib tanlaymiz
        xx = x1[ii]
        if float(((xx - xx.mean()) ** 2).sum()) < 1e-12:
            continue                       # x tarqalmagan --- tashlanadi
        yy = y1[ii]
        bs[k] = (float(((xx - xx.mean()) * (yy - yy.mean())).sum())
                 / float(((xx - xx.mean()) ** 2).sum()))
        k += 1
    return kv7(bs[:k], [0.025, 0.975])


# --- 20 ta HAR XIL urugʻ: «tahlilchi urugʻni belgilamagan» holat -------
UR = [20261401 + k * 7919 for k in range(20)]
lo = np.zeros(20); hi = np.zeros(20)
for k in range(20):
    q = butstrep(UR[k]); lo[k] = q[0]; hi[k] = q[1]

och("w14-t3a")
P_("=== 3-TOPSHIRIQ: URUGʻSIZ BUTSTREP (F204) ===")
P_("Model (12--13-hafta): Qozogʻiston kartoshka 2010-2018, n = %d", o1["n"])
P_("b1 = %.6f t/ga        SE(b1) = %.6f", o1["b1"], o1["seb1"])
P_("Har ishga tushirish: 4000 butstrep takrori, persentil oraligʻi.")
P_("")
P_("%s%s%s%s%s%s", padl("#", 3), padl("urugʻ", 11), padl("quyi", 11),
   padl("yuqori", 11), padl("kenglik", 11), padl("3 xona", 9))
for k in range(20):
    P_("%s%s%s%s%s%s", padl(k + 1, 3), padl(UR[k], 11),
       padl("%.6f" % lo[k], 11), padl("%.6f" % hi[k], 11),
       padl("%.6f" % (hi[k] - lo[k]), 11),
       padl("%.3f" % round(lo[k], 3), 9))
P_("")
P_("Quyi chegara:    %.6f .. %.6f   (tarqoqlik %.6f)",
   lo.min(), lo.max(), lo.max() - lo.min())
P_("Yuqori chegara:  %.6f .. %.6f   (tarqoqlik %.6f)",
   hi.min(), hi.max(), hi.max() - hi.min())
P_("Oraliq kengligi: %.6f .. %.6f", (hi - lo).min(), (hi - lo).max())
P_("")
P_("Hisobotda 3 xonagacha yaxlitlansa, quyi chegara %d xil qiymat oladi.",
   len(set(np.round(lo, 3).tolist())))
P_("Yaʼni: bir xil kod, bir xil maʼlumot --- %d xil hisobot.",
   len(set(np.round(lo, 3).tolist())))
yop()
```

```python
# --- Endi urugʻ BELGILANADI: uch marta ishga tushiramiz ---------------
u3 = [butstrep(UR[0]) for i in range(3)]
xil = len(set("%.9f|%.9f" % (q[0], q[1]) for q in u3))

och("w14-t3b")
P_("-- Endi urugʻ belgilanadi: urugʻ = %d, uch marta ishga tushiramiz --",
   UR[0])
for i in range(3):
    P_("ishga tushirish %d: [%.6f; %.6f]", i + 1, u3[i][0], u3[i][1])
P_("Xil natijalar soni: %d", xil)
P_("")
P_("TEKSHIRUV: uch ishga tushirish AYNAN mosmi?          %s", ha(xil == 1))
P_("TEKSHIRUV: urugʻsiz holatda nechta xil quyi chegara? %d",
   len(set(np.round(lo, 3).tolist())))
P_("")
P_("Urugʻ tasodifni YOʻQOTMAYDI --- u tasodifni QAYTA TIKLASH imkonini")
P_("beradi. Bitta qator kod (set.seed / LCG urugʻi) 3-xil hisobotni")
P_("bitta hisobotga aylantiradi.")
P_("")
P_("NIMA ISBOTLANMADI: urugʻ belgilangani natijani TOʻGʻRI qilmaydi.")
P_("U natijani faqat TAKRORLANUVCHI qiladi. Notoʻgʻri model urugʻ bilan")
P_("ham notoʻgʻri qoladi --- takrorlanuvchanlik POL, shift emas.")
yop()
```

Quyidagi olti bosqich — **sizning yakuniy hisobotingizning skeleti**

```python
och("w14-t5a")
P_("=== 5-TOPSHIRIQ: OLTI BOSQICHLI TAHLIL ===")
P_("")
P_("-- 1-bosqich. Savol -----------------------------------------------")
P_("\"Markaziy Osiyoda bugʻdoy hosildorligi 1992-2018 yillarda oʻsganmi,")
P_(" va oʻsish mamlakatlar boʻyicha bir xilmi?\"")
P_("Savol TEKSHIRILUVCHI: oʻzgaruvchi, davr va obyekt aniq.")
P_("")
P_("-- 2-bosqich. Maʼlumot va uning cheklovlari ------------------------")
P_("Manba: OWID/FAOSTAT. %d qator = %d mamlakat x %d yil.",
   len(hosil), len(mam), len(pd.unique(hosil["yil"])))
P_("Yoʻqolgan qiymatlar: %d. Tozalash: hech narsa chiqarilmadi.",
   int(hosil.isna().sum().sum()))
P_("CHEKLOV: bu MAMLAKAT darajasidagi yigʻma maʼlumot. Fermer xoʻjaligi")
P_("darajasidagi xulosa chiqarib boʻlmaydi --- ekologik xato.")
P_("")
P_("-- 3-bosqich. Tavsifiy tahlil -------------------------------------")
P_("%s%s%s%s%s%s", pad("mamlakat", 14), padl("n", 4), padl("oʻrtacha", 10),
   padl("SD", 9), padl("min", 8), padl("maks", 8))
for i in range(5):
    v = hosil["bugdoy"][hosil["mamlakat"] == mam[i]].values.astype(float)
    P_("%s%s%s%s%s%s", pad(mam[i], 14), padl(len(v), 4),
       padl("%.4f" % v.mean(), 10), padl("%.4f" % sdS(v), 9),
       padl("%.3f" % v.min(), 8), padl("%.3f" % v.max(), 8))
P_("Birlik: t/ga. Tavsif --- XULOSA emas: bu yerda hali hech narsa")
P_("umumlashtirilmadi.")
yop()
```

```python
# --- 4-bosqich: model VA shart (ikkalasi bitta jadvalda) -------------
# QOIDA (12-haftadan): har bir model ortidan SHART TEKSHIRUVI keladi.
mFit = []; mRho = np.zeros(5); mLim = np.zeros(5)
for i in range(5):
    d = hosil[hosil["mamlakat"] == mam[i]].sort_values("yil")
    o = ols(d["yil"].values.astype(float), d["bugdoy"].values.astype(float))
    e = o["e"]
    mFit.append(o)
    mRho[i] = korr(e[:-1], e[1:])           # lag-1 avtokorrelyatsiya
    mLim[i] = 1.959964 / math.sqrt(o["n"])  # chegara
buz = np.abs(mRho) > mLim

och("w14-t5b")
P_("-- 4-bosqich. Model va shartlar -----------------------------------")
P_("Model: bugdoy ~ yil (oddiy chiziqli regressiya, har mamlakat alohida).")
P_("Shart: I (bogʻliqsizlik). Vosita: qoldiqlarning lag-1 korrelyatsiyasi;")
P_("chegara 1.959964/sqrt(n).")
P_("")
P_("%s%s%s%s%s%s%s", pad("mamlakat", 14), padl("b1", 10), padl("SE", 9),
   padl("p", 12), padl("R^2", 8), padl("lag-1", 9), padl("I sharti", 12))
for i in range(5):
    P_("%s%s%s%s%s%s%s", pad(mam[i], 14),
       padl("%.5f" % mFit[i]["b1"], 10), padl("%.5f" % mFit[i]["seb1"], 9),
       padl("%.3e" % mFit[i]["p1"], 12), padl("%.4f" % mFit[i]["r2"], 8),
       padl("%.4f" % mRho[i], 9),
       padl("BUZILGAN" if buz[i] else "bajarilgan", 12))
P_("")
P_("Beshta qatordan %d tasida bogʻliqsizlik sharti BUZILGAN.", int(buz.sum()))
P_("Buzilgan qatorlar: %s",
   ", ".join([mam[i] for i in range(5) if buz[i]]))
P_("")
P_("DIQQAT: buzilgan shart b1 ni buzmaydi --- u b1 ga boʻlgan ISHONCHNI")
P_("buzadi. Shuning uchun quyida SE va p tuzatiladi, b1 emas.")
yop()
```

```python
# --- 5-bosqich: n_eff bilan tuzatish (F199) --------------------------
neff = [None] * 5; se2 = np.zeros(5); p2 = np.zeros(5)
for i in range(5):
    o = mFit[i]
    if buz[i]:
        neff[i] = o["n"] * (1 - mRho[i]) / (1 + mRho[i])          # F199
        infl = math.sqrt((1 + mRho[i]) / (1 - mRho[i]))
        se2[i] = o["seb1"] * infl
        p2[i] = tp2(o["b1"] / se2[i], max(neff[i] - 2, 1))
    else:
        se2[i] = o["seb1"]; p2[i] = o["p1"]
qOld = np.array([mFit[i]["p1"] < 0.05 for i in range(5)])
qNew = p2 < 0.05
ozg = int((qOld != qNew).sum())

och("w14-t5c")
P_("-- 5-bosqich. Xulosa --- shartni hisobga olib (F199) ---------------")
P_("Buzilgan qatorlarda: n_eff = n(1-rho)/(1+rho), SE ni sqrt((1+rho)/(1-rho))")
P_("ga koʻpaytiramiz.")
P_("")
P_("%s%s%s%s%s%s", pad("mamlakat", 14), padl("n_eff", 9),
   padl("tuzatilgan SE", 15), padl("tuzatilgan p", 14),
   padl("oldin", 10), padl("keyin", 10))
for i in range(5):
    P_("%s%s%s%s%s%s", pad(mam[i], 14),
       padl(("%.4f" % neff[i]) if buz[i] else "---", 9),
       padl("%.6f" % se2[i], 15), padl("%.6f" % p2[i], 14),
       padl("RAD" if qOld[i] else "rad emas", 10),
       padl("RAD" if qNew[i] else "rad emas", 10))
P_("")
P_("Xulosasi OʻZGARGAN qatorlar soni: %d", ozg)
P_("Oʻzgarganlari: %s",
   ", ".join([mam[i] for i in range(5) if qOld[i] != qNew[i]]))
P_("")
P_("SEZGIRLIK BAYONOTI (F207). Eng muhim qaror: shartni tekshirish va")
P_("tuzatish. Uni oʻzgartirdik --- natija: %d qatorda xulosa teskarisiga", ozg)
P_("aylandi. Demak 'hosildorlik oʻsdi' degan umumiy xulosa maʼlumotning")
P_("oʻzidan emas, TUZATISH QARORIDAN ham kelib chiqadi.")
P_("")
P_("TEKSHIRUV: buzilgan qatorlar soni = oʻzgargan xulosalar soni?  %s",
   ha(int(buz.sum()) == ozg))
yop()
```
